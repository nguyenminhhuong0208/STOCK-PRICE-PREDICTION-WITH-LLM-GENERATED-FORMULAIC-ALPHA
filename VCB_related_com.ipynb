```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException
import time

# Khởi tạo Edge driver
def init_driver():
    options = webdriver.EdgeOptions()
    options.add_argument("--headless")
    options.add_argument("--disable-gpu")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    driver = webdriver.Edge(options=options)
    return driver

def crawl_data(url, num_pages):
    driver = init_driver()
    driver.get(url)

    data = []
    print(f"Bắt đầu crawl {num_pages} trang từ: {url}")

    for i in range(num_pages):
        print(f"Trang {i+1}...")
        try:
            posts = driver.find_elements(By.XPATH, '//*[@id="divEvents"]/ul/li')

            for post in posts:
                try:
                    time_text = post.find_element(By.XPATH, './span').text.strip()
                    a_tag = post.find_element(By.XPATH, './a')
                    title_text = a_tag.text.strip()
                    link_url = a_tag.get_attribute("href")  # lấy URL

                    data.append({
                        'Thời gian': time_text,
                        'Tiêu đề': title_text,
                        'URL': link_url
                    })
                except NoSuchElementException:
                    continue

            # Nhấn nút "Sau" (AJAX load, không đổi URL)
            next_button = driver.find_element(By.XPATH, '//*[@id="aNext"]')
            if next_button.is_displayed() and next_button.is_enabled():
                driver.execute_script("arguments[0].click();", next_button)
                time.sleep(2)  # chờ dữ liệu mới load
            else:
                print("Không còn nút 'Sau'. Kết thúc.")
                break
        except NoSuchElementException:
            print("Không tìm thấy bài viết. Dừng lại.")
            break

    driver.quit()
    return pd.DataFrame(data)


URL = "https://cafef.vn/du-lieu/tin-doanh-nghiep/vjc/event.chn"
NUM_PAGES = 10
df = crawl_data(URL, NUM_PAGES)

print("\nKết quả crawl:")
print(df.head(10))
```

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.common.exceptions import NoSuchElementException, TimeoutException
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
from dateutil import parser
import datetime

# --- KHAI BÁO CÁC HẰNG SỐ CẬP NHẬT ---
FIREANT_URL = "https://fireant.vn/ma-chung-khoan/BID"
# XPATH MỚI: Tìm tất cả thẻ <a> có thuộc tính href bắt đầu bằng '/bai-viet' (URL bài viết)
# Sau đó lấy phần tử cha của nó (parent::div) để làm container lặp lại.
# Dựa trên ảnh của bạn, tiêu đề nằm trong <a> có href="/bai-viet..."
POSTS_CONTAINER_XPATH = '//a[starts-with(@href, "/bai-viet")]'
TIMEOUT = 15
STOP_DATE = datetime.datetime(2024, 1, 1, 0, 0, 0)
TODAY = datetime.datetime.now().replace(hour=0, minute=0, second=0, microsecond=0)

# Khởi tạo Edge driver (Các hàm cơ bản giữ nguyên)
def init_driver():
    options = webdriver.EdgeOptions()
    options.add_argument("--headless")
    options.add_argument("--disable-gpu")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    driver = webdriver.Edge(options=options)
    return driver

# Hàm parse_fireant_time giữ nguyên
def parse_fireant_time(time_str):
    """
    Chuyển đổi chuỗi thời gian FireAnt (có thể là "Hôm qua", "25/10 10:30",...)
    thành đối tượng datetime.
    """
    time_str = time_str.strip()
    try:
        if 'Hôm qua' in time_str:
            date_part = (TODAY - datetime.timedelta(days=1)).strftime('%Y/%m/%d')
            time_part = time_str.split(' ')[-1]
            return parser.parse(f"{date_part} {time_part}", dayfirst=True)
        elif 'Hôm nay' in time_str:
            date_part = TODAY.strftime('%Y/%m/%d')
            time_part = time_str.split(' ')[-1]
            return parser.parse(f"{date_part} {time_part}", dayfirst=True)
        else:
            try:
                dt = parser.parse(time_str, dayfirst=True)
                if dt.year < TODAY.year:
                    dt = dt.replace(year=TODAY.year)
                return dt
            except:
                return None
    except:
        return None

def extract_posts_data(driver):
    """Trích xuất dữ liệu từ các bài viết hiện có trên trang FireAnt."""
    extracted_data = []

    try:
        # Chờ cho ít nhất một thẻ <a> có link bài viết xuất hiện
        WebDriverWait(driver, TIMEOUT).until(
            EC.presence_of_element_located((By.XPATH, POSTS_CONTAINER_XPATH))
        )
    except TimeoutException:
        print("Cảnh báo: Hết thời gian chờ. Không tìm thấy thẻ <a> chứa bài viết.")
        return extracted_data

    # Tìm TẤT CẢ các thẻ <a> (mỗi thẻ là một bài báo)
    posts = driver.find_elements(By.XPATH, POSTS_CONTAINER_XPATH)

    for a_tag in posts: # Lặp qua từng thẻ <a> (tiêu đề)
        time_text = ""
        title_text = ""
        link_url = ""

        try:
            # 1. Trích xuất Tiêu đề và URL từ thẻ <a> hiện tại
            link_url = a_tag.get_attribute("href")
            title_text = a_tag.text.strip()

            # Phần này rất quan trọng: Ta cần tìm thẻ <span> chứa thời gian
            # Thẻ <span> nằm trong một DIV cha nào đó của thẻ <a>
            # Ta tìm từ <div> cha (grandparent) của thẻ <a>
            grandparent_div = a_tag.find_element(By.XPATH, './ancestor::div[4]') # Giả định khoảng 4 cấp DIV cha

            # 2. Trích xuất Thời gian từ thẻ <span> trong DIV cha đó
            try:
                # Tìm thẻ <span> với class đặc trưng trong phạm vi lớn hơn
                time_element = grandparent_div.find_element(By.XPATH, './/span[contains(@class, "text-gray-400")]')
                time_text = time_element.text.strip()
            except NoSuchElementException:
                time_text = ""

            if title_text and link_url:
                 # Chuyển đổi thời gian
                 dt_object = parse_fireant_time(time_text)

                 extracted_data.append({
                    'Thời gian (string)': time_text,
                    'Thời gian (datetime)': dt_object,
                    'Tiêu đề': title_text,
                    'URL': link_url
                })
        except NoSuchElementException:
            # Lỗi khi tìm thẻ cha hoặc thẻ span
            continue

    return extracted_data

# Hàm crawl_data_fireant giữ nguyên logic cuộn trang và dừng theo ngày

def crawl_data_fireant(url, stop_date):
    """Crawl dữ liệu FireAnt, dừng lại khi bài viết mới nhất đạt đến stop_date."""
    driver = init_driver()
    driver.get(url)

    # Chờ trang tải hoàn toàn
    time.sleep(5)

    all_data = []

    print(f"Bắt đầu crawl từ: {url}")
    print(f"Giới hạn thời gian dừng: {stop_date.strftime('%Y-%m-%d')}")

    scroll_count = 0
    # Thêm một biến để theo dõi bài viết cuối cùng đã crawl để tránh lặp vô hạn nếu không cuộn được
    last_known_url = ""

    while True:
        scroll_count += 1
        print(f"\nLượt thu thập #{scroll_count}...")

        old_data_count = len(all_data)

        # 1. Trích xuất tất cả dữ liệu có sẵn trên màn hình
        current_page_data = extract_posts_data(driver)

        # 2. Kiểm tra điều kiện dừng: Bài viết mới nhất (cuối cùng)
        oldest_post_dt = None
        if current_page_data:
            oldest_post_dt = current_page_data[-1]['Thời gian (datetime)']

        # 3. Loại bỏ trùng lặp và thêm vào danh sách chung
        df_temp = pd.DataFrame(current_page_data).drop_duplicates(subset=['URL'])
        # Chỉ giữ lại các bài viết chưa được crawl (quan trọng khi cuộn trang)
        new_posts = df_temp[~df_temp['URL'].isin([d['URL'] for d in all_data])]
        all_data.extend(new_posts.to_dict('records'))

        newly_added_count = len(all_data) - old_data_count
        print(f"   -> Đã thu thập {len(current_page_data)} bài, thêm mới: {newly_added_count} bài.")

        # --- KIỂM TRA ĐIỀU KIỆN DỪNG ---
        if oldest_post_dt is not None and oldest_post_dt < stop_date:
            print(f"*** ĐIỀU KIỆN DỪNG ĐƯỢC KÍCH HOẠT! ***")
            print(f"Bài viết cũ nhất trên màn hình: {oldest_post_dt.strftime('%Y-%m-%d')}. Đã đạt giới hạn {stop_date.strftime('%Y-%m-%d')}")
            break

        # Nếu không có bài mới nào được thêm vào trong lần cuộn này, dừng lại
        if newly_added_count == 0 and scroll_count > 1:
            print("Cảnh báo: Không có bài viết mới được thêm vào. Có thể đã hết dữ liệu.")
            break

        # Cập nhật URL cuối cùng đã biết
        if new_posts.empty is False:
             last_known_url = new_posts['URL'].iloc[-1]

        # 4. Cuộn trang xuống dưới cùng
        last_height = driver.execute_script("return document.body.scrollHeight")
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

        # Chờ nội dung mới tải
        time.sleep(3)

        # 5. Kiểm tra xem có nội dung mới được tải không (Tùy chọn)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == last_height and newly_added_count == 0 and scroll_count > 1:
            print("Đã cuộn đến cuối trang và không có nội dung mới được tải. Kết thúc.")
            break

    driver.quit()

    # Lọc bài viết và chuẩn hóa kết quả
    df_result = pd.DataFrame(all_data)
    if 'Thời gian (datetime)' in df_result.columns:
        # Chỉ giữ lại các bài viết có thời gian sau hoặc bằng stop_date
        df_result = df_result[df_result['Thời gian (datetime)'] >= stop_date].copy()
        df_result = df_result.sort_values('Thời gian (datetime)', ascending=False) # Sắp xếp lại từ mới nhất
        df_result = df_result.drop(columns=['Thời gian (datetime)'])

    df_result = df_result.rename(columns={'Thời gian (string)': 'Thời gian'}).reset_index(drop=True)

    return df_result

# Crawl thử
URL = FIREANT_URL
STOP_DATE_LIMIT = datetime.datetime(2024, 1, 1, 0, 0, 0)

df = crawl_data_fireant(URL, STOP_DATE_LIMIT)

print("\n" + "="*50)
print("KẾT QUẢ CRAWL FIREANT (Đã sửa XPATH lần 3)")
print("="*50)
print(f"Tổng số bài viết độc nhất thu thập được (sau khi lọc): {len(df)}")
print(df.head(10))
```

```python
import pandas as pd
import time
import os
from datetime import datetime
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException

# --- Khởi tạo Edge driver ---
def init_driver():
    options = webdriver.EdgeOptions()
    options.add_argument("--headless")
    options.add_argument("--disable-gpu")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    driver = webdriver.Edge(options=options)
    return driver


# --- Hàm crawl nội dung 1 bài báo ---
def crawl_article_content(url):
    driver = init_driver()
    try:
        driver.get(url)
        time.sleep(2)  # chờ load JS

        content_texts = []

        # Tìm phần detail-content
        try:
            detail = driver.find_element(
                By.XPATH, '//div[contains(@class, "detail-content")]'
            )
        except NoSuchElementException:
            detail = None

        if detail:
            # Lấy sapo nếu có
            try:
                sapo = detail.find_element(By.XPATH, './/h2').text.strip()
                if sapo:
                    content_texts.append(sapo)
            except NoSuchElementException:
                pass

            # Lấy tất cả các đoạn <p>
            paras = detail.find_elements(By.XPATH, './/p')
            for p in paras:
                txt = p.text.strip()
                if txt and not txt.startswith(">>>"):  # bỏ dòng quảng cáo
                    content_texts.append(txt)

        # Nếu không tìm thấy detail-content thì fallback sang mainContent
        if not content_texts:
            try:
                paras2 = driver.find_elements(By.XPATH, '//*[@id="mainContent"]//p')
                for p in paras2:
                    txt = p.text.strip()
                    if txt:
                        content_texts.append(txt)
            except NoSuchElementException:
                pass

        return "\n".join(content_texts) if content_texts else None

    except Exception as e:
        print(f" Lỗi khi crawl {url}: {e}")
        return None
    finally:
        driver.quit()


# --- File CSV đầu vào ---
file_path = "/content/drive/MyDrive/cafef_news/VCB_related_com/CTG.csv"

if not os.path.exists(file_path):
    print(f" Error: File not found at {file_path}")
else:
    df = pd.read_csv(file_path)

    # Nếu chưa có cột "Nội dung" thì thêm
    if "Nội dung" not in df.columns:
        df["Nội dung"] = None

    # Chuẩn hóa thời gian
    df["Thời gian"] = pd.to_datetime(
        df["Thời gian"], errors="coerce"
    )


    # Lọc khoảng thời gian
    start_date = datetime(2024, 1, 1)
    end_date = datetime(2025, 10, 15)
    mask = (df["Thời gian"] >= start_date) & (df["Thời gian"] <= end_date)
    df_target = df[mask]
    print(f" Có {len(df_target)} bài trong khoảng từ {start_date.date()} đến {end_date.date()}")

    # Crawl nội dung
    for i, row in df_target.iterrows():
        if pd.isna(row["Nội dung"]) or not row["Nội dung"]:
            url = row["URL"]
            print(f" Crawl {i+1}/{len(df_target)}: {url}")
            content = crawl_article_content(url)
            df.at[i, "Nội dung"] = content

            # Lưu tạm để tránh mất dữ liệu
            df.to_csv(file_path, index=False, encoding="utf-8-sig")

    print(f"\n Đã cập nhật nội dung vào file: {file_path}")
```

```python
import pandas as pd
import os

# Định nghĩa đường dẫn file
FILE_PATH = "/content/drive/MyDrive/cafef_news/VCB_related_com/CTG.csv"

def clean_time_column(file_path):
    """
    Tải file CSV, loại bỏ dấu ngoặc đơn () khỏi cột 'Thời gian',
    và lưu lại file đã được làm sạch.
    """
    print(f"Bắt đầu xử lý file: {file_path}")

    # 1. Kiểm tra sự tồn tại của file
    if not os.path.exists(file_path):
        print(f"LỖI: Không tìm thấy file tại đường dẫn: {file_path}. Vui lòng kiểm tra lại.")
        return

    # 2. Đọc file
    try:
        # Thử các encoding phổ biến
        df = pd.read_csv(file_path, encoding='utf-8')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='utf-8-sig')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    # Kiểm tra cột 'Thời gian'
    if 'Thời gian' not in df.columns:
        print("LỖI: File không có cột 'Thời gian'. Vui lòng kiểm tra tên cột.")
        return

    # 3. Loại bỏ dấu () khỏi cột 'Thời gian'
    # Sử dụng .str.replace() để thay thế dấu '(' và ')' bằng chuỗi rỗng
    df['Thời gian'] = df['Thời gian'].astype(str).str.replace(r'[()]', '', regex=True).str.strip()

    # Thử chuyển đổi sang datetime (Optional, nhưng nên làm để kiểm tra kết quả)
    try:
        df['Thời gian'] = pd.to_datetime(df['Thời gian'], dayfirst=True, errors='coerce')
        print("Cột 'Thời gian' đã được chuyển đổi thành công sang định dạng datetime.")
    except Exception as e:
        print(f"Cảnh báo: Không thể chuyển đổi sang datetime. Lỗi: {e}")


    # 4. Lưu lại file đã làm sạch
    # Lưu đè lên file cũ hoặc lưu thành file mới (tùy chọn)
    output_file_path = file_path # Lưu đè lên file cũ
    # output_file_path = file_path.replace(".csv", "_cleaned.csv") # Lưu file mới

    df.to_csv(output_file_path, index=False, encoding='utf-8-sig')

    print(f"✅ Đã loại bỏ dấu ngoặc đơn và lưu lại file thành công tại: {output_file_path}")
    print("\nKiểm tra 5 dòng đầu sau khi làm sạch:")
    print(df.head())

# Chạy hàm làm sạch
clean_time_column(FILE_PATH)
```

```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer, AutoConfig
import numpy as np
import torch

#### Load model
model_path = '5CD-AI/Vietnamese-Sentiment-visobert'
tokenizer = AutoTokenizer.from_pretrained(model_path)
config = AutoConfig.from_pretrained(model_path)
model = AutoModelForSequenceClassification.from_pretrained(model_path)

# Đảm bảo model ở chế độ CPU
device = torch.device("cpu")
model.to(device)

sentence = 'Cũng giống mấy khoá Youtube học cũng được'
print('Sentence:', sentence)

# Tokenize input
input_ids = torch.tensor([tokenizer.encode(sentence, add_special_tokens=True)]).to(device)

# Dự đoán
with torch.no_grad():
    out = model(input_ids)
    scores = out.logits.softmax(dim=-1).cpu().numpy()[0]

# In kết quả
ranking = np.argsort(scores)[::-1]
print("### Sentiment score ####")
for i in range(scores.shape[0]):
    label = config.id2label[ranking[i]]
    score = scores[ranking[i]]
    print(f"{i+1}) {label}: {np.round(float(score), 4)}")
```

```python
import pandas as pd
import numpy as np
import torch
import os
from transformers import AutoModelForSequenceClassification, AutoTokenizer, AutoConfig
from tqdm import tqdm
from datetime import datetime

# =====================================
# --- Cấu hình ---
# =====================================
model_path = '5CD-AI/Vietnamese-Sentiment-visobert'
input_folder = '/content/drive/MyDrive/cafef_news/VCB_related_com/'
text_column = 'Nội dung'
title_column = 'Tiêu đề' # Thêm cột Tiêu đề
time_column = 'Thời gian' # Thêm cột Thời gian
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
aggregation_method = 'weighted'

# --- HẰNG SỐ KIỂM TRA NGÀY (ĐÃ THÊM) ---
START_DATE = pd.to_datetime('2024-01-01')

# --- DANH SÁCH FILE MỤC TIÊU CẦN XỬ LÝ ---
TARGET_FILES = ['VJC.csv','TCB.csv','T_VCB.csv','SSI.csv']
print(f"File mục tiêu: {TARGET_FILES}")
print(f"Chỉ xử lý các hàng từ ngày: {START_DATE.date()} trở đi.")
print("-" * 50)


# --- Load model/tokenizer --- #
tokenizer = AutoTokenizer.from_pretrained(model_path)
config = AutoConfig.from_pretrained(model_path)
model = AutoModelForSequenceClassification.from_pretrained(model_path).to(device)
model.eval()

# --- Xác định kích thước chunk an toàn (Giữ nguyên) --- #
model_max_len = getattr(tokenizer, "model_max_length", None)
if not model_max_len or model_max_len <= 0 or model_max_len > 1e6:
    model_max_len = 512
chunk_size = max(1, int(model_max_len) - 2)

# --- Build mapping label -> index (Giữ nguyên) --- #
label_map = {}
for k, v in config.id2label.items():
    idx = int(k)
    lab = str(v).lower()
    if 'pos' in lab:
        label_map['positive'] = idx
    elif 'neg' in lab:
        label_map['negative'] = idx
    elif 'neu' in lab or 'nat' in lab:
        label_map['neutral'] = idx

if not all(k in label_map for k in ('positive','neutral','negative')):
    ids = sorted([int(k) for k in config.id2label.keys()])
    if len(ids) == 3:
        label_map = {'negative': ids[0], 'neutral': ids[1], 'positive': ids[2]}
    else:
        label_map = {f'label_{i}': idv for i, idv in enumerate(ids)}
print("Label map (resolved):", label_map)

# =====================================
# --- Hàm predict cho một văn bản (Giữ nguyên) ---
# =====================================
def predict_probs_for_text(text, aggregation='weighted'):
    """Trả về vector probabilities (shape = num_labels)."""
    if not isinstance(text, str) or text.strip() == "":
        return None

    token_ids = tokenizer.encode(text, add_special_tokens=False)
    if len(token_ids) == 0:
        return None

    chunks = [token_ids[i:i+chunk_size] for i in range(0, len(token_ids), chunk_size)]
    chunk_preds = []
    chunk_weights = []

    for ch in chunks:
        ch_text = tokenizer.decode(ch, clean_up_tokenization_spaces=True)
        enc = tokenizer(ch_text,
                        truncation=True,
                        max_length=model_max_len,
                        return_tensors='pt',
                        padding=False)
        enc = {k: v.to(device) for k, v in enc.items()}

        with torch.no_grad():
            out = model(**enc)
            probs = torch.softmax(out.logits, dim=-1).cpu().numpy()[0]
        chunk_preds.append(probs)
        chunk_weights.append(len(ch))

    chunk_preds = np.stack(chunk_preds, axis=0)
    chunk_weights = np.array(chunk_weights, dtype=float)

    if aggregation == 'majority':
        votes = np.argmax(chunk_preds, axis=1)
        n_labels = chunk_preds.shape[1]
        counts = np.zeros(n_labels, dtype=float)
        for v in votes:
            counts[int(v)] += 1.0
        return counts / counts.sum()
    else:
        weighted = np.average(chunk_preds, axis=0, weights=chunk_weights)
        return weighted

# =====================================
# --- Xử lý từng file CSV (ĐÃ SỬA LOGIC) ---
# =====================================
for csv_file in TARGET_FILES:
    input_path = os.path.join(input_folder, csv_file)

    if not os.path.exists(input_path):
        print(f"\nCẢNH BÁO: KHÔNG TÌM THẤY file: {csv_file}. Bỏ qua...")
        continue

    print(f"\n--- Đang xử lý file: {csv_file} ---")

    # Đọc CSV
    try:
        df = pd.read_csv(input_path, encoding='utf-8-sig')
    except Exception as e:
        print(f"LỖI: Không thể đọc file {csv_file}: {e}. Bỏ qua.")
        continue

    # --- CHUẨN BỊ VÀ LỌC DỮ LIỆU ---
    # Đổi tên cột Thời gian để dễ dàng chuyển đổi ngày tháng nếu cần (optional, nhưng an toàn hơn)
    df.rename(columns={'Thời gian': time_column}, inplace=True)

    # Đảm bảo các cột sentiment tồn tại và chuyển đổi kiểu dữ liệu ngày tháng
    sentiment_cols = ['positive', 'neutral', 'negative']
    for c in sentiment_cols:
        if c not in df.columns:
            df[c] = np.nan
        # Đảm bảo cột numeric để kiểm tra NaN chính xác
        df[c] = pd.to_numeric(df[c], errors='coerce')

    # Thử các định dạng ngày tháng phổ biến
    try:
        df[time_column] = pd.to_datetime(df[time_column], dayfirst=True)
    except:
        try:
            df[time_column] = pd.to_datetime(df[time_column], format='%Y-%m-%d')
        except:
            df[time_column] = pd.to_datetime(df[time_column], errors='coerce')

    # Lọc các hàng từ ngày 2024-01-01 trở đi
    df_filtered = df[df[time_column] >= START_DATE].copy()

    if df_filtered.empty:
        print(f"Không có hàng nào từ ngày {START_DATE.date()} để xử lý.")
        continue

    # --- LỌC CÁC HÀNG CẦN TÍNH TOÁN (ĐÃ SỬA LOGIC) ---
    # 1. Kiểm tra 3 cột sentiment có phải là NaN không? (Chưa được tính)
    needs_sentiment = df_filtered[sentiment_cols].isnull().all(axis=1)
    # 2. Kiểm tra xem có văn bản nào để tính không? (Nội dung HOẶC Tiêu đề)
    has_text = (df_filtered[text_column].notna() & (df_filtered[text_column].astype(str).str.strip() != '')) | \
               (df_filtered[title_column].notna() & (df_filtered[title_column].astype(str).str.strip() != ''))

    rows_to_process_df = df_filtered[needs_sentiment & has_text]
    rows_to_process_indices = rows_to_process_df.index

    print(f"Tổng số hàng trong file (>=2024-01-01): {len(df_filtered)}")
    print(f"Số hàng cần tính Sentiment (Chưa tính & Có Text): {len(rows_to_process_indices)}")

    if len(rows_to_process_indices) == 0:
        print("Không có hàng mới cần tính sentiment trong khoảng thời gian này.")
        # Lưu file để đảm bảo các cột đã được chuyển sang numeric nếu có NaN
        df.to_csv(input_path, index=False, encoding='utf-8-sig')
        continue

    # Duyệt chỉ các dòng cần xử lý
    for idx in tqdm(rows_to_process_indices, desc=f"Calculating sentiment for {csv_file}"):

        # --- LOGIC XỬ LÝ NỘI DUNG THAY THẾ ---
        text = df.at[idx, text_column]

        if pd.isna(text) or str(text).strip() == "":
            # Nội dung null, dùng Tiêu đề
            text_to_analyze = df.at[idx, title_column]
        else:
            # Nội dung có sẵn, dùng Nội dung
            text_to_analyze = text

        if pd.isna(text_to_analyze) or str(text_to_analyze).strip() == "":
            continue # Không có gì để tính

        try:
            probs = predict_probs_for_text(str(text_to_analyze), aggregation=aggregation_method)

            if probs is None:
                continue

            # Cập nhật vào 3 cột sentiment
            if 'positive' in label_map:
                df.at[idx, 'positive'] = float(probs[int(label_map['positive'])])
            if 'neutral' in label_map:
                df.at[idx, 'neutral'] = float(probs[int(label_map['neutral'])])
            if 'negative' in label_map:
                df.at[idx, 'negative'] = float(probs[int(label_map['negative'])])

        except Exception as e:
            print(f"Lỗi khi xử lý dòng {idx} ({df.at[idx, time_column].date()}) trong {csv_file}: {e}")
            continue

    # Lưu kết quả (ghi đè file gốc với dữ liệu đã được cập nhật)
    df.to_csv(input_path, index=False, encoding='utf-8-sig')
    print(f"Đã cập nhật sentiment và lưu vào file: {csv_file}")

print("\nĐã xử lý xong tất cả file mục tiêu!")
```

```python
import pandas as pd
import os
from datetime import datetime

# 1. Định nghĩa đường dẫn thư mục
folder_path = "/content/drive/MyDrive/cafef_news/VCB_related_com"

# 2. Tên cột cần kiểm tra
column_name = 'Thời gian'

# 3. Định dạng thời gian MONG MUỐN
# Chuỗi định dạng cho "2025-10-14 0:00:00" là: Năm-Tháng-Ngày Giờ:Phút:Giây
target_format = '%Y-%m-%d %H:%M:%S'

print(f" Bắt đầu kiểm tra định dạng cột '{column_name}' trong thư mục: {folder_path}\n")
print("-" * 50)

# 4. Duyệt qua tất cả các file trong thư mục
for filename in os.listdir(folder_path):
    # Chỉ xử lý các file có đuôi .csv
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)

        print(f" Đang xử lý file: {filename}")

        try:
            # Đọc file CSV (chỉ cần đọc một vài hàng để kiểm tra định dạng)
            df = pd.read_csv(file_path, nrows=5)

            # Kiểm tra xem cột 'Thời gian' có tồn tại không
            if column_name not in df.columns:
                print(f"   KHÔNG TÌM THẤY CỘT '{column_name}'")
                continue # Bỏ qua file này và chuyển sang file tiếp theo

            # Lấy giá trị đầu tiên của cột 'Thời gian' để kiểm tra
            first_time_value = df[column_name].iloc[0]

            # Cố gắng phân tích cú pháp chuỗi theo định dạng mong muốn
            try:
                # Nếu phân tích thành công, tức là định dạng đã đúng
                datetime.strptime(str(first_time_value), target_format)
                print(f"   ĐÚNG ĐỊNH DẠNG: '{first_time_value}' đã chuẩn.")

            except ValueError:
                # Nếu phân tích thất bại, tức là định dạng chưa đúng
                print(f"   SAI ĐỊNH DẠNG: Giá trị mẫu: '{first_time_value}'")
                print("   (Cần chuyển đổi sang dạng 'Năm-Tháng-Ngày Giờ:Phút:Giây')")


        except pd.errors.EmptyDataError:
            print("   LỖI: File rỗng.")
        except Exception as e:
            print(f"   LỖI XỬ LÝ: {e}")

        print("-" * 50)

print("\n HOÀN THÀNH việc kiểm tra tất cả các file CSV.")
```

```python
import pandas as pd
import os
import numpy as np

# 1. Định nghĩa đường dẫn và thông tin chung
folder_path = "/content/drive/MyDrive/cafef_news/VCB_related_com"
files_to_fix = ['FPT_sentiment.csv','HPG.csv']
column_name = 'Thời gian'
# Định dạng GỐC được sử dụng để chuyển đổi nếu các file chưa chuẩn: '%m/%d/%Y %H:%M'
# Định dạng ĐÍCH để ghi lại file: 'YYYY-MM-DD H:M:S'
target_format = '%Y-%m-%d %H:%M:%S'


print(" Bắt đầu sửa định dạng thời gian cho các file chưa chuẩn...")
print("-" * 60)

# 2. Duyệt qua danh sách các file cần sửa
for filename in files_to_fix:
    file_path = os.path.join(folder_path, filename)

    print(f" Đang xử lý file: {filename}")

    try:
        # Đọc toàn bộ file CSV
        # Thử đọc với encoding phổ biến
        try:
            df = pd.read_csv(file_path)
        except UnicodeDecodeError:
            df = pd.read_csv(file_path, encoding='latin-1')
        except Exception as e:
            raise ValueError(f"Lỗi đọc file: {e}")


        if column_name not in df.columns:
            print(f"     LỖI: Không tìm thấy cột '{column_name}' trong file này. Bỏ qua.")
            continue

        # Lấy giá trị mẫu trước khi sửa
        old_value = df[column_name].iloc[0]

        # SỬA LỖI Ở ĐÂY: Dùng to_datetime linh hoạt hơn
        # Sử dụng 'mixed' để Pandas tự suy luận định dạng, giúp xử lý:
        # 1. 'M/D/YYYY H:M'
        # 2. 'YYYY-MM-DD' (như FPT_sentiment)
        # 3. 'DD/MM/YYYY' (như HPG)
        # Errors='coerce' sẽ biến các giá trị không hợp lệ thành NaT (Not a Time)
        df[column_name] = pd.to_datetime(df[column_name], errors='coerce', dayfirst=True, format='mixed')

        # Xử lý các giá trị không chuyển đổi được (rất hiếm, nhưng để an toàn)
        if df[column_name].isnull().any():
             invalid_count = df[column_name].isnull().sum()
             print(f"     CẢNH BÁO: Phát hiện {invalid_count} giá trị ngày tháng không hợp lệ (NaT) sau khi chuyển đổi. Sẽ bỏ qua.")

        # B. Chuyển đổi ngược lại sang chuỗi (string) theo định dạng đích
        # fillna('') để tránh lỗi khi chuyển đổi NaT sang strftime
        df[column_name] = df[column_name].dt.strftime(target_format).fillna('')

        # C. Ghi đè file
        df.to_csv(file_path, index=False)

        # Lấy giá trị mẫu sau khi sửa
        new_value = df[column_name].iloc[0]

        print(f"     HOÀN THÀNH. Định dạng đã chuyển đổi:")
        print(f"       Từ: '{old_value}'")
        print(f"       Sang: '{new_value}'")

    except FileNotFoundError:
        print(f"     LỖI: Không tìm thấy file {filename} tại đường dẫn.")
    except Exception as e:
        print(f"     LỖI XỬ LÝ file {filename}: {e}")

    print("-" * 60)

print("\n XONG! Tất cả các file đã được đồng bộ hóa định dạng thời gian.")

# ==============================================================
# Lý giải cho 2 file lỗi:
# FPT_sentiment.csv: Định dạng là YYYY-MM-DD.
# HPG.csv: Định dạng là DD/MM/YYYY.
# Hàm pd.to_datetime(df[column_name], errors='coerce', dayfirst=True, format='mixed')
# sẽ giải quyết cả hai vấn đề này.
# ==============================================================
```

```python
import pandas as pd
import numpy as np
import os
from datetime import datetime
import traceback # Import để in lỗi chi tiết hơn

# --- Cấu hình --- #
input_folder = '/content/drive/MyDrive/cafef_news/VCB_related_com/'
time_column = 'Thời gian'
EXCLUDED_FILE = 'FPT_sentiment.csv' # Tên file cần loại trừ

# --- Hàm chuẩn hóa ngày tháng --- #
def normalize_date(date_str):
    """
    Chuẩn hóa các định dạng ngày tháng về format YYYY-MM-DD
    """
    if pd.isna(date_str):
        return None

    date_str = str(date_str).strip()

    # Cắt bỏ phần giờ phút giây, chỉ lấy phần ngày tháng năm
    # Sử dụng regex hoặc split thông minh hơn để xử lý các trường hợp giờ/phút/giây không có
    date_part = date_str.split(' ')[0]

    # Thử các định dạng ngày tháng phổ biến
    date_formats = [
        '%Y-%m-%d',
        '%d/%m/%Y',
        '%d-%m-%Y',
        '%m/%d/%Y',
        '%Y/%m/%d',
        '%d.%m.%Y',
        '%Y.%m.%d'
    ]

    for fmt in date_formats:
        try:
            date_obj = datetime.strptime(date_part, fmt)
            return date_obj.strftime('%Y-%m-%d')
        except ValueError:
            continue

    # Nếu không parse được, trả về None
    # print(f"Không thể parse ngày: {date_str}") # Tắt in lỗi để tránh làm rối console
    return None

# --- Lấy danh sách tất cả file CSV trong thư mục (Loại trừ các file _sentiment.csv) --- #
csv_files = [
    f for f in os.listdir(input_folder)
    if f.endswith('.csv') and
       not f.endswith('_sentiment.csv') and # Loại trừ tất cả các file đã tổng hợp
       f != EXCLUDED_FILE # Loại trừ file FPT_sentiment.csv (dù nó đã được loại ở điều kiện trên)
]
print(f"Tìm thấy {len(csv_files)} file CSV cần xử lý: {csv_files}")

# --- Xử lý từng file CSV --- #
for csv_file in csv_files:
    # Đảm bảo file output có tên là *_sentiment.csv
    output_file = csv_file.replace('.csv', '_sentiment.csv')

    # Kiểm tra nếu file input là FPT.csv thì output phải là FPT_sentiment.csv
    if csv_file == 'FPT.csv' and output_file == 'FPT_sentiment.csv':
        # Yêu cầu của bạn là xử lý tất cả các file TRỪ file FPT_sentiment.csv.
        # Nếu FPT.csv nằm trong thư mục này, chúng ta cần xử lý nó.
        # Tuy nhiên, nếu bạn muốn loại trừ file gốc FPT.csv (và chỉ giữ lại các file related), bạn cần thay đổi logic.
        # Hiện tại, tôi giữ lại FPT.csv để xử lý nếu nó nằm trong folder này.

        # Tên output: Nếu file là FPT.csv, output sẽ là FPT_sentiment.csv.
        # Nếu output file trùng với EXCLUDED_FILE (FPT_sentiment.csv), chúng ta vẫn tạo nó,
        # nhưng quá trình sau này sẽ sử dụng nó.
        pass

    input_path = os.path.join(input_folder, csv_file)
    output_path = os.path.join(input_folder, output_file)

    print(f"\n{'='*50}")
    print(f"Đang xử lý file: {csv_file}")
    print(f"File output dự kiến: {output_file}")
    print(f"{'='*50}")

    try:
        # Đọc CSV
        # Cố gắng đọc với nhiều encoding
        try:
            df = pd.read_csv(input_path, encoding='utf-8-sig')
        except UnicodeDecodeError:
            try:
                df = pd.read_csv(input_path, encoding='cp1252')
            except:
                df = pd.read_csv(input_path, encoding='latin-1')

        print(f"Tổng số dòng: {len(df)}")

        # Kiểm tra các cột cần thiết
        required_columns = [time_column, 'positive', 'neutral', 'negative']
        missing_columns = [col for col in required_columns if col not in df.columns]

        if missing_columns:
            print(f"BỎ QUA: File {csv_file} thiếu các cột sentiment hoặc thời gian: {missing_columns}")
            continue

        # Chuẩn hóa cột ngày tháng - chỉ lấy phần ngày (bỏ giờ phút)
        df['normalized_date'] = df[time_column].apply(normalize_date)

        # Lọc các dòng có ngày hợp lệ VÀ có sentiment scores
        # Đảm bảo 3 cột sentiment là số và không phải NaN
        sentiment_is_valid = df[['positive', 'neutral', 'negative']].apply(pd.to_numeric, errors='coerce').notna().all(axis=1)
        valid_dates = df['normalized_date'].notna()
        valid_rows = sentiment_is_valid & valid_dates

        print(f"Số dòng có sentiment hợp lệ: {sentiment_is_valid.sum()}")
        print(f"Số dòng có ngày hợp lệ: {valid_dates.sum()}")
        print(f"Số dòng có cả sentiment và ngày hợp lệ: {valid_rows.sum()}")

        if valid_rows.sum() == 0:
            print(f"BỎ QUA: Không có dòng nào có đủ dữ liệu sentiment và ngày hợp lệ trong file {csv_file}")
            continue

        # Tính sentiment trung bình theo ngày
        sentiment_cols = ['positive', 'neutral', 'negative']

        # Chuyển đổi sang numeric trước khi groupby để tránh lỗi
        df_valid = df[valid_rows].copy()
        df_valid[sentiment_cols] = df_valid[sentiment_cols].apply(pd.to_numeric, errors='coerce')

        # Nhóm theo ngày và tính trung bình
        daily_sentiment = df_valid.groupby('normalized_date')[sentiment_cols].mean().reset_index()
        daily_sentiment = daily_sentiment.rename(columns={'normalized_date': 'Thời gian'})

        # Sắp xếp theo thời gian
        daily_sentiment['Thời gian'] = pd.to_datetime(daily_sentiment['Thời gian'])
        daily_sentiment = daily_sentiment.sort_values('Thời gian')
        daily_sentiment['Thời gian'] = daily_sentiment['Thời gian'].dt.strftime('%Y-%m-%d')

        # Làm tròn đến 4 chữ số thập phân
        for col in sentiment_cols:
            daily_sentiment[col] = daily_sentiment[col].round(4)

        # Thêm thông tin số bài báo mỗi ngày
        article_count = df_valid.groupby('normalized_date').size().reset_index(name='số_bài_báo')

        # Merge dựa trên normalized_date trước khi đổi tên cột
        daily_sentiment = daily_sentiment.merge(article_count,
                                                left_on='Thời gian',
                                                right_on=article_count['normalized_date'].apply(lambda x: pd.to_datetime(x).strftime('%Y-%m-%d')),
                                                how='left')
        daily_sentiment = daily_sentiment.drop('normalized_date', axis=1, errors='ignore')


        print(f"\nKết quả tổng hợp theo ngày cho {csv_file}:")
        print(f"Số ngày duy nhất: {len(daily_sentiment)}")
        print(f"Phạm vi thời gian: {daily_sentiment['Thời gian'].min()} đến {daily_sentiment['Thời gian'].max()}")

        # Lưu file kết quả
        daily_sentiment.to_csv(output_path, index=False, encoding='utf-8-sig')
        print(f"ĐÃ LƯU KẾT QUẢ VÀO: {output_file}")

        # Hiển thị preview
        print("\nPreview 5 dòng đầu:")
        print(daily_sentiment.head())

    except Exception as e:
        print(f"LỖI NGHIÊM TRỌNG khi xử lý file {csv_file}: {e}")
        traceback.print_exc()
        continue

print(f"\n{'='*50}")
print("Đã xử lý xong tất cả file được chọn!")
print(f"{'='*50}")
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# --- Cấu hình --- #
STOCK_FILE = "/content/VCB.csv"
# Cột thời gian có thể là 'time' hoặc 'Thời gian'
time_column = 'time'

# --- Hàm tính Technical Indicators (Đã thêm SMA_10, SMA_50, EMA_20) --- #
def calculate_technical_indicators(df):
    """
    Tính toán các technical indicators từ stock data
    """
    # ... (Nội dung hàm calculate_technical_indicators giữ nguyên) ...
    # (Tôi đã lược bỏ nội dung chi tiết để tập trung vào phần cấu hình)

    # Đảm bảo datetime index và sắp xếp
    df[time_column] = pd.to_datetime(df[time_column])
    df = df.sort_values(time_column).reset_index(drop=True)

    print("Calculating technical indicators...")

    # 1. Simple Moving Average (SMA)
    print("Calculating SMA...")
    df['SMA_5'] = df['close'].rolling(window=5, min_periods=1).mean()
    df['SMA_10'] = df['close'].rolling(window=10, min_periods=1).mean() # Thêm SMA_10
    df['SMA_20'] = df['close'].rolling(window=20, min_periods=1).mean()
    df['SMA_50'] = df['close'].rolling(window=50, min_periods=1).mean() # Thêm SMA_50


    # 2. Exponential Moving Average (EMA)
    print("Calculating EMA...")
    df['EMA_10'] = df['close'].ewm(span=10, adjust=False).mean()
    df['EMA_20'] = df['close'].ewm(span=20, adjust=False).mean() # Thêm EMA_20


    # 3. Momentum Indicators
    print("Calculating Momentum...")
    df['Momentum_3'] = df['close'].pct_change(periods=3)
    df['Momentum_5'] = df['close'].pct_change(periods=5) # Thêm Momentum_5
    df['Momentum_10'] = df['close'].pct_change(periods=10)


    # 4. Relative Strength Index (RSI)
    print("Calculating RSI...")
    def calculate_rsi(series, period=14):
        delta = series.diff()
        gain = (delta.where(delta > 0, 0)).rolling(window=period, min_periods=1).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(window=period, min_periods=1).mean()
        loss = loss.replace(0, 0.0001)  # Tránh chia cho 0
        rs = gain / loss
        rsi = 100 - (100 / (1 + rs))
        return rsi

    df['RSI_14'] = calculate_rsi(df['close'], 14)

    # 5. MACD (Moving Average Convergence Divergence)
    print("Calculating MACD...")
    exp1 = df['close'].ewm(span=12, adjust=False).mean()
    exp2 = df['close'].ewm(span=26, adjust=False).mean()
    df['MACD'] = exp1 - exp2
    df['MACD_Signal'] = df['MACD'].ewm(span=9, adjust=False).mean()

    # 6. Bollinger Bands
    print("Calculating Bollinger Bands...")
    df['BB_Middle'] = df['close'].rolling(window=20, min_periods=1).mean()
    bb_std = df['close'].rolling(window=20, min_periods=1).std()
    df['BB_Upper'] = df['BB_Middle'] + (bb_std * 2)
    df['BB_Lower'] = df['BB_Middle'] - (bb_std * 2)

    # 7. On-Balance Volume (OBV)
    print("Calculating OBV...")
    df['OBV'] = (np.sign(df['close'].diff()) * df['volume']).fillna(0).cumsum()

    # Fill NaN values
    print("Filling NaN values...")
    indicator_columns = [
        'SMA_5', 'SMA_10', 'SMA_20', 'SMA_50', 'EMA_10', 'EMA_20', # Cập nhật danh sách TI
        'Momentum_3', 'Momentum_5', 'Momentum_10', # Cập nhật danh sách TI
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Middle', 'BB_Upper', 'BB_Lower', 'OBV'
    ]

    for col in indicator_columns:
        if col in df.columns:
            df[col] = df[col].fillna(method='ffill').fillna(method='bfill')

    df['Momentum_3'] = df['Momentum_3'].fillna(0)
    df['Momentum_5'] = df['Momentum_5'].fillna(0) # Fill Momentum_5
    df['Momentum_10'] = df['Momentum_10'].fillna(0)

    # Loại bỏ cột date tạm thời
    # Code gốc có thêm cột 'date' bên trong hàm, nhưng tôi sửa lại để dùng 'time_column'
    # Nếu file gốc không có cột 'date' thì không cần drop

    return df


# --- Hàm kiểm tra chất lượng indicators (Giữ nguyên) --- #
def validate_indicators(df):
    # ... (Nội dung hàm validate_indicators giữ nguyên) ...
    print("\n=== VALIDATION REPORT ===")
    indicator_columns = [
        'SMA_5', 'SMA_10', 'SMA_20', 'SMA_50', 'EMA_10', 'EMA_20', # Cập nhật danh sách TI
        'Momentum_3', 'Momentum_5', 'Momentum_10', # Cập nhật danh sách TI
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV'
    ]
    for col in indicator_columns:
        if col in df.columns:
            nan_count = df[col].isna().sum()
            inf_count = np.isinf(df[col]).sum()
            mean_val = df[col].mean()
            std_val = df[col].std()
            print(f"{col:15} | NaN: {nan_count:3d} | Inf: {inf_count:2d} | Mean: {mean_val:8.4f} | Std: {std_val:8.4f}")
        else:
            print(f"{col:15} | NOT FOUND")

# --- Hàm mô tả các indicators (Giữ nguyên) --- #
def describe_indicators():
    # ... (Nội dung hàm describe_indicators giữ nguyên) ...
    print("\n=== TECHNICAL INDICATORS DESCRIPTION ===")
    indicators_info = {
        'SMA_5': 'Simple Moving Average (5 periods) - Trung bình giá đóng cửa 5 ngày',
        'SMA_10': 'Simple Moving Average (10 periods) - Trung bình giá đóng cửa 10 ngày', # Thêm mô tả
        'SMA_20': 'Simple Moving Average (20 periods) - Trung bình giá đóng cửa 20 ngày',
        'SMA_50': 'Simple Moving Average (50 periods) - Trung bình giá đóng cửa 50 ngày', # Thêm mô tả
        'EMA_10': 'Exponential Moving Average (10 periods) - Trung bình động hàm mũ 10 ngày',
        'EMA_20': 'Exponential Moving Average (20 periods) - Trung bình động hàm mũ 20 ngày', # Thêm mô tả
        'Momentum_3': 'Momentum Indicator (3 periods) - Động lượng giá 3 ngày',
        'Momentum_5': 'Momentum Indicator (5 periods) - Động lượng giá 5 ngày', # Thêm mô tả
        'Momentum_10': 'Momentum Indicator (10 periods) - Động lượng giá 10 ngày',
        'RSI_14': 'Relative Strength Index (14 periods) - Chỉ số sức mạnh tương đối 14 ngày',
        'MACD': 'MACD Line (fast=12, slow=26) - Đường MACD',
        'MACD_Signal': 'MACD Signal Line (signal=9) - Đường tín hiệu MACD',
        'BB_Upper': 'Bollinger Bands Upper (20-period, 2 std) - Dải Bollinger trên',
        'BB_Lower': 'Bollinger Bands Lower (20-period, 2 std) - Dải Bollinger dưới',
        'OBV': 'On-Balance Volume - Khối lượng cân bằng'
    }
    for indicator, description in indicators_info.items():
        print(f"{indicator:15} : {description}")

# --- Hàm kiểm tra dữ liệu đầu vào (Giữ nguyên) --- #
def check_data_quality(df):
    # ... (Nội dung hàm check_data_quality giữ nguyên) ...
    print("\n=== DATA QUALITY CHECK ===")
    print(f"Total rows: {len(df)}")
    print(f"Total columns: {len(df.columns)}")
    print(f"Column names: {list(df.columns)}")
    print("\nData types:")
    for col in df.columns:
        print(f"  {col}: {df[col].dtype}")
    print("\nMissing values:")
    for col in df.columns:
        missing_count = df[col].isna().sum()
        print(f"  {col}: {missing_count} missing")
    numeric_cols = ['open', 'high', 'low', 'close', 'volume']
    for col in numeric_cols:
        if col in df.columns:
            negative_count = (df[col] < 0).sum()
            zero_count = (df[col] == 0).sum()
            print(f"  {col}: {negative_count} negative, {zero_count} zero values")

# --- Main Execution (Đã sửa) --- #
def main():
    # SỬA: Tiêu đề
    print("=== VCB TECHNICAL INDICATORS CALCULATION ===")

    try:
        # Bước 1: Đọc dữ liệu
        print("1. Reading stock data...")
        # SỬA: Dùng biến STOCK_FILE đã được đổi tên
        df = pd.read_csv(STOCK_FILE)
        print(f"Original data shape: {df.shape}")

        # Kiểm tra chất lượng dữ liệu đầu vào
        check_data_quality(df)

        # Bước 2: Chuyển đổi kiểu dữ liệu
        print("\n2. Converting data types...")
        # Chuyển đổi các cột số
        numeric_cols = ['open', 'high', 'low', 'close', 'volume']
        for col in numeric_cols:
            if col in df.columns:
                df[col] = pd.to_numeric(df[col], errors='coerce')
                print(f"  Converted {col} to numeric")

        # Kiểm tra time format
        # SỬA: Dùng biến time_column đã khai báo
        print(f"  Time column sample: {df[time_column].head(3).tolist()}")

        # Bước 3: Tính technical indicators
        print("\n3. Calculating technical indicators...")
        df_with_indicators = calculate_technical_indicators(df)

        # Bước 4: Kiểm tra chất lượng indicators
        print("\n4. Validating indicators...")
        validate_indicators(df_with_indicators)

        # Bước 5: Hiển thị mô tả
        describe_indicators()

        # Bước 6: Lưu file với indicators
        print("\n5. Saving updated file...")

        # Tạo file mới với đuôi _with_indicators
        # SỬA: Dùng biến STOCK_FILE đã được đổi tên
        output_file = STOCK_FILE.replace('.csv', '_with_indicators.csv')
        df_with_indicators.to_csv(output_file, index=False, encoding='utf-8-sig')
        print(f"Saved to: {output_file}")

        # Cập nhật file gốc
        # SỬA: Dùng biến STOCK_FILE đã được đổi tên
        df_with_indicators.to_csv(STOCK_FILE, index=False, encoding='utf-8-sig')
        print(f"Updated original file: {STOCK_FILE}")

        # Bước 7: Hiển thị thống kê
        print("\n=== FINAL STATISTICS ===")
        print(f"Total rows: {len(df_with_indicators)}")
        print(f"Total columns: {len(df_with_indicators.columns)}")

        # ... (Phần thống kê và hiển thị mẫu giữ nguyên) ...
        # (Lược bỏ phần thống kê để giữ code ngắn gọn)

        return df_with_indicators

    except Exception as e:
        print(f"Error: {e}")
        import traceback
        traceback.print_exc()
        return None

# --- Chạy trực tiếp --- #
if __name__ == "__main__":
    result_df = main()

    if result_df is not None:
        print("\n Technical indicators calculation completed successfully!")
        # Đếm số cột technical indicators (trừ các cột gốc)
        tech_indicator_cols = [col for col in result_df.columns if col not in ['time', 'open', 'high', 'low', 'close', 'volume', 'Date']]
        print(f"\nFile đã được cập nhật với {len(tech_indicator_cols)} technical indicators")
    else:
        print("\n Technical indicators calculation failed!")
```

```python
import pandas as pd
import numpy as np
import os
from datetime import datetime
import warnings
warnings.filterwarnings('ignore') # Tắt cảnh báo khi dùng fillna(method='ffill')

# ====================================================================
#                          I. CẤU HÌNH CHUNG
# ====================================================================
TARGET_STOCK_CODE = "VCB"

# Tên cột thời gian trong các file khác nhau
TIME_COLUMN_STOCK = 'time'       # Tên cột trong /content/SSI.csv
TIME_COLUMN_SENTIMENT = 'Thời gian' # Tên cột trong *_sentiment.csv

# Đường dẫn file và thư mục
STOCK_FILE_PATH = f"/content/{TARGET_STOCK_CODE}.csv"
SENTIMENT_FILE_PATH = f"/content/drive/MyDrive/cafef_news/VCB_related_com/T_{TARGET_STOCK_CODE}_sentiment.csv"
RELATED_COM_FOLDER = "/content/drive/MyDrive/cafef_news/VCB_related_com"
OUTPUT_FILE = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"

START_DATE = "2024-01-01"
END_DATE = "2025-10-15"

# 10 công ty liên quan theo yêu cầu
RELATED_COMPANIES = ['TCB', 'VJC', 'HPG', 'PLX', 'VIC', 'SSI', 'GAS','FPT']

# Tạo thư mục output nếu chưa tồn tại
os.makedirs(os.path.dirname(OUTPUT_FILE), exist_ok=True)
print(f"Thư mục output đã sẵn sàng: {os.path.dirname(OUTPUT_FILE)}")

# ====================================================================
#                     II. HÀM XỬ LÝ DỮ LIỆU CHÍNH
# ====================================================================

# --- 1. Đọc và xử lý Stock Data ---
def load_stock_data():
    """Đọc và xử lý stock data, đổi tên cột 'time' thành 'Thời gian'"""
    print(f"1. Loading stock data ({TARGET_STOCK_CODE}) with technical indicators...")

    if not os.path.exists(STOCK_FILE_PATH):
        print(f"    Lỗi: Không tìm thấy file Stock Data tại: {STOCK_FILE_PATH}")
        return pd.DataFrame()

    df = pd.read_csv(STOCK_FILE_PATH)

    if TIME_COLUMN_STOCK not in df.columns:
        print(f"    Lỗi: Không tìm thấy cột '{TIME_COLUMN_STOCK}' trong file Stock Data.")
        return pd.DataFrame()

    # Chuyển đổi cột 'time' sang datetime và lọc
    df[TIME_COLUMN_STOCK] = pd.to_datetime(df[TIME_COLUMN_STOCK])
    df = df[(df[TIME_COLUMN_STOCK] >= START_DATE) & (df[TIME_COLUMN_STOCK] <= END_DATE)]
    df = df.sort_values(TIME_COLUMN_STOCK).reset_index(drop=True)

    # Đổi tên cột 'time' thành 'Thời gian' để merge
    df = df.rename(columns={TIME_COLUMN_STOCK: TIME_COLUMN_SENTIMENT})

    print(f"   Stock data: {len(df)} rows from {df[TIME_COLUMN_SENTIMENT].min()} to {df[TIME_COLUMN_SENTIMENT].max()}")

    return df

# --- 2. Đọc và tính sentiment daily cho CÔNG TY ĐÍCH (SSI) ---
def load_target_sentiment():
    """Đọc và tính sentiment daily cho công ty đích (SSI)"""
    print(f"2. Loading and calculating daily sentiment for {TARGET_STOCK_CODE}...")

    if not os.path.exists(SENTIMENT_FILE_PATH):
        print(f"    Lỗi: Không tìm thấy file Sentiment {TARGET_STOCK_CODE} tại: {SENTIMENT_FILE_PATH}")
        # Thường thì file SSI_sentiment.csv nên nằm ở đây sau khi bạn chạy code xử lý sentiment
        return pd.DataFrame()

    df = pd.read_csv(SENTIMENT_FILE_PATH, encoding='utf-8-sig')

    if TIME_COLUMN_SENTIMENT not in df.columns:
        print(f"    Lỗi: Không tìm thấy cột '{TIME_COLUMN_SENTIMENT}' trong file Sentiment {TARGET_STOCK_CODE}.")
        return pd.DataFrame()

    df[TIME_COLUMN_SENTIMENT] = pd.to_datetime(df[TIME_COLUMN_SENTIMENT])
    df = df[(df[TIME_COLUMN_SENTIMENT] >= START_DATE) & (df[TIME_COLUMN_SENTIMENT] <= END_DATE)]

    # Tính sentiment score (positive - negative)
    df['sentiment_score'] = df['positive'] - df['negative']

    # Nhóm theo ngày và tính trung bình sentiment
    daily_sentiment = df.groupby(df[TIME_COLUMN_SENTIMENT].dt.date).agg({
        'sentiment_score': 'mean'
    }).reset_index()

    daily_sentiment[TIME_COLUMN_SENTIMENT] = pd.to_datetime(daily_sentiment[TIME_COLUMN_SENTIMENT])
    daily_sentiment = daily_sentiment.rename(columns={'sentiment_score': f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'})

    print(f"   {TARGET_STOCK_CODE} sentiment: {len(daily_sentiment)} days")

    return daily_sentiment

# --- 3. Đọc và tính sentiment daily cho Related Companies ---
def load_related_sentiments():
    """Đọc và tính sentiment daily cho các related companies"""
    print("3. Loading and calculating daily sentiment for related companies...")

    related_sentiments = {}

    for company in RELATED_COMPANIES:
        file_path = os.path.join(RELATED_COM_FOLDER, f"{company}_sentiment.csv")

        if not os.path.exists(file_path):
            print(f"    File not found: {file_path}")
            continue

        try:
            df = pd.read_csv(file_path, encoding='utf-8-sig')

            # Kiểm tra và chuẩn hóa cột Thời gian
            if TIME_COLUMN_SENTIMENT not in df.columns or 'positive' not in df.columns or 'negative' not in df.columns:
                print(f"    Missing required columns in {company}_sentiment.csv")
                continue

            df[TIME_COLUMN_SENTIMENT] = pd.to_datetime(df[TIME_COLUMN_SENTIMENT])
            df = df[(df[TIME_COLUMN_SENTIMENT] >= START_DATE) & (df[TIME_COLUMN_SENTIMENT] <= END_DATE)]

            if len(df) == 0:
                print(f"    No data in range for {company}")
                continue

            # Tính sentiment score và nhóm theo ngày
            df['sentiment_score'] = df['positive'] - df['negative']
            daily_sentiment = df.groupby(df[TIME_COLUMN_SENTIMENT].dt.date).agg({
                'sentiment_score': 'mean'
            }).reset_index()

            daily_sentiment[TIME_COLUMN_SENTIMENT] = pd.to_datetime(daily_sentiment[TIME_COLUMN_SENTIMENT])
            daily_sentiment = daily_sentiment.rename(columns={'sentiment_score': f'{company}_daily_new_sentiment_score'})

            related_sentiments[company] = daily_sentiment
            print(f"    {company}: {len(daily_sentiment)} days in range")

        except Exception as e:
            print(f"    Error loading {company}: {e}")

    return related_sentiments

# --- 4. Merge tất cả dữ liệu (Dùng tên cột chung 'Thời gian') ---
def merge_all_data(stock_df, target_sentiment_df, related_sentiments_dict):
    """Merge tất cả dữ liệu vào một DataFrame hoàn chỉnh"""
    print("4. Merging all data...")

    # Tạo DataFrame với date range đầy đủ
    date_range = pd.date_range(start=START_DATE, end=END_DATE, freq='D')
    full_df = pd.DataFrame({TIME_COLUMN_SENTIMENT: date_range})

    # Merge với stock data (đã đổi tên cột thành 'Thời gian')
    full_df = pd.merge(full_df, stock_df, on=TIME_COLUMN_SENTIMENT, how='left')

    # Merge với SSI sentiment
    full_df = pd.merge(full_df, target_sentiment_df, on=TIME_COLUMN_SENTIMENT, how='left')

    # Merge với từng related company sentiment
    for company, sentiment_df in related_sentiments_dict.items():
        full_df = pd.merge(full_df, sentiment_df, on=TIME_COLUMN_SENTIMENT, how='left')

    full_df = full_df.sort_values(TIME_COLUMN_SENTIMENT).reset_index(drop=True)
    return full_df

# --- 5. Xử lý Missing Values ---
def handle_missing_values(df):
    """Xử lý missing values cho các loại dữ liệu khác nhau"""
    print("5. Handling missing values...")

    # Stock features và technical indicators: forward fill, sau đó backward fill
    stock_tech_columns = [
        'open', 'high', 'low', 'close', 'volume',
        'SMA_5', 'SMA_20', 'EMA_10', 'Momentum_3', 'Momentum_10',
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV'
    ]

    for col in stock_tech_columns:
        if col in df.columns:
            # fillna(method='ffill'): lấp đầy bằng giá trị trước đó (giá trị cũ nhất)
            # fillna(method='bfill'): lấp đầy bằng giá trị sau đó (giá trị mới nhất)
            df[col] = df[col].fillna(method='ffill').fillna(method='bfill')

    # Sentiment scores: fill với 0 (không có tin tức/sentiment)
    sentiment_columns = [col for col in df.columns if 'daily_new_sentiment_score' in col]
    for col in sentiment_columns:
        if col in df.columns:
            df[col] = df[col].fillna(0)

    print(f"   Missing values handled for {len(stock_tech_columns) + len(sentiment_columns)} columns")

    return df

# --- 6. Sắp xếp columns theo thứ tự yêu cầu ---
def organize_columns(df):
    """Sắp xếp columns theo thứ tự yêu cầu"""
    print("6. Organizing columns...")

    column_order = [TIME_COLUMN_SENTIMENT]
    column_order.extend(['open', 'high', 'low', 'close', 'volume'])
    column_order.extend([
        'SMA_5', 'SMA_20', 'EMA_10', 'Momentum_3', 'Momentum_10',
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV'
    ])

    # SSI sentiment
    column_order.extend([f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'])

    # Related companies sentiment
    for company in RELATED_COMPANIES:
        sentiment_col = f'{company}_daily_new_sentiment_score'
        if sentiment_col in df.columns:
            column_order.extend([sentiment_col])

    # Chỉ giữ các columns có trong column_order
    final_columns = [col for col in column_order if col in df.columns]
    return df[final_columns]

# ====================================================================
#                           III. MAIN EXECUTION
# ====================================================================

def main():
    print(f"=== {TARGET_STOCK_CODE} FINAL DATASET CREATION ===")
    print(f"Date range: {START_DATE} to {END_DATE}")
    print(f"Output file: {OUTPUT_FILE}")
    print("=" * 50)

    try:
        # Bước 1: Load tất cả dữ liệu
        stock_df = load_stock_data()
        target_sentiment_df = load_target_sentiment()
        related_sentiments_dict = load_related_sentiments()

        if stock_df.empty:
             print(" KHÔNG THỂ TIẾN HÀNH: Dữ liệu Stock Data rỗng hoặc không tải được.")
             return None

        # Bước 2: Merge dữ liệu
        merged_df = merge_all_data(stock_df, target_sentiment_df, related_sentiments_dict)

        # Bước 3: Xử lý missing values
        cleaned_df = handle_missing_values(merged_df)

        # Bước 4: Sắp xếp columns
        final_df = organize_columns(cleaned_df)

        # Bước 5: Lưu file
        print(f"\n7. Saving final dataset...")
        final_df.to_csv(OUTPUT_FILE, index=False, encoding='utf-8-sig')
        print(f"   Saved to: {OUTPUT_FILE}")

        # Bước 6: Tạo báo cáo
        print("\n=== FINAL DATASET PREVIEW ===")
        print(f"Shape: {final_df.shape}")

        # Thiết lập để hiển thị toàn bộ cột
        pd.set_option('display.max_columns', None)
        pd.set_option('display.width', 1000)

        print(f"\nFirst 5 rows:")
        print(final_df.head())

        return final_df

    except Exception as e:
        print(f"Error: {e}")
        import traceback
        traceback.print_exc()
        return None

if __name__ == "__main__":
    final_dataset = main()

    if final_dataset is not None:
        print(f"\n{'='*50}")
        print(" FINAL DATASET CREATION COMPLETED SUCCESSFULLY!")
        print(f"   File: {OUTPUT_FILE}")
        print(f"   Shape: {final_dataset.shape}")
        print(f"{'='*50}")
    else:
        print(f"\n{'='*50}")
        print(" DATASET CREATION FAILED!")
        print(f"{'='*50}")
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ (ĐÃ SỬA MÃ CỔ PHIẾU VÀ ĐƯỜNG DẪN)
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB"
RELATED_COMPANIES = ['TCB', 'VJC', 'HPG', 'PLX', 'VIC', 'SSI', 'GAS','FPT']

# Đường dẫn file (ĐÃ CẬP NHẬT)
FILE_PATH = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"

SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 32
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA (Giữ nguyên)
# =====================================
def load_stock_data_from_df(file_path, symbol):
    """Tải dữ liệu từ file CSV."""
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    df.rename(columns={'Thời gian': 'date'}, inplace=True)

    # Cố gắng chuyển đổi ngày tháng
    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    df = df.sort_values('date').reset_index(drop=True)

    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    print(f"Đã tải {len(df)} dòng dữ liệu")
    print(f"Ngày bắt đầu: {df['date'].min().date()}")
    print(f"Ngày kết thúc: {df['date'].max().date()}")

    return df

def prepare_stock_data(df): # Đổi tên và loại bỏ tính Alpha
    """Chuẩn bị dữ liệu chỉ với các chỉ báo kỹ thuật và giá."""
    df = df.copy()

    # 1. Thêm cột Pct_Change (Mục tiêu dự đoán)
    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # Xóa các hàng có NaN (nếu có)
    return df.dropna().reset_index(drop=True)

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    """Tạo cửa sổ trượt."""
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)


class StockDataset(Dataset):
    # Giữ nguyên
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

# =====================================
# PHẦN 3 & 4. MODEL & TRAIN (Giữ nguyên)
# =====================================

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0).transpose(0, 1)
        self.register_buffer('pe', pe)
    def forward(self, x):
        return x + self.pe[:x.size(0), :]

class TemporalEmbedding(nn.Module):
    def __init__(self, window_size, d_model):
        super(TemporalEmbedding, self).__init__()
        self.time_embedding = nn.Embedding(window_size, d_model)
        self.register_buffer('positions', torch.arange(window_size).long())
    def forward(self, x):
        batch_size, seq_len, _ = x.shape
        pos = self.positions.unsqueeze(0).repeat(batch_size, 1)
        time_emb = self.time_embedding(pos)
        return time_emb

class StockEncoderDecoderTransformer(nn.Module):
    def __init__(self, input_dim_src, input_dim_tgt, d_model=64, nhead=4, num_layers=3, dropout=0.1, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model

        # --- Encoder Components ---
        self.conv_emb_src = nn.Conv1d(input_dim_src, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(window_size, d_model)
        self.pos_encoder = PositionalEncoding(d_model)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components ---
        self.conv_emb_tgt = nn.Conv1d(input_dim_tgt, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(window_size, d_model)
        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 64), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def _get_embedding(self, x, conv_emb_layer, temporal_emb_layer):
        x_conv = x.transpose(1, 2)
        x_conv = conv_emb_layer(x_conv)
        x_conv = x_conv.transpose(1, 2)
        x_temporal = temporal_emb_layer(x)
        x_emb = x_conv * np.sqrt(self.d_model) + x_temporal
        x_final = self.pos_encoder(x_emb.transpose(0, 1)).transpose(0, 1)
        return x_final

    def forward(self, x_src, x_tgt_in):
        src_emb = self._get_embedding(x_src, self.conv_emb_src, self.temporal_emb_src)
        memory = self.transformer_encoder(src_emb)
        tgt_emb = self._get_embedding(x_tgt_in, self.conv_emb_tgt, self.temporal_emb_tgt)
        output = self.transformer_decoder(tgt_emb, memory)
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)

        if (epoch + 1) % 10 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    """Dự đoán Pct_Change của ngày tiếp theo và chuyển thành giá tuyệt đối."""

    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    # Lấy cửa sổ dữ liệu cuối cùng
    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    # Chuẩn bị input cho Encoder (Source - TIs + Giá)
    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    # Chuẩn bị input cho Decoder (Target Input - Close Price)
    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    # Dự đoán Pct_Change đã được scale
    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    # Đảo ngược scale để có Pct_Change thực
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]

    # Tính giá trị tuyệt đối dự đoán
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price

# =====================================
# PHẦN 5. MAIN EXECUTION (Giữ nguyên features, đã cập nhật SYMBOL)
# =====================================
def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER ===")
    print("=== CHỈ SỬ DỤNG TECHNICAL INDICATORS VÀ DỮ LIỆU GIÁ GỐC (KHÔNG DÙNG SENTIMENT) ===")

    print("\n1. Đang tải dữ liệu từ file CSV...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        print(f"Không có dữ liệu hợp lệ cho mã {symbol}.")
        return None

    # Định nghĩa các cột features (CHỈ CÁC CỘT GIÁ VÀ TECHNICAL INDICATORS)
    source_columns = [
        # Dữ liệu giá gốc
        'open', 'high', 'low', 'volume', 'close',
        # Chỉ báo kỹ thuật
        'SMA_5', 'SMA_20', 'EMA_10', 'Momentum_3', 'Momentum_10',
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV'
    ]

    target_input_columns = ['close']
    target_output_column = 'Pct_Change'

    # Kiểm tra cột có sẵn trong file (Technical & Giá)
    cols_to_check_in_file = [col for col in source_columns if col != 'close'] + target_input_columns
    missing_cols_in_file = [col for col in cols_to_check_in_file if col not in df.columns]

    if missing_cols_in_file:
        print(f"LỖI: Thiếu {len(missing_cols_in_file)} cột dữ liệu cần thiết: {', '.join(missing_cols_in_file)}")
        print("Vui lòng kiểm tra lại file CSV đầu vào.")
        return None

    # Sử dụng hàm đã sửa tên
    df_processed = prepare_stock_data(df)

    if target_output_column not in df_processed.columns:
        print(f"LỖI NGHIÊM TRỌNG: Không thể tạo cột mục tiêu '{target_output_column}'.")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"Số lượng features sử dụng (Không bao gồm Sentiment): {len(source_columns)}")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    print("\n2. Đang chia dữ liệu Train/Test theo ngày...")
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    # Lọc DataFrame
    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    # Dữ liệu test là từ 1/1/2025 đến mới nhất
    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    if df_train.empty:
        print(f"LỖI: Không có dữ liệu training trong khoảng {TRAIN_START_DATE.date()} đến {TRAIN_END_DATE.date()}.")
        return None

    if len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test (từ {TEST_START_DATE.date()}) không đủ để tạo cửa sổ trượt (Cần > {WINDOW_SIZE} ngày, hiện có {len(df_test)} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")

    # --- TẠO CỬA SỔ TRƯỢT RIÊNG BIỆT ---
    print("\n3. Đang tạo cửa sổ trượt...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    if len(X_src_train_raw) == 0:
         print(f"LỖI: Không đủ dữ liệu Train để tạo cửa sổ trượt (Window={WINDOW_SIZE}). Cần ít nhất {WINDOW_SIZE+1} ngày.")
         return None
    if len(X_src_test_raw) == 0:
         print(f"LỖI: Không đủ dữ liệu Test để tạo cửa sổ trượt (Window={WINDOW_SIZE}). Cần ít nhất {WINDOW_SIZE+1} ngày.")
         return None

    # --- CHUẨN HÓA DỮ LIỆU ---
    print("\n4. Đang chuẩn hóa dữ liệu...")
    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n5. Đang khởi tạo và huấn luyện mô hình...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns),
        input_dim_tgt=len(target_input_columns),
        d_model=64,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    print("\n6. Đang đánh giá trên tập Test...")
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()
    pct_change_actual_unscaled = scaler_tgt_output.inverse_transform(y_test.reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    close_price_T = X_tgt_in_test_raw[:, -1].squeeze()

    min_len = min(len(close_price_T), len(pct_change_pred_unscaled))
    if len(close_price_T) != len(pct_change_pred_unscaled):
        close_price_T = close_price_T[:min_len]
        pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
        pct_change_actual_unscaled = pct_change_actual_unscaled[:min_len]
        test_dates = test_dates[:min_len]

    predictions_unscaled = close_price_T * (1 + pct_change_pred_unscaled / 100)
    actuals_unscaled = close_price_T * (1 + pct_change_actual_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*60)
    print("KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST (CHỈ DÙNG TECHNICAL)")
    print("="*60)
    print(f"Mã: {symbol}")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*60)

    # Trả về các kết quả cần thiết để sử dụng bên ngoài
    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# PHẦN 6. VẼ BIỂU ĐỒ SO SÁNH (Giữ nguyên)
# =====================================
def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    """Vẽ biểu đồ so sánh giá thực tế và giá dự đoán trên tập Test."""
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 7))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    plt.title(f'So sánh Dự đoán và Thực tế Cổ phiếu {symbol} (CHỈ TECHNICAL)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16)
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()


# =====================================
# PHẦN 7. THỰC THI CHÍNH (Giữ nguyên)
# =====================================
if __name__ == '__main__':
    # Chạy quy trình huấn luyện và đánh giá
    training_results = main(FILE_PATH, SYMBOL)

    if training_results:
        # Vẽ biểu đồ kết quả trên tập test
        print("\n7. Đang vẽ biểu đồ kết quả trên tập test...")
        plot_predictions_vs_actuals(
            test_dates=training_results["test_dates"],
            predictions=training_results["predictions_unscaled"],
            actuals=training_results["actuals_unscaled"],
            symbol=SYMBOL,
            mape=training_results["metrics"]["mape"] # Thêm tham số MAPE
        )

        # Dự đoán cho ngày tiếp theo sau ngày cuối cùng trong dữ liệu
        print("\n8. Đang dự đoán giá cho ngày giao dịch tiếp theo...")
        next_day_price = predict_next_day(
            model=training_results["model"],
            scaler_src=training_results["scalers"]["src"],
            scaler_tgt_input=training_results["scalers"]["tgt_input"],
            scaler_tgt_output=training_results["scalers"]["tgt_output"],
            recent_data=training_results["df_processed"],
            src_cols=training_results["columns"]["source"],
            tgt_in_cols=training_results["columns"]["target_input"]
        )

        if next_day_price is not None:
            last_date = training_results["df_processed"]['date'].iloc[-1].date()
            last_price = training_results['df_processed']['close'].iloc[-1]
            price_change = next_day_price - last_price
            change_percent = (price_change / last_price) * 100

            print("\n" + "="*50)
            print("DỰ ĐOÁN GIÁ TIẾP THEO")
            print("="*50)
            print(f"Mã cổ phiếu: {SYMBOL}")
            print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
            print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
            print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
            print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
            print("="*50)
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ (ĐÃ SỬA MÃ CỔ PHIẾU VÀ ĐƯỜNG DẪN)
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB"
RELATED_COMPANIES = ['TCB', 'VJC', 'HPG', 'PLX', 'VIC', 'SSI', 'GAS','FPT']

# Đường dẫn file
# Đảm bảo đường dẫn này khớp với vị trí file FPT_final.csv của bạn
FILE_PATH = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"

SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 32
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA (Giữ nguyên)
# =====================================
def load_stock_data_from_df(file_path, symbol):
    """Tải dữ liệu từ file CSV."""
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    df.rename(columns={'Thời gian': 'date'}, inplace=True)

    # Cố gắng chuyển đổi ngày tháng
    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    df = df.sort_values('date').reset_index(drop=True)

    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    print(f"Đã tải {len(df)} dòng dữ liệu")
    print(f"Ngày bắt đầu: {df['date'].min().date()}")
    print(f"Ngày kết thúc: {df['date'].max().date()}")

    return df

def prepare_stock_data(df):
    """Chuẩn bị dữ liệu: Tính Pct_Change và xử lý NaN."""
    df = df.copy()

    # 1. Thêm cột Pct_Change (Mục tiêu dự đoán)
    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # Lưu ý: Các sentiment score được giả định là đã có giá trị 0 hoặc hợp lệ
    # cho các ngày không có tin tức (như trong dữ liệu ví dụ)

    # Xóa các hàng có NaN
    return df.dropna().reset_index(drop=True)

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    """Tạo cửa sổ trượt."""
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)


class StockDataset(Dataset):
    # Giữ nguyên
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

# =====================================
# PHẦN 3 & 4. MODEL & TRAIN (Giữ nguyên)
# =====================================

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0).transpose(0, 1)
        self.register_buffer('pe', pe)
    def forward(self, x):
        return x + self.pe[:x.size(0), :]

class TemporalEmbedding(nn.Module):
    def __init__(self, window_size, d_model):
        super(TemporalEmbedding, self).__init__()
        self.time_embedding = nn.Embedding(window_size, d_model)
        self.register_buffer('positions', torch.arange(window_size).long())
    def forward(self, x):
        batch_size, seq_len, _ = x.shape
        pos = self.positions.unsqueeze(0).repeat(batch_size, 1)
        time_emb = self.time_embedding(pos)
        return time_emb

class StockEncoderDecoderTransformer(nn.Module):
    def __init__(self, input_dim_src, input_dim_tgt, d_model=64, nhead=4, num_layers=3, dropout=0.1, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model

        # --- Encoder Components ---
        self.conv_emb_src = nn.Conv1d(input_dim_src, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(window_size, d_model)
        self.pos_encoder = PositionalEncoding(d_model)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components ---
        self.conv_emb_tgt = nn.Conv1d(input_dim_tgt, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(window_size, d_model)
        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 64), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def _get_embedding(self, x, conv_emb_layer, temporal_emb_layer):
        x_conv = x.transpose(1, 2)
        x_conv = conv_emb_layer(x_conv)
        x_conv = x_conv.transpose(1, 2)
        x_temporal = temporal_emb_layer(x)
        x_emb = x_conv * np.sqrt(self.d_model) + x_temporal
        x_final = self.pos_encoder(x_emb.transpose(0, 1)).transpose(0, 1)
        return x_final

    def forward(self, x_src, x_tgt_in):
        src_emb = self._get_embedding(x_src, self.conv_emb_src, self.temporal_emb_src)
        memory = self.transformer_encoder(src_emb)
        tgt_emb = self._get_embedding(x_tgt_in, self.conv_emb_tgt, self.temporal_emb_tgt)
        output = self.transformer_decoder(tgt_emb, memory)
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)

        if (epoch + 1) % 10 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    """Dự đoán Pct_Change của ngày tiếp theo và chuyển thành giá tuyệt đối."""

    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    # Lấy cửa sổ dữ liệu cuối cùng
    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    # Chuẩn bị input cho Encoder (Source - ALL FEATURES)
    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    # Chuẩn bị input cho Decoder (Target Input - Close Price)
    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    # Dự đoán Pct_Change đã được scale
    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    # Đảo ngược scale để có Pct_Change thực
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]

    # Tính giá trị tuyệt đối dự đoán
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price

# =====================================
# PHẦN 5. MAIN EXECUTION (ĐÃ SỬA CÁC CỘT SENTIMENT)
# =====================================
def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER (FULL FEATURES) ===")
    print("=== SỬ DỤNG TECHNICAL INDICATORS, DỮ LIỆU GIÁ GỐC VÀ 11 SENTIMENT SCORES ===")

    print("\n1. Đang tải dữ liệu từ file CSV...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        print(f"Không có dữ liệu hợp lệ cho mã {symbol}.")
        return None

    # Tạo danh sách các cột sentiment scores dựa trên mã mục tiêu và mã liên quan MỚI
    # FPT là mã mục tiêu (1), RELATED_COMPANIES là 10 mã liên quan (10). Tổng cộng 11 cột.
    all_sentiment_codes = RELATED_COMPANIES + [TARGET_STOCK_CODE]
    sentiment_columns = [f'{code}_daily_new_sentiment_score' for code in all_sentiment_codes]

    technical_and_price_columns = [
        'open', 'high', 'low', 'close', 'volume',
        'SMA_5', 'SMA_20', 'EMA_10', 'Momentum_3', 'Momentum_10',
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV'
    ]

    # Tất cả các cột sẽ là Source Features (Input cho Encoder)
    source_columns = technical_and_price_columns + sentiment_columns
    target_input_columns = ['close'] # Target Input (Input cho Decoder)
    target_output_column = 'Pct_Change'

    # Kiểm tra cột có sẵn trong file
    missing_cols_in_file = [col for col in source_columns if col not in df.columns]

    if missing_cols_in_file:
        print(f"LỖI: Thiếu {len(missing_cols_in_file)} cột dữ liệu cần thiết: {', '.join(missing_cols_in_file)}")
        print("Vui lòng kiểm tra lại file CSV đầu vào để đảm bảo có đủ 11 cột sentiment scores mới.")
        return None

    # Sử dụng hàm đã sửa tên
    df_processed = prepare_stock_data(df)

    if target_output_column not in df_processed.columns:
        print(f"LỖI NGHIÊM TRỌNG: Không thể tạo cột mục tiêu '{target_output_column}'.")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"Số lượng features sử dụng (ALL): {len(source_columns)} (Gồm {len(sentiment_columns)} cột Sentiment)")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    print("\n2. Đang chia dữ liệu Train/Test theo ngày...")
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    # Lọc DataFrame
    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    # Dữ liệu test là từ 1/1/2025 đến mới nhất
    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    if df_train.empty:
        print(f"LỖI: Không có dữ liệu training trong khoảng {TRAIN_START_DATE.date()} đến {TRAIN_END_DATE.date()}.")
        return None

    if len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test (từ {TEST_START_DATE.date()}) không đủ để tạo cửa sổ trượt (Cần > {WINDOW_SIZE} ngày, hiện có {len(df_test)} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")

    # --- TẠO CỬA SỔ TRƯỢT RIÊNG BIỆT ---
    print("\n3. Đang tạo cửa sổ trượt...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    if len(X_src_train_raw) == 0:
         print(f"LỖI: Không đủ dữ liệu Train để tạo cửa sổ trượt (Window={WINDOW_SIZE}). Cần ít nhất {WINDOW_SIZE+1} ngày.")
         return None
    if len(X_src_test_raw) == 0:
         print(f"LỖI: Dữ liệu Test để tạo cửa sổ trượt (Window={WINDOW_SIZE}). Cần ít nhất {WINDOW_SIZE+1} ngày.")
         return None

    # --- CHUẨN HÓA DỮ LIỆU ---
    print("\n4. Đang chuẩn hóa dữ liệu...")
    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n5. Đang khởi tạo và huấn luyện mô hình...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns), # SỐ LƯỢNG INPUT CHO ENCODER ĐÃ ĐƯỢC CẬP NHẬT
        input_dim_tgt=len(target_input_columns),
        d_model=64,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    print("\n6. Đang đánh giá trên tập Test...")
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()
    pct_change_actual_unscaled = scaler_tgt_output.inverse_transform(y_test.reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    close_price_T = X_tgt_in_test_raw[:, -1].squeeze()

    min_len = min(len(close_price_T), len(pct_change_pred_unscaled))
    if len(close_price_T) != len(pct_change_pred_unscaled):
        close_price_T = close_price_T[:min_len]
        pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
        pct_change_actual_unscaled = pct_change_actual_unscaled[:min_len]
        test_dates = test_dates[:min_len]

    predictions_unscaled = close_price_T * (1 + pct_change_pred_unscaled / 100)
    actuals_unscaled = close_price_T * (1 + pct_change_actual_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*60)
    print("KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST (FULL FEATURES: TECHNICAL + SENTIMENT)")
    print("="*60)
    print(f"Mã: {symbol}")
    print(f"Tổng số features: {len(source_columns)}")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*60)

    # Trả về các kết quả cần thiết để sử dụng bên ngoài
    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# PHẦN 6. VẼ BIỂU ĐỒ SO SÁNH (Giữ nguyên)
# =====================================
def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    """Vẽ biểu đồ so sánh giá thực tế và giá dự đoán trên tập Test."""
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 7))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    # Cập nhật tiêu đề để phản ánh việc sử dụng Full Features
    plt.title(f'So sánh Dự đoán và Thực tế Cổ phiếu {symbol} (FULL FEATURES: TECH + SENTIMENT)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16)
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()


# =====================================
# PHẦN 7. THỰC THI CHÍNH (Giữ nguyên)
# =====================================
if __name__ == '__main__':
    # Chạy quy trình huấn luyện và đánh giá
    training_results = main(FILE_PATH, SYMBOL)

    if training_results:
        # Vẽ biểu đồ kết quả trên tập test
        print("\n7. Đang vẽ biểu đồ kết quả trên tập test...")
        plot_predictions_vs_actuals(
            test_dates=training_results["test_dates"],
            predictions=training_results["predictions_unscaled"],
            actuals=training_results["actuals_unscaled"],
            symbol=SYMBOL,
            mape=training_results["metrics"]["mape"] # Thêm tham số MAPE
        )

        # Dự đoán cho ngày tiếp theo sau ngày cuối cùng trong dữ liệu
        print("\n8. Đang dự đoán giá cho ngày giao dịch tiếp theo...")
        next_day_price = predict_next_day(
            model=training_results["model"],
            scaler_src=training_results["scalers"]["src"],
            scaler_tgt_input=training_results["scalers"]["tgt_input"],
            scaler_tgt_output=training_results["scalers"]["tgt_output"],
            recent_data=training_results["df_processed"],
            src_cols=training_results["columns"]["source"],
            tgt_in_cols=training_results["columns"]["target_input"]
        )

        if next_day_price is not None:
            last_date = training_results["df_processed"]['date'].iloc[-1].date()
            last_price = training_results['df_processed']['close'].iloc[-1]
            price_change = next_day_price - last_price
            change_percent = (price_change / last_price) * 100

            print("\n" + "="*50)
            print("DỰ ĐOÁN GIÁ TIẾP THEO")
            print("="*50)
            print(f"Mã cổ phiếu: {SYMBOL}")
            print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
            print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
            print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
            print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
            print("="*50)
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os
import traceback # Import để in lỗi chi tiết

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ (ĐÃ CẬP NHẬT CHO VCB)
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB" # <--- MÃ CỔ PHIẾU MỤC TIÊU MỚI
RELATED_COMPANIES = ['TCB', 'VJC', 'HPG', 'PLX', 'VIC', 'SSI', 'GAS','FPT'] # Danh sách công ty liên quan MỚI
# Dùng để xác định các cột sentiment MỚI sẽ có trong file
ALL_SENTIMENT_COMPANIES = [TARGET_STOCK_CODE] + RELATED_COMPANIES

# Đường dẫn file đã được sửa
FILE_PATH = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"

SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 32
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA (CẬP NHẬT CÔNG THỨC ALPHA MỚI CHO VCB)
# =====================================
def load_stock_data_from_df(file_path, symbol):
    """Tải dữ liệu từ file CSV đã có đầy đủ chỉ báo kỹ thuật và sentiment."""
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        # Đọc file CSV với encoding phù hợp
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    # Đổi tên cột 'Thời gian' thành 'date'
    df.rename(columns={'Thời gian': 'date'}, inplace=True)

    # Cố gắng chuyển đổi ngày tháng
    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    # Sắp xếp theo ngày
    df = df.sort_values('date').reset_index(drop=True)

    # Lọc dữ liệu từ 2024-01-01 trở đi
    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    print(f"Đã tải {len(df)} dòng dữ liệu")
    print(f"Ngày bắt đầu: {df['date'].min().date()}")
    print(f"Ngày kết thúc: {df['date'].max().date()}")

    return df

def calculate_alpha_features(df):
    """
    Tính toán 5 trường alpha mới cho VCB:
    alpha1: ((close - open) / (BB_Upper - BB_Lower).replace(0, 1e-6)) * Momentum_5 * (1 + VCB_daily_new_sentiment_score)
    alpha2: log(RSI_14) * ((MACD - MACD_Signal) / close).replace(0, 1e-6)
    alpha3: (VCB_daily_new_sentiment_score - TCB_daily_new_sentiment_score) * (Momentum_10_VCB / Momentum_10_TCB.replace(0, 1e-6))
    alpha4: ((close - EMA_20) / volume.replace(0, 1e-6)) * (OBV - OBV.shift(1)).fillna(0)
    alpha5: (VCB_daily_new_sentiment_score - HPG_daily_new_sentiment_score) * Momentum_3
    """
    df = df.copy()

    # Định nghĩa các cột sentiment và TI cần thiết (cột TI của VCB không cần hậu tố)
    sentiment_vcb = f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'
    sentiment_tcb = 'TCB_daily_new_sentiment_score'
    sentiment_hpg = 'HPG_daily_new_sentiment_score'
    momentum_tcb = 'Momentum_10_TCB'
    momentum_vcb = 'Momentum_10' # Cột Momentum_10 của VCB là Momentum_10

    # Lọc các cột có thể bị thiếu (ví dụ: các chỉ báo của công ty liên quan)
    required_cols_for_alpha = [sentiment_vcb, 'close', 'open', 'BB_Upper', 'BB_Lower',
                               'Momentum_5', 'RSI_14', 'MACD', 'MACD_Signal',
                               sentiment_tcb, momentum_tcb, 'EMA_20', 'volume', 'OBV',
                               sentiment_hpg, 'Momentum_3']

    missing_alpha_cols = [col for col in required_cols_for_alpha if col not in df.columns]
    if missing_alpha_cols:
        print(f"CẢNH BÁO: Thiếu các cột sau để tính Alpha Features: {missing_alpha_cols}. Thay thế bằng 0.")
        for col in missing_alpha_cols:
            df[col] = 0.0

    # --- CÁC CÔNG THỨC ALPHA MỚI CHO VCB ---

    # Cột phụ cần thiết
    df['BB_Range'] = df['BB_Upper'] - df['BB_Lower']
    df['OBV_Change'] = df['OBV'] - df['OBV'].shift(1).fillna(df['OBV'])

    # alpha1: ((close - open) / (BB_Upper - BB_Lower)) * Momentum_5 * (1 + VCB_daily_new_sentiment_score)
    df['alpha1'] = np.where(
        df['BB_Range'] == 0,
        0,
        (df['close'] - df['open']) / df['BB_Range']
    ) * df['Momentum_5'] * (1 + df[sentiment_vcb])

    # alpha2: log(RSI_14) * ((MACD - MACD_Signal) / close)
    # Xử lý log(0) và chia cho 0
    df['RSI_Safe'] = np.where(df['RSI_14'] <= 1e-6, 1e-6, df['RSI_14']) # Tránh log(0)

    df['alpha2_ratio'] = np.where(
        df['close'] == 0,
        0,
        (df['MACD'] - df['MACD_Signal']) / df['close']
    )

    df['alpha2'] = np.log(df['RSI_Safe']) * df['alpha2_ratio']

    # alpha3: (VCB_daily_new_sentiment_score - TCB_daily_new_sentiment_score) * (Momentum_10_VCB / Momentum_10_TCB)
    df['alpha3_ratio'] = np.where(
        df[momentum_tcb] == 0,
        0,
        df[momentum_vcb] / df[momentum_tcb]
    )

    df['alpha3'] = (df[sentiment_vcb] - df[sentiment_tcb]) * df['alpha3_ratio']

    # alpha4: ((close - EMA_20) / volume) * (OBV - OBV.shift(1))
    df['alpha4_ratio'] = np.where(
        df['volume'] == 0,
        0,
        (df['close'] - df['EMA_20']) / df['volume']
    )

    df['alpha4'] = df['alpha4_ratio'] * df['OBV_Change']

    # alpha5: (VCB_daily_new_sentiment_score - HPG_daily_new_sentiment_score) * Momentum_3
    df['alpha5'] = (df[sentiment_vcb] - df[sentiment_hpg]) * df['Momentum_3']

    # Xóa các cột tạm
    df = df.drop(['BB_Range', 'OBV_Change', 'RSI_Safe', 'alpha2_ratio', 'alpha3_ratio', 'alpha4_ratio'],
                 axis=1, errors='ignore')

    # Thay thế các giá trị vô cùng và NaN bằng 0
    df = df.replace([np.inf, -np.inf], 0)
    df = df.fillna(0)

    print("Đã tính toán xong 5 alpha features mới cho VCB.")
    return df

def prepare_data_with_sentiment(df):
    """Chuẩn bị dữ liệu với các chỉ báo kỹ thuật, sentiment scores và alpha features."""
    df = df.copy()

    # 1. Tính toán các alpha features
    df = calculate_alpha_features(df)

    # 2. Thêm cột Pct_Change (Mục tiêu dự đoán)
    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # Xóa các hàng có NaN (nếu có)
    # Chú ý: sau khi fillna(0) ở alpha, bước này chủ yếu loại bỏ các hàng NaN do TI (ví dụ SMA/Momentum ban đầu)
    df_cleaned = df.dropna().reset_index(drop=True)

    print(f"Sau khi làm sạch: {len(df_cleaned)} dòng dữ liệu")
    return df_cleaned

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    """Tạo cửa sổ trượt."""
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)

class StockDataset(Dataset):
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

# =====================================
# PHẦN 3. ĐỊNH NGHĨA MODEL (Giữ nguyên)
# =====================================
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0).transpose(0, 1)
        self.register_buffer('pe', pe)
    def forward(self, x):
        return x + self.pe[:x.size(0), :]

class TemporalEmbedding(nn.Module):
    def __init__(self, window_size, d_model):
        super(TemporalEmbedding, self).__init__()
        self.time_embedding = nn.Embedding(window_size, d_model)
        self.register_buffer('positions', torch.arange(window_size).long())
    def forward(self, x):
        batch_size, seq_len, _ = x.shape
        pos = self.positions.unsqueeze(0).repeat(batch_size, 1)
        time_emb = self.time_embedding(pos)
        return time_emb

class StockEncoderDecoderTransformer(nn.Module):
    def __init__(self, input_dim_src, input_dim_tgt, d_model=64, nhead=4, num_layers=3, dropout=0.1, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model

        # --- Encoder Components ---
        self.conv_emb_src = nn.Conv1d(input_dim_src, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(window_size, d_model)
        self.pos_encoder = PositionalEncoding(d_model)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components ---
        self.conv_emb_tgt = nn.Conv1d(input_dim_tgt, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(window_size, d_model)
        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 64), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def _get_embedding(self, x, conv_emb_layer, temporal_emb_layer):
        x_conv = x.transpose(1, 2)
        x_conv = conv_emb_layer(x_conv)
        x_conv = x_conv.transpose(1, 2)
        x_temporal = temporal_emb_layer(x)
        x_emb = x_conv * np.sqrt(self.d_model) + x_temporal
        x_final = self.pos_encoder(x_emb.transpose(0, 1)).transpose(0, 1)
        return x_final

    def forward(self, x_src, x_tgt_in):
        src_emb = self._get_embedding(x_src, self.conv_emb_src, self.temporal_emb_src)
        memory = self.transformer_encoder(src_emb)
        tgt_emb = self._get_embedding(x_tgt_in, self.conv_emb_tgt, self.temporal_emb_tgt)
        output = self.transformer_decoder(tgt_emb, memory)
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

# =====================================
# PHẦN 4. TRAIN & EVALUATE (Giữ nguyên)
# =====================================
def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        train_losses.append(train_loss)
        val_losses.append(val_loss)

        if (epoch + 1) % 10 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    """Dự đoán giá đóng cửa của ngày tiếp theo."""

    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    # Lấy cửa sổ dữ liệu cuối cùng
    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    # Chuẩn bị input cho Encoder (Source - TIs + Sentiment + Alpha)
    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    # Chuẩn bị input cho Decoder (Target Input - Close Price)
    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    # Dự đoán Pct_Change đã được scale
    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    # Đảo ngược scale để có Pct_Change thực
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]

    # Tính giá trị tuyệt đối dự đoán
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price

# =====================================
# PHẦN 5. MAIN EXECUTION (CẬP NHẬT COLUMNS CHO VCB)
# =====================================
def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER ===")
    print("=== VỚI 5 ALPHA FEATURES MỚI KẾT HỢP TECHNICAL INDICATORS VÀ SENTIMENT CỦA VCB VÀ CÁC CÔNG TY LIÊN QUAN ===")

    print("\n1. Đang tải dữ liệu từ file CSV...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        print(f"Không có dữ liệu hợp lệ cho mã {symbol}.")
        return None

    df_processed = prepare_data_with_sentiment(df)

    # Lấy danh sách cột sentiment score cần thiết cho Source Columns
    sentiment_columns = [f'{sym}_daily_new_sentiment_score' for sym in ALL_SENTIMENT_COMPANIES]

    # Loại bỏ các cột sentiment không có trong df (ví dụ: nếu file thiếu sentiment của công ty liên quan)
    sentiment_columns = [col for col in sentiment_columns if col in df_processed.columns]

    # Thêm cột Momentum_10_TCB nếu nó tồn tại (cần cho alpha3)
    ti_related_columns = ['Momentum_10_TCB']
    ti_related_columns = [col for col in ti_related_columns if col in df_processed.columns]


    # Định nghĩa các cột features - bao gồm cả chỉ báo kỹ thuật, sentiment và alpha features
    source_columns = [
        # Chỉ báo kỹ thuật cơ bản của VCB
        'open', 'high', 'low', 'close', 'volume',
        'SMA_5', 'SMA_20', 'EMA_10', 'EMA_20', 'Momentum_3', 'Momentum_5', 'Momentum_10',
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV',
    ] + ti_related_columns + [
        # Sentiment scores
    ] + sentiment_columns + [
        # Alpha features mới
        'alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5'
    ]

    # Loại bỏ các cột trùng lặp (nếu có) và các cột không tồn tại trong DataFrame
    source_columns = list(set([col for col in source_columns if col in df_processed.columns]))

    target_input_columns = ['close']
    target_output_column = 'Pct_Change'

    # Kiểm tra cột có sẵn trong file
    missing_cols_in_df = [col for col in source_columns if col not in df_processed.columns]
    if missing_cols_in_df:
        # Lỗi này chỉ xảy ra nếu logic ở trên có vấn đề, vì chúng ta đã lọc các cột không tồn tại.
        print(f"\nLỖI LỌC CỘT: Thiếu các cột dữ liệu cần thiết: {', '.join(missing_cols_in_df)}")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"Số lượng features sử dụng (TIs + {len(sentiment_columns)} Sentiments + 5 Alpha): {len(source_columns)}")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    print("\n2. Đang chia dữ liệu Train/Test theo ngày...")
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    # Lọc DataFrame
    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    # Dữ liệu test là từ 1/1/2025 đến mới nhất
    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    if df_train.empty or len(df_train) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu training không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None
    if df_test.empty or len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")

    # --- TẠO CỬA SỔ TRƯỢT RIÊNG BIỆT ---
    print("\n3. Đang tạo cửa sổ trượt...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    # Lấy ngày cho tập test (bỏ đi N ngày đầu tiên do window)
    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    # --- CHUẨN HÓA DỮ LIỆU ---
    print("\n4. Đang chuẩn hóa dữ liệu...")
    # Scaler 1: Cho dữ liệu nguồn (TIs + Sentiment + Alpha features)
    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    # Scaler 2: Cho dữ liệu đầu vào của decoder (Close Price)
    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    # Scaler 3: Cho dữ liệu đầu ra (Pct_Change)
    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n5. Đang khởi tạo và huấn luyện mô hình...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns),
        input_dim_tgt=len(target_input_columns),
        d_model=64,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    print("\n6. Đang đánh giá trên tập Test...")
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()
    pct_change_actual_unscaled = scaler_tgt_output.inverse_transform(y_test.reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    # Lấy giá đóng cửa T-1 (Giá tại cuối window)
    close_price_T = df_test['close'].iloc[WINDOW_SIZE:].values

    # Đảm bảo close_price_T có cùng số phần tử với pct_change_pred_unscaled
    min_len = min(len(close_price_T), len(pct_change_pred_unscaled))

    close_price_T = close_price_T[:min_len]
    pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
    pct_change_actual_unscaled = pct_change_actual_unscaled[:min_len]
    test_dates = test_dates[:min_len]

    predictions_unscaled = close_price_T * (1 + pct_change_pred_unscaled / 100)
    actuals_unscaled = close_price_T * (1 + pct_change_actual_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    # Thêm 1e-6 để tránh lỗi chia cho 0
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*60)
    print("KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST (5 ALPHA MỚI + SENTIMENT CỦA VCB + LIÊN QUAN)")
    print("="*60)
    print(f"Mã: {symbol}")
    print(f"Tổng số features: {len(source_columns)}")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*60)

    # Trả về các kết quả cần thiết để sử dụng bên ngoài
    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# PHẦN 6. VẼ BIỂU ĐỒ SO SÁNH (CẬP NHẬT TIÊU ĐỀ)
# =====================================
def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    """Vẽ biểu đồ so sánh giá thực tế và giá dự đoán trên tập Test."""
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 8))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2, alpha=0.8)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    # Cập nhật tiêu đề
    plt.title(f'SO SÁNH DỰ ĐOÁN VÀ THỰC TẾ CỔ PHIẾU {symbol} (5 ALPHA MỚI + SENTIMENT VCB + LIÊN QUAN)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16, fontweight='bold')
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend(fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# =====================================
# THỰC THI CHÍNH
# =====================================
if __name__ == '__main__':
    # Chạy quy trình huấn luyện và đánh giá
    training_results = main(FILE_PATH, SYMBOL)

    if training_results:
        # Vẽ biểu đồ kết quả trên tập test
        print("\n7. Đang vẽ biểu đồ kết quả...")
        plot_predictions_vs_actuals(
            test_dates=training_results["test_dates"],
            predictions=training_results["predictions_unscaled"],
            actuals=training_results["actuals_unscaled"],
            symbol=SYMBOL,
            mape=training_results["metrics"]["mape"]
        )

        # Dự đoán cho ngày tiếp theo sau ngày cuối cùng trong dữ liệu
        print("\n8. Dự đoán giá cho ngày giao dịch tiếp theo...")
        next_day_price = predict_next_day(
            model=training_results["model"],
            scaler_src=training_results["scalers"]["src"],
            scaler_tgt_input=training_results["scalers"]["tgt_input"],
            scaler_tgt_output=training_results["scalers"]["tgt_output"],
            recent_data=training_results["df_processed"],
            src_cols=training_results["columns"]["source"],
            tgt_in_cols=training_results["columns"]["target_input"]
        )

        if next_day_price is not None:
            last_date = training_results["df_processed"]['date'].iloc[-1].date()
            last_price = training_results['df_processed']['close'].iloc[-1]
            price_change = next_day_price - last_price
            change_percent = (price_change / last_price) * 100

            print("\n" + "="*50)
            print("DỰ ĐOÁN GIÁ TIẾP THEO")
            print("="*50)
            print(f"Mã cổ phiếu: {SYMBOL}")
            print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
            print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
            print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
            print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
            print("="*50)
```

```python
import pandas as pd
import numpy as np
import os
from datetime import datetime
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import traceback # Import để in lỗi chi tiết hơn

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ (ĐÃ CẬP NHẬT VCB VÀ FILE_PATH)
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB"
RELATED_COMPANIES = ['TCB', 'VJC', 'HPG', 'PLX', 'VIC', 'SSI', 'GAS', 'FPT']
ALL_SENTIMENT_COMPANIES = [TARGET_STOCK_CODE] + RELATED_COMPANIES

# Đường dẫn file đã được sửa theo yêu cầu của bạn
FILE_PATH = "/content/VCB_final.csv" # <--- ĐÃ SỬA FILE PATH

SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 32
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA (CẬP NHẬT CÔNG THỨC ALPHA VCB)
# =====================================

def calculate_missing_technical_indicators(df):
    """
    Tính toán các chỉ báo kỹ thuật còn thiếu để phục vụ tính toán Alpha.
    Các chỉ báo cần thiếu: SMA_10, SMA_50, EMA_20.
    """
    df = df.copy()

    # 1. SMA 10 ngày (Bổ sung do thiếu)
    if 'SMA_10' not in df.columns:
        print("-> Đang tính toán chỉ báo SMA_10 (10 ngày)...") # <--- BỔ SUNG
        df['SMA_10'] = df['close'].rolling(window=10).mean() # <--- BỔ SUNG

    # 2. SMA 50 ngày
    if 'SMA_50' not in df.columns:
        print("-> Đang tính toán chỉ báo SMA_50...")
        df['SMA_50'] = df['close'].rolling(window=50).mean()

    # 3. EMA 20 ngày
    if 'EMA_20' not in df.columns:
        print("-> Đang tính toán chỉ báo EMA_20...")
        df['EMA_20'] = df['close'].ewm(span=20, adjust=False).mean()

    # Tạm thời điền 0 cho các giá trị NaN do cửa sổ lăn ban đầu
    df = df.fillna(0)

    return df

def load_stock_data_from_df(file_path, symbol):
    """Tải dữ liệu từ file CSV, đổi tên cột và tính toán TIs còn thiếu."""
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        # Đọc file CSV với encoding phù hợp
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    # Đổi tên cột 'Thời gian' thành 'date'
    df.rename(columns={'Thời gian': 'date'}, inplace=True)

    # Cố gắng chuyển đổi ngày tháng
    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    # Sắp xếp theo ngày
    df = df.sort_values('date').reset_index(drop=True)

    # Lọc dữ liệu từ 2024-01-01 trở đi
    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    # Thêm bước tính TIs còn thiếu
    print("\n-> Đang kiểm tra và tính toán các Chỉ báo Kỹ thuật còn thiếu (SMA_50, EMA_20)...")
    df = calculate_missing_technical_indicators(df)

    print(f"Đã tải {len(df)} dòng dữ liệu")
    print(f"Ngày bắt đầu: {df['date'].min().date()}")
    print(f"Ngày kết thúc: {df['date'].max().date()}")

    return df

def calculate_alpha_features(df):
    """
    Tính toán 5 trường alpha mới cho VCB dựa trên công thức đã cung cấp.
    """
    df = df.copy()

    # 1. TÍNH CÁC CỘT PHỤ CẦN THIẾT

    # A. Sentiment Score của VCB và Trung bình các công ty liên quan
    sentiment_vcb_col = f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'
    related_sentiment_cols = [f'{c}_daily_new_sentiment_score' for c in RELATED_COMPANIES if f'{c}_daily_new_sentiment_score' in df.columns]

    if sentiment_vcb_col not in df.columns:
         print(f"LỖI: Thiếu cột sentiment của VCB: {sentiment_vcb_col}. Các Alpha có sentiment sẽ bị set về 0.")
         df[sentiment_vcb_col] = 0.0

    # Tính trung bình sentiment của các công ty liên quan
    if related_sentiment_cols:
        # Chỉ tính trung bình trên các cột không phải NaN, nếu tất cả là NaN thì trả về 0
        df['avg_related_sent'] = df[related_sentiment_cols].mean(axis=1).fillna(0)
    else:
        # print("CẢNH BÁO: Không tìm thấy cột sentiment của công ty liên quan. avg_related_sent set về 0.")
        df['avg_related_sent'] = 0.0

    # Độ phân kỳ tâm lý
    df['sent_divergence'] = df[sentiment_vcb_col] - df['avg_related_sent']

    # B. Chỉ báo kỹ thuật phụ
    df['ret_5d'] = df['close'].pct_change(periods=5).fillna(0) * 100 # Lợi suất 5 ngày
    df['vol_spike'] = df['volume'] / df['volume'].rolling(window=20).mean().replace(0, 1e-6) # Volume Spike
    df['MACD_Hist'] = df['MACD'] - df['MACD_Signal']
    df['BB_Range'] = df['BB_Upper'] - df['BB_Lower']

    # --- CÁC CÔNG THỨC ALPHA MỚI CHO VCB ---

    # alpha1: ret_5d + 0.5 * (vcb_sentiment - avg_related_sent)
    df['alpha1'] = df['ret_5d'] + 0.5 * df['sent_divergence']

    # alpha2: ((sma_10 - sma_50) / close.replace(0, 1e-6)) * (1 - rsi_14 / 100)
    df['close_safe'] = df['close'].replace(0, 1e-6)
    df['alpha2'] = ((df['SMA_10'] - df['SMA_50']) / df['close_safe']) * (1 - df['RSI_14'] / 100)

    # alpha3: macd_hist * np.sign(sent_divergence)
    df['alpha3'] = df['MACD_Hist'] * np.sign(df['sent_divergence'])

    # alpha4: vol_spike * ((close - ema_20) / ema_20.replace(0, 1e-6)) + 0.3 * (vcb_sentiment - avg_related_sent)
    df['EMA_20_safe'] = df['EMA_20'].replace(0, 1e-6)
    df['alpha4_ratio'] = (df['close'] - df['EMA_20']) / df['EMA_20_safe']
    df['alpha4'] = df['vol_spike'] * df['alpha4_ratio'] + 0.3 * df['sent_divergence']

    # alpha5: -np.abs((bb_width - bb_width.mean()) / bb_width.std().replace(0, 1e-6)) * ret_5d + 0.7 * (vcb_sentiment - avg_related_sent)
    # Sử dụng BB_Range = BB_Upper - BB_Lower làm BB_Width
    df['BB_Range_Mean'] = df['BB_Range'].rolling(window=20).mean().fillna(0) # Dùng 20 ngày cho tính trung bình
    df['BB_Range_Std'] = df['BB_Range'].rolling(window=20).std().replace(0, 1e-6).fillna(1e-6) # Tránh std=0

    df['alpha5_ratio'] = (df['BB_Range'] - df['BB_Range_Mean']) / df['BB_Range_Std']
    df['alpha5'] = -np.abs(df['alpha5_ratio']) * df['ret_5d'] + 0.7 * df['sent_divergence']

    # Xóa các cột tạm
    cols_to_drop = [
        'avg_related_sent', 'sent_divergence', 'ret_5d', 'vol_spike',
        'MACD_Hist', 'BB_Range', 'close_safe', 'EMA_20_safe',
        'alpha4_ratio', 'BB_Range_Mean', 'BB_Range_Std', 'alpha5_ratio'
    ]
    df = df.drop(cols_to_drop, axis=1, errors='ignore')

    # Thay thế các giá trị vô cùng và NaN bằng 0 (do các phép tính log/chia có thể tạo ra)
    df = df.replace([np.inf, -np.inf], 0)
    df = df.fillna(0) # Fillna sau khi tính toán xong

    print("Đã tính toán xong 5 alpha features mới cho VCB.")
    return df

def prepare_data_with_sentiment(df):
    """Chuẩn bị dữ liệu với các chỉ báo kỹ thuật, sentiment scores và alpha features."""
    df = df.copy()

    # 1. Tính toán các alpha features
    df = calculate_alpha_features(df)

    # 2. Thêm cột Pct_Change (Mục tiêu dự đoán)
    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # Xóa các hàng có NaN (chủ yếu là do cửa sổ T-50 ngày của alpha/TIs)
    # Cần lọc bỏ các hàng có TI và Alpha là 0 do thiếu dữ liệu lịch sử ban đầu (vd: 50 ngày đầu tiên)
    cols_check_nan = ['close', 'SMA_5', 'RSI_14', 'SMA_50', 'EMA_20', 'alpha1', 'alpha5']
    # Lọc các hàng mà giá trị trong các cột quan trọng không phải 0 (tức là đã tính được)
    # Điều này hiệu quả hơn dropna vì ta đã điền 0 cho NaN
    df_cleaned = df[~((df[cols_check_nan] == 0).any(axis=1))].reset_index(drop=True)

    print(f"Sau khi làm sạch: {len(df_cleaned)} dòng dữ liệu")
    return df_cleaned

# --- Giữ nguyên các hàm còn lại (Model Definition, Train, Predict, Plot) ---
# Tối ưu lại phần Main Execution để sử dụng các chỉ báo mới

class StockDataset(Dataset):
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    """Tạo cửa sổ trượt."""
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0).transpose(0, 1)
        self.register_buffer('pe', pe)
    def forward(self, x):
        return x + self.pe[:x.size(0), :]

class TemporalEmbedding(nn.Module):
    def __init__(self, window_size, d_model):
        super(TemporalEmbedding, self).__init__()
        self.time_embedding = nn.Embedding(window_size, d_model)
        self.register_buffer('positions', torch.arange(window_size).long())
    def forward(self, x):
        batch_size, seq_len, _ = x.shape
        pos = self.positions.unsqueeze(0).repeat(batch_size, 1)
        time_emb = self.time_embedding(pos)
        return time_emb

class StockEncoderDecoderTransformer(nn.Module):
    def __init__(self, input_dim_src, input_dim_tgt, d_model=64, nhead=4, num_layers=3, dropout=0.1, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model

        # --- Encoder Components ---
        self.conv_emb_src = nn.Conv1d(input_dim_src, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(window_size, d_model)
        self.pos_encoder = PositionalEncoding(d_model)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components ---
        self.conv_emb_tgt = nn.Conv1d(input_dim_tgt, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(window_size, d_model)
        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 64), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def _get_embedding(self, x, conv_emb_layer, temporal_emb_layer):
        x_conv = x.transpose(1, 2)
        x_conv = conv_emb_layer(x_conv)
        x_conv = x_conv.transpose(1, 2)
        x_temporal = temporal_emb_layer(x)
        x_emb = x_conv * np.sqrt(self.d_model) + x_temporal
        x_final = self.pos_encoder(x_emb.transpose(0, 1)).transpose(0, 1)
        return x_final

    def forward(self, x_src, x_tgt_in):
        src_emb = self._get_embedding(x_src, self.conv_emb_src, self.temporal_emb_src)
        memory = self.transformer_encoder(src_emb)
        tgt_emb = self._get_embedding(x_tgt_in, self.conv_emb_tgt, self.temporal_emb_tgt)
        output = self.transformer_decoder(tgt_emb, memory)
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        train_losses.append(train_loss)
        val_losses.append(val_loss)

        if (epoch + 1) % 10 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    """Dự đoán giá đóng cửa của ngày tiếp theo."""

    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    # Lấy cửa sổ dữ liệu cuối cùng
    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    # Chuẩn bị input cho Encoder (Source - TIs + Sentiment + Alpha)
    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    # Chuẩn bị input cho Decoder (Target Input - Close Price)
    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    # Dự đoán Pct_Change đã được scale
    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    # Đảo ngược scale để có Pct_Change thực
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]

    # Tính giá trị tuyệt đối dự đoán
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price

def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    """Vẽ biểu đồ so sánh giá thực tế và giá dự đoán trên tập Test."""
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 8))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2, alpha=0.8)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    plt.title(f'SO SÁNH DỰ ĐOÁN VÀ THỰC TẾ CỔ PHIẾU {symbol} (5 ALPHA MỚI VCB + SENTIMENT LIÊN QUAN)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16, fontweight='bold')
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend(fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# =====================================
# PHẦN 5. MAIN EXECUTION
# =====================================
def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER ===")
    print("=== VỚI 5 ALPHA FEATURES MỚI KẾT HỢP TECHNICAL INDICATORS VÀ SENTIMENT CỦA VCB VÀ CÁC MÃ LIÊN QUAN ===")

    print("\n1. Đang tải dữ liệu từ file CSV...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        print(f"Không có dữ liệu hợp lệ cho mã {symbol}.")
        return None

    df_processed = prepare_data_with_sentiment(df)

    # Lấy danh sách cột sentiment score - VCB và TẤT CẢ các mã liên quan
    sentiment_columns = [f'{c}_daily_new_sentiment_score' for c in ALL_SENTIMENT_COMPANIES if f'{c}_daily_new_sentiment_score' in df_processed.columns]

    # Định nghĩa các cột features - ĐÃ CẬP NHẬT SMA_50 VÀ EMA_20
    source_columns = [
        # Chỉ báo kỹ thuật cơ bản
        'open', 'high', 'low', 'close', 'volume',
        'SMA_5', 'SMA_10', 'SMA_20', 'SMA_50', 'EMA_10', 'EMA_20', # <--- ĐÃ THÊM SMA_50 VÀ EMA_20
        'Momentum_3', 'Momentum_10',
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV',
    ] + sentiment_columns + [
        # Alpha features mới
        'alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5'
    ]

    target_input_columns = ['close']
    target_output_column = 'Pct_Change'

    # Kiểm tra cột có sẵn trong file và lọc lại source_columns
    source_columns = [col for col in source_columns if col in df_processed.columns]

    missing_cols_in_df = [col for col in source_columns if col not in df_processed.columns]
    if missing_cols_in_df:
         if any(a in missing_cols_in_df for a in ['alpha1', 'alpha5', 'SMA_50', 'EMA_20']):
             print(f"\nLỖI QUAN TRỌNG: Thiếu các cột dữ liệu cần thiết (Alpha/Sentiment/TIs): {', '.join(missing_cols_in_df)}")
             return None

    if not source_columns:
        print("KHÔNG ĐỦ DỮ LIỆU ĐỂ TIẾP TỤC.")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"Số lượng features sử dụng (TIs + Sentiments + 5 Alpha): {len(source_columns)}")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    print("\n2. Đang chia dữ liệu Train/Test theo ngày...")
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    # Lọc DataFrame
    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    # Dữ liệu test là từ 1/1/2025 đến mới nhất
    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    if df_train.empty or len(df_train) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu training không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None
    if df_test.empty or len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")

    # --- TẠO CỬA SỔ TRƯỢT RIÊNG BIỆT VÀ CHUẨN HÓA DỮ LIỆU ---
    print("\n3. Đang tạo cửa sổ trượt và chuẩn hóa dữ liệu...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    # Scaler 1: Cho dữ liệu nguồn (TIs + Sentiment + Alpha features)
    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    # Scaler 2: Cho dữ liệu đầu vào của decoder (Close Price)
    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    # Scaler 3: Cho dữ liệu đầu ra (Pct_Change)
    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n4. Đang khởi tạo và huấn luyện mô hình...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns),
        input_dim_tgt=len(target_input_columns),
        d_model=64,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    # ... (Phần đánh giá giữ nguyên) ...
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()
    pct_change_actual_unscaled = scaler_tgt_output.inverse_transform(y_test.reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    close_price_T = df_test['close'].iloc[WINDOW_SIZE:].values
    min_len = min(len(close_price_T), len(pct_change_pred_unscaled))
    close_price_T = close_price_T[:min_len]
    pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
    pct_change_actual_unscaled = pct_change_actual_unscaled[:min_len]
    test_dates = test_dates[:min_len]

    predictions_unscaled = close_price_T * (1 + pct_change_pred_unscaled / 100)
    actuals_unscaled = close_price_T * (1 + pct_change_actual_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*60)
    print(f"KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST ({symbol}: 5 ALPHA MỚI VCB + SENTIMENT CỦA VCB & CÁC MÃ LIÊN QUAN)")
    print("="*60)
    print(f"Mã: {symbol}")
    print(f"Tổng số features: {len(source_columns)}")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*60)

    # Trả về các kết quả cần thiết để sử dụng bên ngoài
    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# THỰC THI CHÍNH
# =====================================
if __name__ == '__main__':
    try:
        # Chạy quy trình huấn luyện và đánh giá
        training_results = main(FILE_PATH, SYMBOL)

        if training_results:
            # Vẽ biểu đồ kết quả trên tập test
            print("\n5. Đang vẽ biểu đồ kết quả...")
            plot_predictions_vs_actuals(
                test_dates=training_results["test_dates"],
                predictions=training_results["predictions_unscaled"],
                actuals=training_results["actuals_unscaled"],
                symbol=SYMBOL,
                mape=training_results["metrics"]["mape"]
            )

            # Dự đoán cho ngày tiếp theo sau ngày cuối cùng trong dữ liệu
            print("\n6. Dự đoán giá cho ngày giao dịch tiếp theo...")
            next_day_price = predict_next_day(
                model=training_results["model"],
                scaler_src=training_results["scalers"]["src"],
                scaler_tgt_input=training_results["scalers"]["tgt_input"],
                scaler_tgt_output=training_results["scalers"]["tgt_output"],
                recent_data=training_results["df_processed"],
                src_cols=training_results["columns"]["source"],
                tgt_in_cols=training_results["columns"]["target_input"]
            )

            if next_day_price is not None:
                last_date = training_results["df_processed"]['date'].iloc[-1].date()
                last_price = training_results['df_processed']['close'].iloc[-1]
                price_change = next_day_price - last_price
                change_percent = (price_change / last_price) * 100

                print("\n" + "="*50)
                print("DỰ ĐOÁN GIÁ TIẾP THEO")
                print("="*50)
                print(f"Mã cổ phiếu: {SYMBOL}")
                print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
                print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
                print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
                print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
                print("="*50)
    except Exception as e:
        print(f"\nLỖI CHƯƠNG TRÌNH: Đã xảy ra lỗi không mong muốn trong quá trình thực thi: {e}")
        traceback.print_exc()
```

```python
import pandas as pd
import numpy as np
import os
from datetime import datetime
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import traceback

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB"
RELATED_COMPANIES = ['TCB', 'VJC', 'HPG', 'PLX', 'VIC', 'SSI', 'GAS', 'FPT']
ALL_SENTIMENT_COMPANIES = [TARGET_STOCK_CODE] + RELATED_COMPANIES
FILE_PATH = "/content/VCB_final.csv"
SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 32
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA (ĐÃ CẬP NHẬT TÍNH TOÁN TI VÀ ALPHA)
# =====================================

def calculate_missing_technical_indicators(df):
    """
    Tính toán các chỉ báo kỹ thuật còn thiếu để phục vụ tính toán Alpha.
    Chỉ báo cần thiếu: SMA_10, SMA_50, EMA_20 (đã có từ yêu cầu trước), và RSI cần thiết cho alpha2.
    """
    df = df.copy()

    print("-> Đang kiểm tra và tính toán các Chỉ báo Kỹ thuật còn thiếu...")

    # 1. SMA 10 ngày (Bổ sung để sửa lỗi KeyError)
    if 'SMA_10' not in df.columns:
        print("   - Đang tính toán SMA_10...")
        df['SMA_10'] = df['close'].rolling(window=10).mean()

    # 2. SMA 50 ngày (Từ yêu cầu trước)
    if 'SMA_50' not in df.columns:
        print("   - Đang tính toán SMA_50...")
        df['SMA_50'] = df['close'].rolling(window=50).mean()

    # 3. EMA 20 ngày (Từ yêu cầu trước)
    if 'EMA_20' not in df.columns:
        print("   - Đang tính toán EMA_20...")
        df['EMA_20'] = df['close'].ewm(span=20, adjust=False).mean()

    # 4. Standard Deviation (std) cho cửa sổ 20 ngày (Cần cho alpha1 và alpha5)
    if 'Std_20' not in df.columns:
        print("   - Đang tính toán Std_20...")
        df['Std_20'] = df['close'].rolling(window=20).std()

    # 5. MACD Histogram (Cần cho alpha3 và alpha5)
    if 'MACD_Hist' not in df.columns and 'MACD' in df.columns and 'MACD_Signal' in df.columns:
        print("   - Đang tính toán MACD_Hist...")
        df['MACD_Hist'] = df['MACD'] - df['MACD_Signal']

    # 6. RSI (Cần cho alpha2, mặc dù đã có RSI_14 nhưng đề phòng)
    if 'RSI_14' not in df.columns:
         # Đây là công thức đầy đủ nếu RSI_14 thiếu
         print("   - CẢNH BÁO: RSI_14 bị thiếu. Các Alpha sử dụng RSI sẽ có lỗi nếu không tính lại.")
         # Việc tính lại RSI phức tạp, ta giả định nó có hoặc sẽ lỗi nếu thiếu
         pass

    # Tạm thời điền 0 cho các giá trị NaN do cửa sổ lăn ban đầu
    df = df.fillna(0)

    return df

def calculate_sentiment_derivatives(df):
    """
    Tính toán các chỉ báo sentiment phái sinh cần thiết cho 5 Alpha mới.
    Các chỉ báo cần thiết: Sentiment_7d_Divergence, Sentiment_7d_Momentum.
    """
    df = df.copy()
    vcb_sent_col = f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'
    tcb_sent_col = f'TCB_daily_new_sentiment_score'

    # Kiểm tra cột sentiment VCB
    if vcb_sent_col not in df.columns:
        print(f"LỖI: Thiếu cột sentiment VCB: {vcb_sent_col}. Các Alpha có sentiment sẽ bị set về 0.")
        df[vcb_sent_col] = 0.0

    # Kiểm tra cột sentiment TCB (cần cho alpha4)
    if tcb_sent_col not in df.columns:
        print(f"CẢNH BÁO: Thiếu cột sentiment TCB: {tcb_sent_col}. Alpha4 sẽ bị set về 0.")
        df[tcb_sent_col] = 0.0

    # 1. Sentiment_7d_Divergence (Cần cho alpha2)
    # Giả định Divergence là VCB sentiment hiện tại so với TCB 7 ngày trước (ví dụ)
    # Hoặc VCB 7d AVG - TCB 7d AVG. Dựa vào tên, ta dùng VCB 7d AVG - TCB 7d AVG cho tính tổng quát

    # Tính trung bình 7 ngày của sentiment VCB
    df['VCB_Sentiment_7d_Avg'] = df[vcb_sent_col].rolling(window=7).mean().fillna(0)
    # Tính trung bình 7 ngày của sentiment TCB
    df['TCB_Sentiment_7d_Avg'] = df[tcb_sent_col].rolling(window=7).mean().fillna(0)

    # Sentiment_7d_Divergence: Chênh lệch trung bình 7 ngày giữa VCB và TCB
    if 'Sentiment_7d_Divergence' not in df.columns:
        print("-> Đang tính toán Sentiment_7d_Divergence...")
        df['Sentiment_7d_Divergence'] = df['VCB_Sentiment_7d_Avg'] - df['TCB_Sentiment_7d_Avg']

    # 2. Sentiment_7d_Momentum (Cần cho alpha5)
    # Giả định là Momentum của Sentiment VCB trong 7 ngày
    if 'Sentiment_7d_Momentum' not in df.columns:
        print("-> Đang tính toán Sentiment_7d_Momentum...")
        # Sử dụng VCB Sentiment hiện tại / VCB Sentiment 7 ngày trước (hoặc đơn giản là trung bình 7 ngày)
        df['Sentiment_7d_Momentum'] = df['VCB_Sentiment_7d_Avg']

    df = df.fillna(0)
    return df

def load_stock_data_from_df(file_path, symbol):
    """Tải dữ liệu từ file CSV, đổi tên cột và tính toán TIs còn thiếu."""
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        # Đọc file CSV với encoding phù hợp
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    # Đổi tên cột 'Thời gian' thành 'date'
    df.rename(columns={'Thời gian': 'date'}, inplace=True)

    # Cố gắng chuyển đổi ngày tháng
    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    # Sắp xếp theo ngày
    df = df.sort_values('date').reset_index(drop=True)

    # Lọc dữ liệu từ 2024-01-01 trở đi
    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    # Thêm bước tính TIs còn thiếu
    df = calculate_missing_technical_indicators(df)

    print(f"Đã tải {len(df)} dòng dữ liệu")
    print(f"Ngày bắt đầu: {df['date'].min().date()}")
    print(f"Ngày kết thúc: {df['date'].max().date()}")

    return df

def calculate_alpha_features(df):
    """
    Tính toán 5 trường alpha mới cho VCB dựa trên công thức mới được cung cấp.
    """
    df = df.copy()

    # 1. Tính các chỉ báo Sentiment phái sinh
    df = calculate_sentiment_derivatives(df)

    vcb_sent = f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'
    tcb_sent = f'TCB_daily_new_sentiment_score'

    # Đảm bảo các cột chia không bị chia cho 0
    df['close_safe'] = df['close'].replace(0, 1e-9)
    df['Std_20_safe'] = df['Std_20'].replace(0, 1e-9)
    df['BB_Range'] = df['BB_Upper'] - df['BB_Lower']

    # --- CÁC CÔNG THỨC ALPHA MỚI (ĐÃ CẬP NHẬT) ---

    # alpha1: ((close.shift(1) / close.shift(5)) - 1) / (std(close.rolling(20)) + 1e-9) * (1 + 0.8 * VCB_daily_new_sentiment_score.shift(1))
    df['alpha1_numerator'] = (df['close_safe'].shift(1) / df['close_safe'].shift(5)) - 1
    df['alpha1_momentum_ratio'] = df['alpha1_numerator'] / df['Std_20_safe']
    df['alpha1_sentiment_term'] = 1 + 0.8 * df[vcb_sent].shift(1)
    df['alpha1'] = df['alpha1_momentum_ratio'] * df['alpha1_sentiment_term']

    # alpha2: -((close / close.shift(1)) - 1) * (1 - ((BB_Upper - BB_Lower) / close).rolling(10).mean()) * (1 + 0.5 * Sentiment_7d_Divergence)
    df['alpha2_return'] = (df['close_safe'] / df['close_safe'].shift(1)) - 1
    df['alpha2_bb_ratio'] = (df['BB_Range'] / df['close_safe']).rolling(window=10).mean().fillna(0)
    df['alpha2_bb_term'] = 1 - df['alpha2_bb_ratio']
    df['alpha2_sentiment_term'] = 1 + 0.5 * df['Sentiment_7d_Divergence']
    df['alpha2'] = -df['alpha2_return'] * df['alpha2_bb_term'] * df['alpha2_sentiment_term']

    # alpha3: ((volume / volume.rolling(20).mean()) > 1.5).astype(int) * ((MACD - MACD_Signal) + 0.5 * (VCB_daily_new_sentiment_score - VCB_daily_new_sentiment_score.shift(3))) * (close > BB_Upper).astype(int)
    df['volume_20d_mean'] = df['volume'].rolling(window=20).mean().replace(0, 1e-9)
    df['alpha3_vol_spike'] = (df['volume'] / df['volume_20d_mean'] > 1.5).astype(int)
    df['alpha3_macd_sent_term'] = df['MACD_Hist'] + 0.5 * (df[vcb_sent] - df[vcb_sent].shift(3))
    df['alpha3_breakout'] = (df['close'] > df['BB_Upper']).astype(int)
    df['alpha3'] = df['alpha3_vol_spike'] * df['alpha3_macd_sent_term'] * df['alpha3_breakout']

    # alpha4: ((VCB_daily_new_sentiment_score - TCB_daily_new_sentiment_score) * ((close / close.shift(3)) - 1) / (1 + abs((close / close.shift(20)) - 1)))
    df['alpha4_sent_spread'] = df[vcb_sent] - df[tcb_sent]
    df['alpha4_return_3d'] = (df['close_safe'] / df['close_safe'].shift(3)) - 1
    df['alpha4_return_20d'] = (df['close_safe'] / df['close_safe'].shift(20)) - 1
    df['alpha4_denominator'] = 1 + np.abs(df['alpha4_return_20d'])
    df['alpha4'] = (df['alpha4_sent_spread'] * df['alpha4_return_3d']) / df['alpha4_denominator'].replace(0, 1e-9)

    # alpha5: (0.4 * (((close / close.shift(5)) - 1) / (std(close.rolling(20)) + 1e-9)) + 0.3 * ((MACD - MACD_Signal) / (abs(MACD_Signal) + 1e-9)) + 0.3 * ((Sentiment_7d_Momentum - Sentiment_7d_Momentum.shift(1)) / (abs(Sentiment_7d_Momentum.shift(1)) + 1e-9)))

    # Term 1: Risk-Adjusted Momentum (Same as alpha1's ratio)
    df['alpha5_term1'] = df['alpha1_momentum_ratio'] # Tái sử dụng

    # Term 2: MACD Histogram / abs(MACD_Signal)
    df['MACD_Signal_safe'] = df['MACD_Signal'].replace(0, 1e-9)
    df['alpha5_term2'] = df['MACD_Hist'] / np.abs(df['MACD_Signal_safe'])

    # Term 3: Sentiment Momentum Change / abs(Sentiment Momentum Shift)
    df['Sent_Mom_Shift_safe'] = df['Sentiment_7d_Momentum'].shift(1).replace(0, 1e-9)
    df['alpha5_term3_numerator'] = df['Sentiment_7d_Momentum'] - df['Sentiment_7d_Momentum'].shift(1)
    df['alpha5_term3'] = df['alpha5_term3_numerator'] / np.abs(df['Sent_Mom_Shift_safe'])

    df['alpha5'] = 0.4 * df['alpha5_term1'] + 0.3 * df['alpha5_term2'] + 0.3 * df['alpha5_term3']

    # Xóa các cột tạm và xử lý NaN/Inf
    cols_to_drop = [
        'close_safe', 'Std_20_safe', 'BB_Range',
        'VCB_Sentiment_7d_Avg', 'TCB_Sentiment_7d_Avg',
        'alpha1_numerator', 'alpha1_momentum_ratio', 'alpha1_sentiment_term',
        'alpha2_return', 'alpha2_bb_ratio', 'alpha2_bb_term', 'alpha2_sentiment_term',
        'volume_20d_mean', 'alpha3_vol_spike', 'alpha3_macd_sent_term', 'alpha3_breakout',
        'alpha4_sent_spread', 'alpha4_return_3d', 'alpha4_return_20d', 'alpha4_denominator',
        'MACD_Signal_safe', 'Sent_Mom_Shift_safe', 'alpha5_term3_numerator',
        'alpha5_term1', 'alpha5_term2', 'alpha5_term3'
    ]
    df = df.drop(cols_to_drop, axis=1, errors='ignore')

    # Thay thế các giá trị vô cùng và NaN bằng 0
    df = df.replace([np.inf, -np.inf], 0)
    df = df.fillna(0)

    print("Đã tính toán xong 5 alpha features mới cho VCB.")
    return df

def prepare_data_with_sentiment(df):
    """Chuẩn bị dữ liệu với các chỉ báo kỹ thuật, sentiment scores và alpha features."""
    df = df.copy()

    # 1. Tính toán các alpha features
    df = calculate_alpha_features(df)

    # 2. Thêm cột Pct_Change (Mục tiêu dự đoán)
    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # Xóa các hàng có NaN (chủ yếu là do cửa sổ T-20 ngày của alpha/TIs)
    # Cần lọc bỏ các hàng có TI và Alpha là 0 do thiếu dữ liệu lịch sử ban đầu (vd: 20 ngày đầu tiên)
    cols_check_nan = ['close', 'SMA_5', 'RSI_14', 'Std_20', 'alpha1', 'alpha5']
    # Lọc các hàng mà giá trị trong các cột quan trọng không phải 0 (tức là đã tính được)
    df_cleaned = df[~((df[cols_check_nan] == 0).any(axis=1))].reset_index(drop=True)

    print(f"Sau khi làm sạch: {len(df_cleaned)} dòng dữ liệu")
    return df_cleaned

# --- Các hàm khác (Model Definition, Train, Predict, Plot) giữ nguyên ---

class StockDataset(Dataset):
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    """Tạo cửa sổ trượt."""
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)

class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0).transpose(0, 1)
        self.register_buffer('pe', pe)
    def forward(self, x):
        return x + self.pe[:x.size(0), :]

class TemporalEmbedding(nn.Module):
    def __init__(self, window_size, d_model):
        super(TemporalEmbedding, self).__init__()
        self.time_embedding = nn.Embedding(window_size, d_model)
        self.register_buffer('positions', torch.arange(window_size).long())
    def forward(self, x):
        batch_size, seq_len, _ = x.shape
        pos = self.positions.unsqueeze(0).repeat(batch_size, 1)
        time_emb = self.time_embedding(pos)
        return time_emb

class StockEncoderDecoderTransformer(nn.Module):
    def __init__(self, input_dim_src, input_dim_tgt, d_model=64, nhead=4, num_layers=3, dropout=0.1, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model

        # --- Encoder Components ---
        self.conv_emb_src = nn.Conv1d(input_dim_src, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(window_size, d_model)
        self.pos_encoder = PositionalEncoding(d_model)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components ---
        self.conv_emb_tgt = nn.Conv1d(input_dim_tgt, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(window_size, d_model)
        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=256, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 64), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(64, 1)
        )

    def _get_embedding(self, x, conv_emb_layer, temporal_emb_layer):
        x_conv = x.transpose(1, 2)
        x_conv = conv_emb_layer(x_conv)
        x_conv = x_conv.transpose(1, 2)
        x_temporal = temporal_emb_layer(x)
        x_emb = x_conv * np.sqrt(self.d_model) + x_temporal
        x_final = self.pos_encoder(x_emb.transpose(0, 1)).transpose(0, 1)
        return x_final

    def forward(self, x_src, x_tgt_in):
        src_emb = self._get_embedding(x_src, self.conv_emb_src, self.temporal_emb_src)
        memory = self.transformer_encoder(src_emb)
        tgt_emb = self._get_embedding(x_tgt_in, self.conv_emb_tgt, self.temporal_emb_tgt)
        output = self.transformer_decoder(tgt_emb, memory)
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        train_losses.append(train_loss)
        val_losses.append(val_loss)

        if (epoch + 1) % 10 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    """Dự đoán giá đóng cửa của ngày tiếp theo."""

    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    # Lấy cửa sổ dữ liệu cuối cùng
    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    # Chuẩn bị input cho Encoder (Source - TIs + Sentiment + Alpha)
    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    # Chuẩn bị input cho Decoder (Target Input - Close Price)
    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    # Dự đoán Pct_Change đã được scale
    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    # Đảo ngược scale để có Pct_Change thực
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]

    # Tính giá trị tuyệt đối dự đoán
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price

def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    """Vẽ biểu đồ so sánh giá thực tế và giá dự đoán trên tập Test."""
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 8))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2, alpha=0.8)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    plt.title(f'SO SÁNH DỰ ĐOÁN VÀ THỰC TẾ CỔ PHIẾU {symbol} (5 ALPHA MỚI VCB + SENTIMENT LIÊN QUAN)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16, fontweight='bold')
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend(fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# =====================================
# PHẦN 5. MAIN EXECUTION
# =====================================
def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER ===")
    print("=== VỚI 5 ALPHA FEATURES MỚI KẾT HỢP TECHNICAL INDICATORS VÀ SENTIMENT CỦA VCB VÀ CÁC MÃ LIÊN QUAN ===")

    print("\n1. Đang tải dữ liệu từ file CSV...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        print(f"Không có dữ liệu hợp lệ cho mã {symbol}.")
        return None

    df_processed = prepare_data_with_sentiment(df)

    # Lấy danh sách cột sentiment score - VCB và TẤT CẢ các mã liên quan
    sentiment_columns = [f'{c}_daily_new_sentiment_score' for c in ALL_SENTIMENT_COMPANIES if f'{c}_daily_new_sentiment_score' in df_processed.columns]

    # Định nghĩa các cột features - ĐÃ CẬP NHẬT TẤT CẢ CÁC CỘT CẦN THIẾT
    source_columns = [
        # Chỉ báo kỹ thuật cơ bản
        'open', 'high', 'low', 'close', 'volume',
        'SMA_5', 'SMA_10', 'SMA_20', 'SMA_50', 'EMA_10', 'EMA_20', # TI bổ sung
        'Momentum_3', 'Momentum_10', 'Std_20', # TI bổ sung
        'RSI_14', 'MACD', 'MACD_Signal', 'BB_Upper', 'BB_Lower', 'OBV', 'MACD_Hist' # TI bổ sung
    ] + sentiment_columns + [
        # Alpha features mới
        'alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5',
        # Sentiment phái sinh
        'Sentiment_7d_Divergence', 'Sentiment_7d_Momentum' # Sentiment bổ sung
    ]

    target_input_columns = ['close']
    target_output_column = 'Pct_Change'

    # Kiểm tra cột có sẵn trong file và lọc lại source_columns
    source_columns = [col for col in source_columns if col in df_processed.columns]

    missing_cols_in_df = [col for col in source_columns if col not in df_processed.columns and col not in ['Sentiment_7d_Divergence', 'Sentiment_7d_Momentum']]
    if missing_cols_in_df:
         if any(a in missing_cols_in_df for a in ['alpha1', 'alpha5']):
             print(f"\nLỖI QUAN TRỌNG: Thiếu các cột dữ liệu cần thiết (Alpha/Sentiment/TIs): {', '.join(missing_cols_in_df)}")
             return None

    if not source_columns:
        print("KHÔNG ĐỦ DỮ LIỆU ĐỂ TIẾP TỤC.")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"Số lượng features sử dụng (TIs + Sentiments + 5 Alpha + Phái sinh): {len(source_columns)}")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    print("\n2. Đang chia dữ liệu Train/Test theo ngày...")
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    # Lọc DataFrame
    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    # Dữ liệu test là từ 1/1/2025 đến mới nhất
    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    if df_train.empty or len(df_train) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu training không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None
    if df_test.empty or len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")

    # --- TẠO CỬA SỔ TRƯỢT RIÊNG BIỆT VÀ CHUẨN HÓA DỮ LIỆU ---
    print("\n3. Đang tạo cửa sổ trượt và chuẩn hóa dữ liệu...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    # Scaler 1: Cho dữ liệu nguồn (TIs + Sentiment + Alpha features)
    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    # Scaler 2: Cho dữ liệu đầu vào của decoder (Close Price)
    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    # Scaler 3: Cho dữ liệu đầu ra (Pct_Change)
    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n4. Đang khởi tạo và huấn luyện mô hình...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns),
        input_dim_tgt=len(target_input_columns),
        d_model=64,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()
    pct_change_actual_unscaled = scaler_tgt_output.inverse_transform(y_test.reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    close_price_T = df_test['close'].iloc[WINDOW_SIZE:].values
    min_len = min(len(close_price_T), len(pct_change_pred_unscaled))
    close_price_T = close_price_T[:min_len]
    pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
    pct_change_actual_unscaled = pct_change_actual_unscaled[:min_len]
    test_dates = test_dates[:min_len]

    predictions_unscaled = close_price_T * (1 + pct_change_pred_unscaled / 100)
    actuals_unscaled = close_price_T * (1 + pct_change_actual_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*60)
    print(f"KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST ({symbol}: 5 ALPHA MỚI VCB + SENTIMENT CỦA VCB & CÁC MÃ LIÊN QUAN)")
    print("="*60)
    print(f"Mã: {symbol}")
    print(f"Tổng số features: {len(source_columns)}")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*60)

    # Trả về các kết quả cần thiết để sử dụng bên ngoài
    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# THỰC THI CHÍNH
# =====================================
if __name__ == '__main__':
    try:
        # Chạy quy trình huấn luyện và đánh giá
        training_results = main(FILE_PATH, SYMBOL)

        if training_results:
            # Vẽ biểu đồ kết quả trên tập test
            print("\n5. Đang vẽ biểu đồ kết quả...")
            plot_predictions_vs_actuals(
                test_dates=training_results["test_dates"],
                predictions=training_results["predictions_unscaled"],
                actuals=training_results["actuals_unscaled"],
                symbol=SYMBOL,
                mape=training_results["metrics"]["mape"]
            )

            # Dự đoán cho ngày tiếp theo sau ngày cuối cùng trong dữ liệu
            print("\n6. Dự đoán giá cho ngày giao dịch tiếp theo...")
            next_day_price = predict_next_day(
                model=training_results["model"],
                scaler_src=training_results["scalers"]["src"],
                scaler_tgt_input=training_results["scalers"]["tgt_input"],
                scaler_tgt_output=training_results["scalers"]["tgt_output"],
                recent_data=training_results["df_processed"],
                src_cols=training_results["columns"]["source"],
                tgt_in_cols=training_results["columns"]["target_input"]
            )

            if next_day_price is not None:
                last_date = training_results["df_processed"]['date'].iloc[-1].date()
                last_price = training_results['df_processed']['close'].iloc[-1]
                price_change = next_day_price - last_price
                change_percent = (price_change / last_price) * 100

                print("\n" + "="*50)
                print("DỰ ĐOÁN GIÁ TIẾP THEO")
                print("="*50)
                print(f"Mã cổ phiếu: {SYMBOL}")
                print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
                print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
                print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
                print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
                print("="*50)
    except Exception as e:
        print(f"\nLỖI CHƯƠNG TRÌNH: Đã xảy ra lỗi không mong muốn trong quá trình thực thi: {e}")
        traceback.print_exc()
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os
import traceback

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ VÀ HẰNG SỐ (CẬP NHẬT: VCB)
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB" # <--- CẬP NHẬT: VCB
# Các mã sentiment cần thiết: VCB, TCB
ADDITIONAL_SENTIMENT_COMPANIES = ['TCB'] # <--- CẬP NHẬT: Chỉ cần TCB (VCB đã có)

# Đường dẫn file đã được sửa
FILE_PATH = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"

SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 64
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA: TÍNH TIs, ALPHA VÀ TEMPORAL FEATURES
# =====================================

def z_score(series, window):
    """Tính toán Z-score (chuẩn hóa) của một chuỗi trong cửa sổ lăn."""
    return (series - series.rolling(window=window).mean()) / series.rolling(window=window).std().replace(0, 1e-6)

def calculate_missing_technical_indicators(df):
    """
    Tính toán các chỉ báo kỹ thuật còn thiếu, bao gồm các chỉ báo cần thiết
    cho 5 Alpha mới (BB, Std, MACD).
    """
    df = df.copy()
    window_bb = 20
    window_macd_fast = 12
    window_macd_slow = 26
    window_macd_signal = 9

    # 1. BB (BB_Upper, BB_Lower, BB_Range) & STD 20
    df['SMA_20'] = df['close'].rolling(window=window_bb).mean()
    df['STD_20'] = df['close'].rolling(window=window_bb).std().replace(0, 1e-9)
    df['BB_Upper'] = df['SMA_20'] + 2 * df['STD_20']
    df['BB_Lower'] = df['SMA_20'] - 2 * df['STD_20']
    df['BB_Range'] = df['BB_Upper'] - df['BB_Lower']

    # 2. MACD
    ema_fast = df['close'].ewm(span=window_macd_fast, adjust=False).mean()
    ema_slow = df['close'].ewm(span=window_macd_slow, adjust=False).mean()
    df['MACD'] = ema_fast - ema_slow
    df['MACD_Signal'] = df['MACD'].ewm(span=window_macd_signal, adjust=False).mean()

    # Thêm các cột shift/rolling cần thiết
    df['close_shift_1'] = df['close'].shift(1)
    df['close_shift_5'] = df['close'].shift(5)
    df['close_shift_3'] = df['close'].shift(3)
    df['close_shift_20'] = df['close'].shift(20)
    df['volume_rolling_20_mean'] = df['volume'].rolling(window=20).mean()

    df = df.fillna(0)

    return df

def load_stock_data_from_df(file_path, symbol):
    # Giữ nguyên logic tải và xử lý ngày
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    df.rename(columns={'Thời gian': 'date'}, inplace=True)

    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    df = df.sort_values('date').reset_index(drop=True)
    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    print("\n-> Đang kiểm tra và tính toán các Chỉ báo Kỹ thuật còn thiếu...")
    df = calculate_missing_technical_indicators(df)

    print(f"Đã tải {len(df)} dòng dữ liệu")
    return df

def calculate_alpha_features(df):
    """
    Tính toán 5 trường alpha mới cho VCB dựa trên công thức yêu cầu.
    """
    df = df.copy()
    sentiment_vcb = f'{TARGET_STOCK_CODE}_daily_new_sentiment_score' # VCB_daily_new_sentiment_score
    sentiment_tcb = 'TCB_daily_new_sentiment_score'

    # Tính các Sentiment Tùy chỉnh (Cần cho Alpha 2 và Alpha 5)
    # 7-day Sentiment Divergence (Cần cho Alpha 2)
    df['Sentiment_7d_Divergence'] = df[sentiment_vcb] - df[sentiment_vcb].rolling(window=7).mean().fillna(0)
    # 7-day Sentiment Momentum (Cần cho Alpha 5)
    df['Sentiment_7d_Momentum'] = df[sentiment_vcb] - df[sentiment_vcb].shift(7).fillna(0)

    # --- CÁC CÔNG THỨC ALPHA MỚI ---

    # Alpha 1: ((close.shift(1) / close.shift(5)) - 1) / (std(close.rolling(20)) + 1e-9) * (1 + 0.8 * VCB_daily_new_sentiment_score.shift(1))
    df['Alpha1_Price_Momentum'] = (df['close_shift_1'] / df['close_shift_5']) - 1
    df['Alpha1_Sent_Term'] = 1 + 0.8 * df[sentiment_vcb].shift(1).fillna(0)

    df['alpha1'] = (df['Alpha1_Price_Momentum'] / (df['STD_20'] + 1e-9)) * \
                   df['Alpha1_Sent_Term']

    # Alpha 2: -((close / close.shift(1)) - 1) * (1 - ((BB_Upper - BB_Lower) / close).rolling(10).mean()) * (1 + 0.5 * Sentiment_7d_Divergence)
    df['Alpha2_Price_Reversal'] = -((df['close'] / df['close'].shift(1)) - 1)
    df['Alpha2_Volatility_Term'] = (df['BB_Range'] / df['close']).rolling(window=10).mean().fillna(0)
    df['Alpha2_Sent_Term'] = 1 + 0.5 * df['Sentiment_7d_Divergence']

    df['alpha2'] = df['Alpha2_Price_Reversal'] * \
                   (1 - df['Alpha2_Volatility_Term']) * \
                   df['Alpha2_Sent_Term']

    # Alpha 3: ((volume / volume.rolling(20).mean()) > 1.5).astype(int) * ((MACD - MACD_Signal) + 0.5 * (VCB_daily_new_sentiment_score - VCB_daily_new_sentiment_score.shift(3))) * (close > BB_Upper).astype(int)
    df['Alpha3_Volume_Condition'] = (df['volume'] / df['volume_rolling_20_mean']).fillna(0) > 1.5
    df['Alpha3_Price_Condition'] = df['close'] > df['BB_Upper']
    df['Alpha3_Sent_Diff'] = df[sentiment_vcb] - df[sentiment_vcb].shift(3).fillna(0)
    df['Alpha3_MACD_Term'] = (df['MACD'] - df['MACD_Signal']) + 0.5 * df['Alpha3_Sent_Diff']

    df['alpha3'] = df['Alpha3_Volume_Condition'].astype(int) * \
                   df['Alpha3_MACD_Term'] * \
                   df['Alpha3_Price_Condition'].astype(int)

    # Alpha 4: ((VCB_daily_new_sentiment_score - TCB_daily_new_sentiment_score) * ((close / close.shift(3)) - 1) / (1 + abs((close / close.shift(20)) - 1)))
    df['Alpha4_Sent_Diff'] = df[sentiment_vcb] - df[sentiment_tcb]
    df['Alpha4_Price_Momentum_3d'] = (df['close'] / df['close_shift_3']) - 1
    df['Alpha4_Price_Momentum_20d_Abs'] = np.abs((df['close'] / df['close_shift_20']) - 1)

    df['alpha4'] = (df['Alpha4_Sent_Diff'] * df['Alpha4_Price_Momentum_3d']) / \
                   (1 + df['Alpha4_Price_Momentum_20d_Abs'])

    # Alpha 5: (0.4 * (((close / close.shift(5)) - 1) / (std(close.rolling(20)) + 1e-9)) + 0.3 * ((MACD - MACD_Signal) / (abs(MACD_Signal) + 1e-9)) + 0.3 * ((Sentiment_7d_Momentum - Sentiment_7d_Momentum.shift(1)) / (abs(Sentiment_7d_Momentum.shift(1)) + 1e-9)))"
    df['Alpha5_Term1'] = ((df['close'] / df['close_shift_5']) - 1) / (df['STD_20'] + 1e-9)
    df['Alpha5_Term2'] = (df['MACD'] - df['MACD_Signal']) / (np.abs(df['MACD_Signal']) + 1e-9)
    df['Alpha5_Term3_Div'] = (np.abs(df['Sentiment_7d_Momentum'].shift(1)) + 1e-9)
    df['Alpha5_Term3'] = (df['Sentiment_7d_Momentum'] - df['Sentiment_7d_Momentum'].shift(1)) / df['Alpha5_Term3_Div']

    df['alpha5'] = (0.4 * df['Alpha5_Term1']) + \
                   (0.3 * df['Alpha5_Term2']) + \
                   (0.3 * df['Alpha5_Term3'])

    # --- Dọn dẹp và làm sạch ---

    # Xóa các cột tạm và shift
    cols_to_drop = [col for col in df.columns if col.startswith('Alpha') or col.endswith('_Term') or col.endswith('_Diff') or col.endswith('_Condition') or col.endswith('_Momentum')]
    cols_to_drop.extend(['close_shift_1', 'close_shift_5', 'close_shift_3', 'close_shift_20', 'volume_rolling_20_mean', 'BB_Range', 'STD_20', 'SMA_20'])

    df = df.drop(cols_to_drop, axis=1, errors='ignore')

    # Thay thế các giá trị vô cùng và NaN bằng 0
    df = df.replace([np.inf, -np.inf], 0)
    df = df.fillna(0)

    print(f"Đã tính toán xong 5 alpha features mới cho {TARGET_STOCK_CODE}.")
    return df

def add_temporal_features(df):
    """
    Tạo các tính năng liên quan đến thời gian (Temporal Features):
    Day of Week (1-5), Day of Month (1-31), Month of Year (1-12)
    """
    df['day_of_week'] = df['date'].dt.dayofweek + 1 # 1=Thứ 2, 5=Thứ 6
    df['day_of_month'] = df['date'].dt.day
    df['month_of_year'] = df['date'].dt.month
    return df

def prepare_data_with_sentiment(df):
    df = df.copy()

    df = calculate_alpha_features(df)
    df = add_temporal_features(df)

    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # Loại bỏ các hàng mà các Alpha quan trọng không tính được
    check_cols = ['MACD', 'alpha1']

    initial_len = len(df)
    df_cleaned = df[~(df[check_cols] == 0).any(axis=1)].reset_index(drop=True)

    print(f"Sau khi làm sạch: {len(df_cleaned)} dòng dữ liệu (Đã loại bỏ {initial_len - len(df_cleaned)} dòng do thiếu dữ liệu lịch sử).")
    return df_cleaned

# ... (Giữ nguyên các hàm: create_sliding_windows, StockDataset) ...

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)

class StockDataset(Dataset):
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

# =====================================
# PHẦN 3. ĐỊNH NGHĨA MODEL (ĐÃ SỬA LỖI PositionalEncoding)
# =====================================
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0) # Shape: (1, max_len, d_model)
        self.register_buffer('pe', pe)

    def forward(self, x):
        # x shape: (batch_size, seq_len, d_model)
        # Sửa lỗi: Lấy pe theo seq_len (chiều 1 của x)
        x = x + self.pe[:, :x.size(1), :] # Dạng (batch, seq_len, d_model) + (1, seq_len, d_model)
        return x

class TemporalEmbedding(nn.Module):
    def __init__(self, input_temporal_dim, d_model):
        super(TemporalEmbedding, self).__init__()
        self.linear_temporal = nn.Linear(input_temporal_dim, d_model)

    def forward(self, x):
        return self.linear_temporal(x)

class StockEncoderDecoderTransformer(nn.Module):
    def __init__(self, input_dim_src, input_dim_tgt, d_model=128, nhead=8, num_layers=2, dropout=0.2, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model
        self.window_size = window_size

        self.temporal_dim = 3
        self.alpha_dim = 5
        self.price_dim = 1

        # --- Encoder Components (5 Alpha + Temporal) ---
        self.conv_emb_src_alpha = nn.Conv1d(self.alpha_dim, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(self.temporal_dim, d_model)

        self.pos_encoder = PositionalEncoding(d_model, max_len=window_size)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=512, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components (1 Close Price + Temporal) ---
        self.conv_emb_tgt_price = nn.Conv1d(self.price_dim, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(self.temporal_dim, d_model)

        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=512, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 256), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def _get_embedding(self, x_data, x_temporal, conv_emb_layer, temporal_emb_layer):
        # 1. Conv Embedding (Áp dụng cho Data Features: Alpha hoặc Price)
        x_data_conv = x_data.transpose(1, 2)
        x_conv = conv_emb_layer(x_data_conv)
        x_conv = x_conv.transpose(1, 2)

        # 2. Temporal Embedding (Áp dụng cho Temporal Features)
        x_temporal_emb = temporal_emb_layer(x_temporal)

        # 3. Kết hợp (Combine)
        x_emb = x_conv + x_temporal_emb

        # 4. Position Embedding (Thêm vào sau cùng)
        x_final = self.pos_encoder(x_emb)
        return x_final

    def forward(self, x_src, x_tgt_in):
        # Tách Alpha (5) và Temporal (3)
        x_src_alpha = x_src[:, :, :self.alpha_dim]
        x_src_temporal = x_src[:, :, self.alpha_dim:]

        # Tách Price (1) và Temporal (3)
        x_tgt_in_price = x_tgt_in[:, :, :self.price_dim]
        x_tgt_in_temporal = x_tgt_in[:, :, self.price_dim:]

        # Encoder Embedding
        src_emb = self._get_embedding(
            x_src_alpha, x_src_temporal,
            self.conv_emb_src_alpha, self.temporal_emb_src
        )
        memory = self.transformer_encoder(src_emb)

        # Decoder Embedding
        tgt_emb = self._get_embedding(
            x_tgt_in_price, x_tgt_in_temporal,
            self.conv_emb_tgt_price, self.temporal_emb_tgt
        )
        output = self.transformer_decoder(tgt_emb, memory)

        # Prediction Head
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

# ... (Giữ nguyên các hàm: train_model, predict_next_day) ...

def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        train_losses.append(train_loss)
        val_losses.append(val_loss)

        if (epoch + 1) % 5 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price


# =====================================
# PHẦN 5. MAIN EXECUTION
# =====================================
def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER (CẤU HÌNH MỚI) ===")
    print(f"=== Cấu hình: Window Size={WINDOW_SIZE}, Encoder Input: 5 Alpha + Temporal | Decoder Input: Close + Temporal ===")

    print("\n1. Đang tải và xử lý dữ liệu...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        return None

    df_processed = prepare_data_with_sentiment(df)

    temporal_columns = ['day_of_week', 'day_of_month', 'month_of_year']
    alpha_columns = ['alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5']

    source_columns = alpha_columns + temporal_columns
    target_input_columns = ['close'] + temporal_columns
    target_output_column = 'Pct_Change'

    source_columns = [col for col in source_columns if col in df_processed.columns]
    target_input_columns = [col for col in target_input_columns if col in df_processed.columns]

    if len(source_columns) != 8:
        print(f"\nLỖI: Thiếu cột Encoder Input. Chỉ tìm thấy {len(source_columns)}/8. Vui lòng kiểm tra file data.")
        return None
    if len(target_input_columns) != 4:
        print(f"\nLỖI: Thiếu cột Decoder Input. Chỉ tìm thấy {len(target_input_columns)}/4. Vui lòng kiểm tra file data.")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"SỐ LƯỢNG FEATURES ENCODER: {len(source_columns)} (5 Alpha + 3 Temporal)")
    print(f"SỐ LƯỢNG FEATURES DECODER INPUT: {len(target_input_columns)} (1 Close + 3 Temporal)")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    # ... (Kiểm tra dữ liệu Train/Test) ...
    if df_train.empty or len(df_train) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu training không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None
    if df_test.empty or len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")


    # --- TẠO CỬA SỔ TRƯỢT VÀ CHUẨN HÓA DỮ LIỆU ---
    print("\n3. Đang tạo cửa sổ trượt và chuẩn hóa dữ liệu...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n4. Đang khởi tạo và huấn luyện mô hình (d_model=128, nhead=8, num_layers=2, dropout=0.2)...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns),
        input_dim_tgt=len(target_input_columns),
        d_model=128,
        nhead=8,
        num_layers=2,
        dropout=0.2,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    print("\n5. Đang đánh giá trên tập Test...")
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    close_price_T_minus_1 = df_test['close'].iloc[WINDOW_SIZE-1:WINDOW_SIZE-1 + len(pct_change_pred_unscaled)].values
    actuals_unscaled = df_test['close'].iloc[WINDOW_SIZE:WINDOW_SIZE + len(pct_change_pred_unscaled)].values

    min_len = min(len(close_price_T_minus_1), len(pct_change_pred_unscaled))

    close_price_T_minus_1 = close_price_T_minus_1[:min_len]
    pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
    actuals_unscaled = actuals_unscaled[:min_len]
    test_dates = test_dates[:min_len]

    # Giá dự đoán ngày T = Giá T-1 * (1 + Pct_Change T / 100)
    predictions_unscaled = close_price_T_minus_1 * (1 + pct_change_pred_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*90)
    print(f"KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST ({symbol}: CẤU HÌNH ENCODER-DECODER MỚI)")
    print("="*90)
    print(f"Cấu hình: Window={WINDOW_SIZE}, d_model=128, Encoder (5 Alpha + Temporal), Decoder (Close + Temporal)")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*90)

    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# PHẦN 6. VẼ BIỂU ĐỒ SO SÁNH
# =====================================
def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 8))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2, alpha=0.8)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    plt.title(f'SO SÁNH DỰ ĐOÁN VÀ THỰC TẾ CỔ PHIẾU {symbol} (CẤU HÌNH ENCODER-DECODER MỚI)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16, fontweight='bold')
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend(fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# =====================================
# THỰC THI CHÍNH
# =====================================
if __name__ == '__main__':
    try:
        training_results = main(FILE_PATH, SYMBOL)

        if training_results:
            print("\n6. Đang vẽ biểu đồ kết quả...")
            plot_predictions_vs_actuals(
                test_dates=training_results["test_dates"],
                predictions=training_results["predictions_unscaled"],
                actuals=training_results["actuals_unscaled"],
                symbol=SYMBOL,
                mape=training_results["metrics"]["mape"]
            )

            print("\n7. Dự đoán giá cho ngày giao dịch tiếp theo...")
            next_day_price = predict_next_day(
                model=training_results["model"],
                scaler_src=training_results["scalers"]["src"],
                scaler_tgt_input=training_results["scalers"]["tgt_input"],
                scaler_tgt_output=training_results["scalers"]["tgt_output"],
                recent_data=training_results["df_processed"],
                src_cols=training_results["columns"]["source"],
                tgt_in_cols=training_results["columns"]["target_input"]
            )

            if next_day_price is not None:
                last_date = training_results["df_processed"]['date'].iloc[-1].date()
                last_price = training_results['df_processed']['close'].iloc[-1]
                price_change = next_day_price - last_price
                change_percent = (price_change / last_price) * 100

                print("\n" + "="*50)
                print("DỰ ĐOÁN GIÁ TIẾP THEO")
                print("="*50)
                print(f"Mã cổ phiếu: {SYMBOL}")
                print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
                print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
                print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
                print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
                print("="*50)
    except Exception as e:
        print(f"\nLỖI CHƯƠNG TRÌNH: Đã xảy ra lỗi không mong muốn trong quá trình thực thi: {e}")
        traceback.print_exc()
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os
import traceback

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ VÀ HẰNG SỐ (CẬP NHẬT: VCB)
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB" # <--- CẬP NHẬT: VCB
# Các mã sentiment cần thiết: VCB, TCB
ADDITIONAL_SENTIMENT_COMPANIES = ['TCB'] # <--- CẬP NHẬT: Chỉ cần TCB (VCB đã có)

# Đường dẫn file đã được sửa
FILE_PATH = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"

SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 64
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA: TÍNH TIs, ALPHA VÀ TEMPORAL FEATURES
# =====================================

def z_score(series, window=None):
    """
    Tính toán Z-score (chuẩn hóa) của một chuỗi.
    SỬA LỖI: Xử lý an toàn khi mẫu số (std) là float hoặc Series và bằng 0.
    """
    if window:
        # Trường hợp rolling: Mẫu số là Series, dùng .replace()
        std_val = series.rolling(window=window).std().replace(0, 1e-6)
        return (series - series.rolling(window=window).mean()) / std_val
    else:
        # Trường hợp z-score toàn cục: Mẫu số là float.
        # Phải kiểm tra và thay thế 0 bằng 1e-6 cho mẫu số (float)
        mean_val = series.mean()
        std_val = series.std()

        # SỬA LỖI: Dùng np.where để kiểm tra std_val
        safe_std = np.where(std_val == 0, 1e-6, std_val)

        # Nếu series là một Series (khi gọi trong alpha), phép tính này sẽ broadcast đúng
        return (series - mean_val) / safe_std

# Định nghĩa lại hàm z_score cho dễ đọc trong công thức alpha
z = z_score
def calculate_missing_technical_indicators(df):
    """
    Tính toán các chỉ báo kỹ thuật còn thiếu (hoặc cần thiết cho công thức Alpha mới).
    Sử dụng các cột đã có sẵn trong file CSV nếu có (SMA_20, BB_Upper, BB_Lower, MACD, MACD_Signal, OBV).
    """
    df = df.copy()

    # Đảm bảo các cột sentiment tồn tại và được đặt tên ngắn gọn
    sentiment_vcb = f'{TARGET_STOCK_CODE}_daily_new_sentiment_score' # VCB_daily_new_sentiment_score
    sentiment_tcb = 'TCB_daily_new_sentiment_score'

    # Đổi tên cho dễ dùng trong Alpha
    df['vcb_sent'] = df[sentiment_vcb].fillna(0)
    df['tcb_sent'] = df[sentiment_tcb].fillna(0) # Đảm bảo TCB_daily_new_sentiment_score được tải

    # 1. Các cột giá/volume an toàn và shift (cần cho Alpha)
    df['close_safe'] = df['close'].replace(0, 1e-9) # Giá đóng cửa an toàn (tránh chia cho 0)
    df['close_safe_shift_1'] = df['close_safe'].shift(1)
    df['close_safe_shift_6'] = df['close_safe'].shift(6)
    df['close_safe_shift_11'] = df['close_safe'].shift(11)
    df['close_safe_shift_3'] = df['close_safe'].shift(3)
    df['close_safe_shift_20'] = df['close_safe'].shift(20)

    # 2. Cột chỉ báo dựa trên các cột đã có trong CSV
    # Cần: SMA_20, BB_Upper, BB_Lower, MACD, MACD_Signal, OBV, volume

    # Tính lại hoặc dùng lại các chỉ báo cần thiết
    window_bb = 20
    df['SMA_20'] = df['close'].rolling(window=window_bb).mean()
    df['Std_20_safe'] = df['close'].rolling(window=window_bb).std().replace(0, 1e-9).fillna(0)
    df['BB_Upper'] = df['SMA_20'] + 2 * df['Std_20_safe']
    df['BB_Lower'] = df['SMA_20'] - 2 * df['Std_20_safe']

    ema_fast = df['close'].ewm(span=12, adjust=False).mean()
    ema_slow = df['close'].ewm(span=26, adjust=False).mean()
    df['MACD'] = ema_fast - ema_slow
    df['MACD_Signal'] = df['MACD'].ewm(span=9, adjust=False).mean()


    # 3. Các chỉ báo phụ cần thiết cho Alpha mới
    df['vol_sma_20'] = df['volume'].rolling(window=window_bb).mean().replace(0, 1e-9).fillna(0)
    df['MACD_Hist'] = df['MACD'] - df['MACD_Signal']
    df['bb_width'] = (df['BB_Upper'] - df['BB_Lower']) / df['SMA_20'].replace(0, 1e-9)

    # OBV đã có trong file, đảm bảo nó không NaN
    if 'OBV' not in df.columns:
        print("CẢNH BÁO: Cột 'OBV' không tìm thấy trong dữ liệu gốc. Alpha 3 có thể bị ảnh hưởng.")
        # Nếu không có, tạo cột OBV tạm thời bằng 0
        df['OBV'] = 0.0
    df['OBV'] = df['OBV'].fillna(0)

    # Đảm bảo các cột shift/rolling cơ bản từ code gốc vẫn tồn tại (nếu cần)
    df['close_shift_1'] = df['close'].shift(1)
    df['close_shift_5'] = df['close'].shift(5)
    df['close_shift_3'] = df['close'].shift(3)
    df['close_shift_20'] = df['close'].shift(20)
    df['volume_rolling_20_mean'] = df['volume'].rolling(window=20).mean()

    # Fill NA cho các cột mới tạo
    df = df.fillna(0)

    return df

def load_stock_data_from_df(file_path, symbol):
    # Giữ nguyên logic tải và xử lý ngày
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    # Xử lý tên cột: đảm bảo các cột quan trọng có tên chuẩn
    rename_map = {
        'Thời gian': 'date',
        'VCB_daily_new_sentiment_score': 'VCB_daily_new_sentiment_score',
        'TCB_daily_new_sentiment_score': 'TCB_daily_new_sentiment_score'
        # Các cột khác như close, volume, SMA_20, BB_Upper, MACD,... đã có tên chuẩn
    }
    df.rename(columns=rename_map, inplace=True)

    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    df = df.sort_values('date').reset_index(drop=True)
    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    print("\n-> Đang kiểm tra và tính toán các Chỉ báo Kỹ thuật còn thiếu...")
    # TÍNH TOÁN CÁC CHỈ BÁO PHỤ CẦN THIẾT CHO CÔNG THỨC ALPHA MỚI
    df = calculate_missing_technical_indicators(df)

    print(f"Đã tải {len(df)} dòng dữ liệu")
    return df

def calculate_alpha_features(df):
    """
    Tính toán 5 trường alpha mới cho VCB dựa trên công thức yêu cầu.
    """
    df = df.copy()

    # Đặt lại tên cột sentiment đã chuẩn hóa trong hàm calculate_missing_technical_indicators
    # df['vcb_sent'] và df['tcb_sent'] đã được tạo

    # Hàm z_score toàn cục được định nghĩa bên ngoài hàm này

    # --- CÁC CÔNG THỨC ALPHA MỚI ---

    # Alpha 1: Tương quan giữa Momentum ngắn (6d), dài (11d), Biến động (Std_20) và Sentiment hiện tại/momentum sentiment.
    df['alpha1'] = ((0.6 * ((df['close_safe_shift_1'] / df['close_safe_shift_6']) - 1) +
                     0.4 * ((df['close_safe_shift_1'] / df['close_safe_shift_11']) - 1)) /
                    (df['Std_20_safe'].replace(0, 1e-9))) * \
                   (1 + 0.95 * df['vcb_sent'].shift(1).fillna(0)) * \
                   (1 - 0.35 * np.tanh(-df['vcb_sent'].diff(3).fillna(0)))

    # Alpha 2: Volume/SMA_Vol_20, MACD History, Vị thế BB (nằm ngoài BB Upper/Lower), Sentiment
    df['alpha2'] = ((df['volume'] / df['vol_sma_20'].replace(0, 1e-9)) * (np.sign(df['MACD_Hist'].fillna(0)) * np.abs(df['MACD_Hist'].fillna(0))**0.7)) * \
                   ((df['close_safe'] > df['BB_Upper']).astype(float) -
                    (df['close_safe'] < df['BB_Lower']).astype(float)) * \
                   (1 + 0.5 * df['vcb_sent'].shift(1).fillna(0)) / \
                   (1 + df['bb_width'].fillna(0))

    # Alpha 3: Giá/SMA_20, Chiều rộng BB (z-score), Sentiment, Tốc độ thay đổi OBV
    df['alpha3'] = -1 * ((df['close_safe'] - df['SMA_20']) / df['SMA_20'].replace(0, 1e-9)) * \
                   (1 - z_score(df['bb_width'].fillna(0))) * \
                   (1 + 0.6 * df['vcb_sent'].shift(1).fillna(0)) * \
                   ((df['OBV'] - df['OBV'].shift(1)).fillna(0) / df['volume'].replace(0, 1e-9))

    # Alpha 4: Sentiment tương đối (VCB vs TCB), Momentum 3d, Biến động 20d, Tương quan sentiment 7d
    df['alpha4'] = ((df['vcb_sent'].shift(1).fillna(0) - df['tcb_sent'].shift(1).fillna(0)) * ((df['close_safe'] / df['close_safe_shift_3']) - 1).fillna(0)) / \
                   (1 + np.abs(((df['close_safe'] / df['close_safe_shift_20']) - 1).fillna(0)).replace(0, 1e-9)) * \
                   (1 + 0.3 * df['vcb_sent'].rolling(7).corr(df['tcb_sent']).fillna(0))

    # Alpha 5: Kết hợp chuẩn hóa (z) của Momentum 5d, MACD Hist và Momentum Sentiment 7d, điều chỉnh theo Volume.
    df['alpha5'] = np.tanh(
                       0.5 * z(((df['close_safe'] / df['close_safe'].shift(5)) - 1).fillna(0) / (df['Std_20_safe'].replace(0, 1e-9))) +
                       0.3 * z(df['MACD_Hist'].fillna(0)) +
                       0.2 * z(df['vcb_sent'].rolling(7).mean().diff().fillna(0))
                   ) * (1 + 0.15 * ((df['volume'] / df['vol_sma_20'].replace(0, 1e-9)) - 1).fillna(0))

    # --- Dọn dẹp và làm sạch ---

    # Xóa các cột tạm và shift
    cols_to_drop = [col for col in df.columns if col.startswith('close_safe') or col.endswith('_safe') or col.startswith('vol_') or col.endswith('_Hist') or col.startswith('bb_')]

    # Giữ lại các cột TI, Sentiment gốc: SMA_20, BB_Upper, BB_Lower, MACD, MACD_Signal, OBV, vcb_sent, tcb_sent
    # Xóa các cột shift/temp khác nếu muốn

    cols_to_drop.extend(['close_shift_1', 'close_shift_5', 'close_shift_3', 'close_shift_20', 'volume_rolling_20_mean'])

    # Thêm cột sentiment ngắn gọn vào danh sách xóa nếu không cần thiết sau khi tính alpha
    # cols_to_drop.extend(['vcb_sent', 'tcb_sent']) # Bỏ qua vì chúng có thể cần cho các bước khác (như chuẩn hóa)

    df = df.drop(cols_to_drop, axis=1, errors='ignore')

    # Thay thế các giá trị vô cùng và NaN bằng 0
    df = df.replace([np.inf, -np.inf], 0)
    df = df.fillna(0)

    print(f"Đã tính toán xong 5 alpha features mới cho {TARGET_STOCK_CODE}.")
    return df

def add_temporal_features(df):
    """
    Tạo các tính năng liên quan đến thời gian (Temporal Features):
    Day of Week (1-5), Day of Month (1-31), Month of Year (1-12)
    """
    df['day_of_week'] = df['date'].dt.dayofweek + 1 # 1=Thứ 2, 5=Thứ 6
    df['day_of_month'] = df['date'].dt.day
    df['month_of_year'] = df['date'].dt.month
    return df

def prepare_data_with_sentiment(df):
    df = df.copy()

    df = calculate_alpha_features(df)
    df = add_temporal_features(df)

    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # Loại bỏ các hàng mà các Alpha quan trọng không tính được
    check_cols = ['MACD', 'alpha1']

    initial_len = len(df)
    df_cleaned = df[~(df[check_cols] == 0).any(axis=1)].reset_index(drop=True)

    print(f"Sau khi làm sạch: {len(df_cleaned)} dòng dữ liệu (Đã loại bỏ {initial_len - len(df_cleaned)} dòng do thiếu dữ liệu lịch sử).")
    return df_cleaned

# ... (Giữ nguyên các hàm: create_sliding_windows, StockDataset) ...

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)

class StockDataset(Dataset):
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

# =====================================
# PHẦN 3. ĐỊNH NGHĨA MODEL (ĐÃ SỬA LỖI PositionalEncoding)
# =====================================
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0) # Shape: (1, max_len, d_model)
        self.register_buffer('pe', pe)

    def forward(self, x):
        # x shape: (batch_size, seq_len, d_model)
        # Sửa lỗi: Lấy pe theo seq_len (chiều 1 của x)
        x = x + self.pe[:, :x.size(1), :] # Dạng (batch, seq_len, d_model) + (1, seq_len, d_model)
        return x

class TemporalEmbedding(nn.Module):
    def __init__(self, input_temporal_dim, d_model):
        super(TemporalEmbedding, self).__init__()
        self.linear_temporal = nn.Linear(input_temporal_dim, d_model)

    def forward(self, x):
        return self.linear_temporal(x)

class StockEncoderDecoderTransformer(nn.Module):
    def __init__(self, input_dim_src, input_dim_tgt, d_model=128, nhead=8, num_layers=2, dropout=0.2, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model
        self.window_size = window_size

        self.temporal_dim = 3
        self.alpha_dim = 5
        self.price_dim = 1

        # --- Encoder Components (5 Alpha + Temporal) ---
        self.conv_emb_src_alpha = nn.Conv1d(self.alpha_dim, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(self.temporal_dim, d_model)

        self.pos_encoder = PositionalEncoding(d_model, max_len=window_size)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=512, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components (1 Close Price + Temporal) ---
        self.conv_emb_tgt_price = nn.Conv1d(self.price_dim, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(self.temporal_dim, d_model)

        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=512, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 256), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def _get_embedding(self, x_data, x_temporal, conv_emb_layer, temporal_emb_layer):
        # 1. Conv Embedding (Áp dụng cho Data Features: Alpha hoặc Price)
        x_data_conv = x_data.transpose(1, 2)
        x_conv = conv_emb_layer(x_data_conv)
        x_conv = x_conv.transpose(1, 2)

        # 2. Temporal Embedding (Áp dụng cho Temporal Features)
        x_temporal_emb = temporal_emb_layer(x_temporal)

        # 3. Kết hợp (Combine)
        x_emb = x_conv + x_temporal_emb

        # 4. Position Embedding (Thêm vào sau cùng)
        x_final = self.pos_encoder(x_emb)
        return x_final

    def forward(self, x_src, x_tgt_in):
        # Tách Alpha (5) và Temporal (3)
        x_src_alpha = x_src[:, :, :self.alpha_dim]
        x_src_temporal = x_src[:, :, self.alpha_dim:]

        # Tách Price (1) và Temporal (3)
        x_tgt_in_price = x_tgt_in[:, :, :self.price_dim]
        x_tgt_in_temporal = x_tgt_in[:, :, self.price_dim:]

        # Encoder Embedding
        src_emb = self._get_embedding(
            x_src_alpha, x_src_temporal,
            self.conv_emb_src_alpha, self.temporal_emb_src
        )
        memory = self.transformer_encoder(src_emb)

        # Decoder Embedding
        tgt_emb = self._get_embedding(
            x_tgt_in_price, x_tgt_in_temporal,
            self.conv_emb_tgt_price, self.temporal_emb_tgt
        )
        output = self.transformer_decoder(tgt_emb, memory)

        # Prediction Head
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

# ... (Giữ nguyên các hàm: train_model, predict_next_day) ...

def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        train_losses.append(train_loss)
        val_losses.append(val_loss)

        if (epoch + 1) % 5 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price


# =====================================
# PHẦN 5. MAIN EXECUTION
# =====================================
def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER (CẤU HÌNH MỚI) ===")
    print(f"=== Cấu hình: Window Size={WINDOW_SIZE}, Encoder Input: 5 Alpha + Temporal | Decoder Input: Close + Temporal ===")

    print("\n1. Đang tải và xử lý dữ liệu...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        return None

    df_processed = prepare_data_with_sentiment(df)

    temporal_columns = ['day_of_week', 'day_of_month', 'month_of_year']
    alpha_columns = ['alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5']

    source_columns = alpha_columns + temporal_columns
    target_input_columns = ['close'] + temporal_columns
    target_output_column = 'Pct_Change'

    source_columns = [col for col in source_columns if col in df_processed.columns]
    target_input_columns = [col for col in target_input_columns if col in df_processed.columns]

    if len(source_columns) != 8:
        print(f"\nLỖI: Thiếu cột Encoder Input. Chỉ tìm thấy {len(source_columns)}/8. Vui lòng kiểm tra file data.")
        return None
    if len(target_input_columns) != 4:
        print(f"\nLỖI: Thiếu cột Decoder Input. Chỉ tìm thấy {len(target_input_columns)}/4. Vui lòng kiểm tra file data.")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"SỐ LƯỢNG FEATURES ENCODER: {len(source_columns)} (5 Alpha + 3 Temporal)")
    print(f"SỐ LƯỢNG FEATURES DECODER INPUT: {len(target_input_columns)} (1 Close + 3 Temporal)")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    # ... (Kiểm tra dữ liệu Train/Test) ...
    if df_train.empty or len(df_train) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu training không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None
    if df_test.empty or len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")


    # --- TẠO CỬA SỔ TRƯỢT VÀ CHUẨN HÓA DỮ LIỆU ---
    print("\n3. Đang tạo cửa sổ trượt và chuẩn hóa dữ liệu...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n4. Đang khởi tạo và huấn luyện mô hình (d_model=128, nhead=8, num_layers=2, dropout=0.2)...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns),
        input_dim_tgt=len(target_input_columns),
        d_model=128,
        nhead=8,
        num_layers=2,
        dropout=0.2,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    print("\n5. Đang đánh giá trên tập Test...")
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    close_price_T_minus_1 = df_test['close'].iloc[WINDOW_SIZE-1:WINDOW_SIZE-1 + len(pct_change_pred_unscaled)].values
    actuals_unscaled = df_test['close'].iloc[WINDOW_SIZE:WINDOW_SIZE + len(pct_change_pred_unscaled)].values

    min_len = min(len(close_price_T_minus_1), len(pct_change_pred_unscaled))

    close_price_T_minus_1 = close_price_T_minus_1[:min_len]
    pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
    actuals_unscaled = actuals_unscaled[:min_len]
    test_dates = test_dates[:min_len]

    # Giá dự đoán ngày T = Giá T-1 * (1 + Pct_Change T / 100)
    predictions_unscaled = close_price_T_minus_1 * (1 + pct_change_pred_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*90)
    print(f"KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST ({symbol}: CẤU HÌNH ENCODER-DECODER MỚI)")
    print("="*90)
    print(f"Cấu hình: Window={WINDOW_SIZE}, d_model=128, Encoder (5 Alpha + Temporal), Decoder (Close + Temporal)")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*90)

    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# PHẦN 6. VẼ BIỂU ĐỒ SO SÁNH
# =====================================
def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 8))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2, alpha=0.8)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    plt.title(f'SO SÁNH DỰ ĐOÁN VÀ THỰC TẾ CỔ PHIẾU {symbol} (CẤU HÌNH ENCODER-DECODER MỚI)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16, fontweight='bold')
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend(fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# =====================================
# THỰC THI CHÍNH
# =====================================
if __name__ == '__main__':
    try:
        training_results = main(FILE_PATH, SYMBOL)

        if training_results:
            print("\n6. Đang vẽ biểu đồ kết quả...")
            plot_predictions_vs_actuals(
                test_dates=training_results["test_dates"],
                predictions=training_results["predictions_unscaled"],
                actuals=training_results["actuals_unscaled"],
                symbol=SYMBOL,
                mape=training_results["metrics"]["mape"]
            )

            print("\n7. Dự đoán giá cho ngày giao dịch tiếp theo...")
            next_day_price = predict_next_day(
                model=training_results["model"],
                scaler_src=training_results["scalers"]["src"],
                scaler_tgt_input=training_results["scalers"]["tgt_input"],
                scaler_tgt_output=training_results["scalers"]["tgt_output"],
                recent_data=training_results["df_processed"],
                src_cols=training_results["columns"]["source"],
                tgt_in_cols=training_results["columns"]["target_input"]
            )

            if next_day_price is not None:
                last_date = training_results["df_processed"]['date'].iloc[-1].date()
                last_price = training_results['df_processed']['close'].iloc[-1]
                price_change = next_day_price - last_price
                change_percent = (price_change / last_price) * 100

                print("\n" + "="*50)
                print("DỰ ĐOÁN GIÁ TIẾP THEO")
                print("="*50)
                print(f"Mã cổ phiếu: {SYMBOL}")
                print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
                print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
                print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
                print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
                print("="*50)
    except Exception as e:
        print(f"\nLỖI CHƯƠNG TRÌNH: Đã xảy ra lỗi không mong muốn trong quá trình thực thi: {e}")
        traceback.print_exc()
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import warnings
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os
import traceback

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# =====================================
# PHẦN 1. CHUẨN BỊ VÀ HẰNG SỐ (CẬP NHẬT: VCB)
# =====================================
# Khai báo các hằng số
TARGET_STOCK_CODE = "VCB" # <--- CẬP NHẬT: VCB
# Các mã sentiment cần thiết: VCB, TCB, VIC (VIC có trong CSV của bạn)
ADDITIONAL_SENTIMENT_COMPANIES = ['TCB', 'VIC'] # Đã sửa, chỉ dùng các mã có trong file

# Đường dẫn file đã được sửa
FILE_PATH = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"

SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 64
EPOCHS = 50
LEARNING_RATE = 5e-5

# =====================================
# PHẦN 2. XỬ LÝ DATA: TÍNH TIs, ALPHA VÀ TEMPORAL FEATURES
# =====================================

def z_score(series, window=None):
    """
    Tính toán Z-score (chuẩn hóa) an toàn (Đã sửa lỗi float object).
    """
    if window:
        # Trường hợp rolling: Mẫu số là Series, dùng .replace()
        std_val = series.rolling(window=window).std().replace(0, 1e-6)
        return (series - series.rolling(window=window).mean()) / std_val
    else:
        # Trường hợp z-score toàn cục: Mẫu số là float.
        mean_val = series.mean()
        std_val = series.std()
        safe_std = np.where(std_val == 0, 1e-6, std_val)
        return (series - mean_val) / safe_std

# Định nghĩa lại hàm z_score và z
z = z_score

def rankpct(series):
    """Tính toán rank của series và chuẩn hóa về [0, 1]."""
    return series.rank(pct=True).fillna(0)


def calculate_missing_technical_indicators(df):
    """
    Tính toán các chỉ báo kỹ thuật còn thiếu, CHỈ SỬ DỤNG CÁC CỘT CÓ SẴN
    (VCB, TCB, VIC sentiment). Đã khắc phục lỗi KeyError cho cột shift.
    """
    df = df.copy()

    # 1. Xử lý và Đổi tên cột Sentiment (CHỈ DÙNG VCB, TCB, VIC)
    df['vcb_sent'] = df[f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'].fillna(0)
    df['tcb_sent'] = df['TCB_daily_new_sentiment_score'].fillna(0)

    if 'VIC_daily_new_sentiment_score' in df.columns:
        df['vic_sent'] = df['VIC_daily_new_sentiment_score'].fillna(0)
    else:
        df['vic_sent'] = 0.0

    # 2. Các cột giá/volume an toàn và shift (ĐÃ KHẮC PHỤC KEYERROR)
    df['close_safe'] = df['close'].replace(0, 1e-9)

    # Tạo tất cả các cột shift cần thiết cho 5 Alpha
    df['close_safe_shift_1'] = df['close_safe'].shift(1)
    df['close_safe_shift_3'] = df['close_safe'].shift(3)
    df['close_safe_shift_5'] = df['close_safe'].shift(5)
    df['close_safe_shift_6'] = df['close_safe'].shift(6)
    df['close_safe_shift_11'] = df['close_safe'].shift(11)
    df['close_safe_shift_12'] = df['close_safe'].shift(12) # <--- MỚI: Thêm shift 12
    df['close_safe_shift_20'] = df['close_safe'].shift(20)

    # Cột phụ cho Alpha 5 mới
    df['close_safe_sma_20'] = df['close_safe'].rolling(window=20).mean().replace(0, 1e-9) # <--- MỚI: SMA 20 của close_safe

    # ... (Đảm bảo các cột gốc từ code cũ vẫn tồn tại) ...
    df['close_shift_1'] = df['close'].shift(1)
    df['close_shift_5'] = df['close'].shift(5)
    df['close_shift_3'] = df['close'].shift(3)
    df['close_shift_20'] = df['close'].shift(20)
    df['volume_rolling_20_mean'] = df['volume'].rolling(window=20).mean()

    # 3. Tính toán các Chỉ báo Kỹ thuật Phụ
    window_bb = 20

    df['SMA_20'] = df['close'].rolling(window=window_bb).mean()
    df['Std_20_safe'] = df['close'].rolling(window=window_bb).std().replace(0, 1e-9).fillna(0)
    df['BB_Upper'] = df['SMA_20'] + 2 * df['Std_20_safe']
    df['BB_Lower'] = df['SMA_20'] - 2 * df['Std_20_safe']

    df['vol_sma_20'] = df['volume'].rolling(window=window_bb).mean().replace(0, 1e-9).fillna(0)
    df['MACD_Hist'] = df['MACD'] - df['MACD_Signal']
    df['bb_width'] = (df['BB_Upper'] - df['BB_Lower']) / df['SMA_20'].replace(0, 1e-9)

    if 'OBV' not in df.columns:
        df['OBV'] = 0.0
    df['OBV'] = df['OBV'].fillna(0)

    # Fill NA cho tất cả
    df = df.fillna(0)

    return df

def load_stock_data_from_df(file_path, symbol):
    # Giữ nguyên logic tải và xử lý ngày
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except UnicodeDecodeError:
        try:
            df = pd.read_csv(file_path, encoding='cp1252')
        except:
            df = pd.read_csv(file_path, encoding='latin-1')

    # Xử lý tên cột: đảm bảo các cột quan trọng có tên chuẩn
    rename_map = {
        'Thời gian': 'date',
        # ... (các cột khác)
    }
    df.rename(columns=rename_map, inplace=True)

    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    df = df.sort_values('date').reset_index(drop=True)
    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    print("\n-> Đang kiểm tra và tính toán các Chỉ báo Kỹ thuật còn thiếu...")
    # TÍNH TOÁN CÁC CHỈ BÁO PHỤ CẦN THIẾT CHO CÔNG THỨC ALPHA MỚI
    df = calculate_missing_technical_indicators(df)

    print(f"Đã tải {len(df)} dòng dữ liệu")
    return df

def calculate_alpha_features(df):
    """
    Tính toán 5 trường alpha mới cho VCB dựa trên công thức yêu cầu.
    (Đã cập nhật theo 5 công thức mới)
    """
    df = df.copy()

    # --- CÁC CÔNG THỨC ALPHA MỚI ---

    # Alpha 1: Momentum (6d, 12d) / Biến động (Std_20), điều chỉnh bởi Sentiment và Tốc độ thay đổi Sentiment
    df['alpha1'] = ((0.7*((df['close_safe'].shift(1)/df['close_safe'].shift(6))-1) +
                     0.3*((df['close_safe'].shift(1)/df['close_safe'].shift(12))-1)) /
                    (df['Std_20_safe'].replace(0,1e-9))) * \
                   (1 + 0.95 * df['vcb_sent'].shift(1).fillna(0)) * \
                   (1 - 0.4 * np.sign(df['vcb_sent'].diff(3).fillna(0)))

    # Alpha 2: Volume/SMA_Vol_20 * MACD Hist * Vị thế BB, điều chỉnh bởi Sentiment
    df['alpha2'] = ((df['volume'] / df['vol_sma_20'].replace(0,1e-9)).replace(0,1e-9) * \
                   (np.sign(df['MACD_Hist'].fillna(0)) * np.abs(df['MACD_Hist'].fillna(0))**0.6) * \
                   ((df['close_safe'] > df['BB_Upper']).astype(float) - (df['close_safe'] < df['BB_Lower']).astype(float))) / \
                   (1 + df['bb_width'].fillna(0)) * \
                   (1 + 0.5 * df['vcb_sent'].shift(1).fillna(0))

    # Alpha 3: Giá/SMA_20 (đảo ngược) * (1 - Z-score của BB Width) * Tốc độ thay đổi OBV * Sentiment
    # Lưu ý: Cần tính toán Z-score toàn cục cho bb_width
    bb_width_z = (df['bb_width'].fillna(0) - df['bb_width'].mean()) / (df['bb_width'].std(ddof=0) + 1e-9)

    df['alpha3'] = -1 * ((df['close_safe'] - df['SMA_20']) / df['SMA_20'].replace(0,1e-9)) * \
                   (1 - bb_width_z.fillna(0)) * \
                   ((df['OBV'] - df['OBV'].shift(1)).fillna(0) / df['volume'].replace(0,1e-9)) * \
                   (1 + 0.6 * df['vcb_sent'].shift(1).fillna(0))

    # Alpha 4: Sentiment Tương đối VCB/TCB * Momentum 3d / Biến động 20d * Correlation Sentiment
    df['alpha4'] = ((df['vcb_sent'].shift(1).fillna(0) - df['tcb_sent'].shift(1).fillna(0)) * \
                   ((df['close_safe'] / df['close_safe'].shift(3)) - 1).fillna(0)) / \
                   (1 + np.abs(((df['close_safe'] / df['close_safe'].shift(20)) - 1).fillna(0))).replace(0,1e-9) * \
                   (1 + 0.25 * df['vcb_sent'].rolling(7).corr(df['tcb_sent']).fillna(0))

    # Alpha 5: Z-score tổng hợp (Momentum 5d/Std, MACD Hist, Momentum Sentiment 7d), điều chỉnh bởi Vị thế giá so với SMA 20
    df['alpha5'] = np.tanh(
                       0.5 * z_score(((df['close_safe'] / df['close_safe'].shift(5)) - 1).fillna(0) / (df['Std_20_safe'].replace(0,1e-9))) +
                       0.3 * z_score(df['MACD_Hist'].fillna(0)) +
                       0.2 * z_score(df['vcb_sent'].rolling(7).mean().diff().fillna(0))
                   ) * (1 + 0.2 * ((df['close_safe'] / df['close_safe_sma_20']) - 1).fillna(0))

    # --- Dọn dẹp và làm sạch ---

    # Xóa các cột tạm và shift (Thêm 'close_safe_shift_12' và 'close_safe_sma_20' vào list dọn dẹp)
    cols_to_drop = [col for col in df.columns if col.startswith('close_safe') or col.endswith('_safe') or col.startswith('vol_') or col.endswith('_Hist') or col.startswith('bb_')]

    cols_to_drop.extend(['close_shift_1', 'close_shift_5', 'close_shift_3', 'close_shift_20', 'volume_rolling_20_mean'])

    # Loại bỏ các cột phụ đã thêm
    cols_to_drop.extend(['Alpha4_Avg_Sent', 'close_safe_shift_12', 'close_safe_sma_20'])

    df = df.drop(cols_to_drop, axis=1, errors='ignore')

    df = df.replace([np.inf, -np.inf], 0)
    df = df.fillna(0)

    print(f"Đã tính toán xong 5 alpha features mới cho {TARGET_STOCK_CODE}.")
    return df

def add_temporal_features(df):
    """
    Tạo các tính năng liên quan đến thời gian (Temporal Features):
    Day of Week (1-5), Day of Month (1-31), Month of Year (1-12)
    """
    df['day_of_week'] = df['date'].dt.dayofweek + 1 # 1=Thứ 2, 5=Thứ 6
    df['day_of_month'] = df['date'].dt.day
    df['month_of_year'] = df['date'].dt.month
    return df

def prepare_data_with_sentiment(df):
    df = df.copy()

    # 1. Tính Alpha và Temporal
    df = calculate_alpha_features(df)
    df = add_temporal_features(df)

    # 2. Tính Target Output
    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    # 3. LÀM SẠCH DỮ LIỆU (ĐÃ SỬA LỖI: Không loại bỏ hết 654 dòng)
    initial_len = len(df)

    # Độ trễ lớn nhất là 20 ngày (Std_20, shift_20). Loại bỏ 30 hàng đầu tiên cho an toàn.
    MIN_ROWS_TO_DROP = 30

    if initial_len > MIN_ROWS_TO_DROP:
        # Loại bỏ các hàng bị ảnh hưởng bởi rolling/shift
        df_cleaned = df.iloc[MIN_ROWS_TO_DROP:].reset_index(drop=True)
        # Loại bỏ NaN nếu còn sót lại ở các cột quan trọng
        df_cleaned = df_cleaned.dropna(subset=['close', 'alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5'])
    else:
        df_cleaned = pd.DataFrame()

    rows_dropped = initial_len - len(df_cleaned)

    # Đã sửa logic in ra để phản ánh đúng lỗi
    print(f"Sau khi làm sạch: {len(df_cleaned)} dòng dữ liệu (Đã loại bỏ {rows_dropped} dòng do thiếu dữ liệu lịch sử).")
    return df_cleaned

# ... (Giữ nguyên các hàm: create_sliding_windows, StockDataset, PositionalEncoding, TemporalEmbedding, StockEncoderDecoderTransformer, train_model, predict_next_day, main, plot_predictions_vs_actuals) ...

# ⚠️ LƯU Ý: Phần còn lại của code (PHẦN 3, 4, 5, 6 và THỰC THI CHÍNH)
# được giữ nguyên như trong code gốc của bạn. Tôi chỉ sửa lại các hàm ở PHẦN 1 & 2.

# =====================================
# PHẦN 3. ĐỊNH NGHĨA MODEL (ĐÃ SỬA LỖI PositionalEncoding)
# =====================================
class PositionalEncoding(nn.Module):
    # ... (Giữ nguyên)
    def __init__(self, d_model, max_len=5000):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0) # Shape: (1, max_len, d_model)
        self.register_buffer('pe', pe)

    def forward(self, x):
        # x shape: (batch_size, seq_len, d_model)
        # Sửa lỗi: Lấy pe theo seq_len (chiều 1 của x)
        x = x + self.pe[:, :x.size(1), :] # Dạng (batch, seq_len, d_model) + (1, seq_len, d_model)
        return x

class TemporalEmbedding(nn.Module):
    # ... (Giữ nguyên)
    def __init__(self, input_temporal_dim, d_model):
        super(TemporalEmbedding, self).__init__()
        self.linear_temporal = nn.Linear(input_temporal_dim, d_model)

    def forward(self, x):
        return self.linear_temporal(x)

class StockEncoderDecoderTransformer(nn.Module):
    # ... (Giữ nguyên)
    def __init__(self, input_dim_src, input_dim_tgt, d_model=128, nhead=8, num_layers=2, dropout=0.2, window_size=5):
        super(StockEncoderDecoderTransformer, self).__init__()
        self.d_model = d_model
        self.window_size = window_size

        self.temporal_dim = 3
        self.alpha_dim = 5
        self.price_dim = 1
        # ... (Phần còn lại của __init__ giữ nguyên)
        self.conv_emb_src_alpha = nn.Conv1d(self.alpha_dim, d_model, kernel_size=3, padding=1)
        self.temporal_emb_src = TemporalEmbedding(self.temporal_dim, d_model)

        self.pos_encoder = PositionalEncoding(d_model, max_len=window_size)
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=512, dropout=dropout, batch_first=True
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # --- Decoder Components (1 Close Price + Temporal) ---
        self.conv_emb_tgt_price = nn.Conv1d(self.price_dim, d_model, kernel_size=3, padding=1)
        self.temporal_emb_tgt = TemporalEmbedding(self.temporal_dim, d_model)

        decoder_layer = nn.TransformerDecoderLayer(
            d_model=d_model, nhead=nhead, dim_feedforward=512, dropout=dropout, batch_first=True
        )
        self.transformer_decoder = nn.TransformerDecoder(decoder_layer, num_layers=num_layers)

        # --- Prediction Head ---
        self.projection_head = nn.Sequential(
            nn.Linear(d_model * window_size, 256), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(128, 1)
        )

    def _get_embedding(self, x_data, x_temporal, conv_emb_layer, temporal_emb_layer):
        # 1. Conv Embedding (Áp dụng cho Data Features: Alpha hoặc Price)
        x_data_conv = x_data.transpose(1, 2)
        x_conv = conv_emb_layer(x_data_conv)
        x_conv = x_conv.transpose(1, 2)

        # 2. Temporal Embedding (Áp dụng cho Temporal Features)
        x_temporal_emb = temporal_emb_layer(x_temporal)

        # 3. Kết hợp (Combine)
        x_emb = x_conv + x_temporal_emb

        # 4. Position Embedding (Thêm vào sau cùng)
        x_final = self.pos_encoder(x_emb)
        return x_final

    def forward(self, x_src, x_tgt_in):
        # Tách Alpha (5) và Temporal (3)
        x_src_alpha = x_src[:, :, :self.alpha_dim]
        x_src_temporal = x_src[:, :, self.alpha_dim:]

        # Tách Price (1) và Temporal (3)
        x_tgt_in_price = x_tgt_in[:, :, :self.price_dim]
        x_tgt_in_temporal = x_tgt_in[:, :, self.price_dim:]

        # Encoder Embedding
        src_emb = self._get_embedding(
            x_src_alpha, x_src_temporal,
            self.conv_emb_src_alpha, self.temporal_emb_src
        )
        memory = self.transformer_encoder(src_emb)

        # Decoder Embedding
        tgt_emb = self._get_embedding(
            x_tgt_in_price, x_tgt_in_temporal,
            self.conv_emb_tgt_price, self.temporal_emb_tgt
        )
        output = self.transformer_decoder(tgt_emb, memory)

        # Prediction Head
        output_flat = output.reshape(output.size(0), -1)
        prediction = self.projection_head(output_flat)
        return prediction

def create_sliding_windows(source_data, target_input_data, target_output_data, window_size=5):
    X_src, X_tgt_in, y = [], [], []
    for i in range(len(source_data) - window_size):
        X_src.append(source_data[i:(i + window_size)])
        X_tgt_in.append(target_input_data[i:(i + window_size)])
        y.append(target_output_data[i + window_size])
    return np.array(X_src), np.array(X_tgt_in), np.array(y)

class StockDataset(Dataset):
    def __init__(self, X_src, X_tgt_in, y):
        self.X_src = torch.FloatTensor(X_src)
        self.X_tgt_in = torch.FloatTensor(X_tgt_in)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X_src)
    def __getitem__(self, idx):
        return self.X_src[idx], self.X_tgt_in[idx], self.y[idx]

def train_model(model, train_loader, val_loader, epochs, learning_rate):
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    train_losses, val_losses = [], []

    print(f"\nBắt đầu huấn luyện {epochs} epochs...")
    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for batch_X_src, batch_X_tgt_in, batch_y in train_loader:
            optimizer.zero_grad()
            predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
            loss = criterion(predictions, batch_y)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for batch_X_src, batch_X_tgt_in, batch_y in val_loader:
                predictions = model(batch_X_src, batch_X_tgt_in).squeeze()
                loss = criterion(predictions, batch_y)
                val_loss += loss.item()

        train_loss /= len(train_loader)
        val_loss /= len(val_loader)
        train_losses.append(train_loss)
        val_losses.append(val_loss)

        if (epoch + 1) % 5 == 0 or epoch == epochs - 1:
            print(f'Epoch [{epoch+1}/{epochs}], Train Loss: {train_loss:.6f}, Val Loss: {val_loss:.6f}')

    return train_losses, val_losses

def predict_next_day(model, scaler_src, scaler_tgt_input, scaler_tgt_output, recent_data, src_cols, tgt_in_cols):
    if len(recent_data) < WINDOW_SIZE:
        print(f"Lỗi: Cần ít nhất {WINDOW_SIZE} ngày dữ liệu để dự đoán.")
        return None

    last_window_data = recent_data.tail(WINDOW_SIZE)
    last_actual_close = last_window_data['close'].iloc[-1]

    features_src = last_window_data[src_cols].values
    features_src_scaled = scaler_src.transform(features_src)
    last_window_src_tensor = torch.FloatTensor(features_src_scaled).unsqueeze(0)

    features_tgt_in = last_window_data[tgt_in_cols].values
    features_tgt_in_scaled = scaler_tgt_input.transform(features_tgt_in)
    last_window_tgt_in_tensor = torch.FloatTensor(features_tgt_in_scaled).unsqueeze(0)

    model.eval()
    with torch.no_grad():
        scaled_pct_change_pred = model(last_window_src_tensor, last_window_tgt_in_tensor).item()

    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array([[scaled_pct_change_pred]]))[0, 0]
    predicted_price = last_actual_close * (1 + pct_change_pred_unscaled / 100)

    return predicted_price

def main(file_path, symbol):
    print(f"=== DỰ BÁO GIÁ CỔ PHIẾU {symbol} BẰNG TRANSFORMER ENCODER-DECODER (CẤU HÌNH MỚI) ===")
    print(f"=== Cấu hình: Window Size={WINDOW_SIZE}, Encoder Input: 5 Alpha + Temporal | Decoder Input: Close + Temporal ===")

    print("\n1. Đang tải và xử lý dữ liệu...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty:
        return None

    df_processed = prepare_data_with_sentiment(df)

    # Kiểm tra lại sau khi làm sạch
    if len(df_processed) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu training không đủ sau khi làm sạch (Cần > {WINDOW_SIZE} ngày).")
        return None

    temporal_columns = ['day_of_week', 'day_of_month', 'month_of_year']
    alpha_columns = ['alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5']

    source_columns = alpha_columns + temporal_columns
    target_input_columns = ['close'] + temporal_columns

    source_columns = [col for col in source_columns if col in df_processed.columns]
    target_input_columns = [col for col in target_input_columns if col in df_processed.columns]
    target_output_column = 'Pct_Change'

    if len(source_columns) != 8:
        print(f"\nLỖI: Thiếu cột Encoder Input. Chỉ tìm thấy {len(source_columns)}/8. Vui lòng kiểm tra file data.")
        return None
    if len(target_input_columns) != 4:
        print(f"\nLỖI: Thiếu cột Decoder Input. Chỉ tìm thấy {len(target_input_columns)}/4. Vui lòng kiểm tra file data.")
        return None

    print(f"Tổng số ngày dữ liệu sau khi làm sạch: {len(df_processed)}")
    print(f"SỐ LƯỢNG FEATURES ENCODER: {len(source_columns)} (5 Alpha + 3 Temporal)")
    print(f"SỐ LƯỢNG FEATURES DECODER INPUT: {len(target_input_columns)} (1 Close + 3 Temporal)")

    # --- CHIA TRAIN/TEST THEO NGÀY ---
    TRAIN_START_DATE = pd.to_datetime('2024-01-01')
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')
    TEST_START_DATE = pd.to_datetime('2025-01-01')

    df_train = df_processed[
        (df_processed['date'] >= TRAIN_START_DATE) &
        (df_processed['date'] <= TRAIN_END_DATE)
    ].copy()

    df_test = df_processed[
        (df_processed['date'] >= TEST_START_DATE)
    ].copy()

    # ... (Kiểm tra dữ liệu Train/Test) ...
    if df_train.empty or len(df_train) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu training không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None
    if df_test.empty or len(df_test) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu test không đủ (Cần > {WINDOW_SIZE} ngày).")
        return None

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")


    # --- TẠO CỬA SỔ TRƯỢT VÀ CHUẨN HÓA DỮ LIỆU ---
    print("\n3. Đang tạo cửa sổ trượt và chuẩn hóa dữ liệu...")
    X_src_train_raw, X_tgt_in_train_raw, y_train_raw = create_sliding_windows(
        df_train[source_columns].values,
        df_train[target_input_columns].values,
        df_train[target_output_column].values,
        WINDOW_SIZE
    )

    X_src_test_raw, X_tgt_in_test_raw, y_test_raw = create_sliding_windows(
        df_test[source_columns].values,
        df_test[target_input_columns].values,
        df_test[target_output_column].values,
        WINDOW_SIZE
    )

    test_dates = df_test['date'].dt.date.values[WINDOW_SIZE:]

    scaler_src = StandardScaler()
    X_src_train = scaler_src.fit_transform(X_src_train_raw.reshape(-1, len(source_columns))).reshape(X_src_train_raw.shape)
    X_src_test = scaler_src.transform(X_src_test_raw.reshape(-1, len(source_columns))).reshape(X_src_test_raw.shape)

    scaler_tgt_input = StandardScaler()
    X_tgt_in_train = scaler_tgt_input.fit_transform(X_tgt_in_train_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_train_raw.shape)
    X_tgt_in_test = scaler_tgt_input.transform(X_tgt_in_test_raw.reshape(-1, len(target_input_columns))).reshape(X_tgt_in_test_raw.shape)

    scaler_tgt_output = StandardScaler()
    y_train = scaler_tgt_output.fit_transform(y_train_raw.reshape(-1, 1)).squeeze()
    y_test = scaler_tgt_output.transform(y_test_raw.reshape(-1, 1)).squeeze()

    print(f"Train samples (đã tạo window): {len(X_src_train)}")
    print(f"Test samples (đã tạo window): {len(X_src_test)}")

    # Tạo DataLoader
    train_loader = DataLoader(StockDataset(X_src_train, X_tgt_in_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(StockDataset(X_src_test, X_tgt_in_test, y_test), batch_size=BATCH_SIZE, shuffle=False)

    # Định nghĩa và Train Model
    print(f"\n4. Đang khởi tạo và huấn luyện mô hình (d_model=128, nhead=8, num_layers=2, dropout=0.2)...")
    model = StockEncoderDecoderTransformer(
        input_dim_src=len(source_columns),
        input_dim_tgt=len(target_input_columns),
        d_model=128,
        nhead=8,
        num_layers=2,
        dropout=0.2,
        window_size=WINDOW_SIZE
    )
    train_model(model, train_loader, test_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE)

    # Đánh giá cuối cùng trên Test set
    print("\n5. Đang đánh giá trên tập Test...")
    model.eval()
    predictions_scaled = []
    with torch.no_grad():
        for batch_X_src, batch_X_tgt_in, _ in test_loader:
            batch_pred = model(batch_X_src, batch_X_tgt_in).squeeze()
            if batch_pred.dim() == 0:
                predictions_scaled.append(batch_pred.item())
            else:
                predictions_scaled.extend(batch_pred.cpu().numpy().tolist())

    # Inverse Transform Pct_Change
    pct_change_pred_unscaled = scaler_tgt_output.inverse_transform(np.array(predictions_scaled).reshape(-1, 1)).squeeze()

    # Tính lại giá trị tuyệt đối từ Pct_Change
    close_price_T_minus_1 = df_test['close'].iloc[WINDOW_SIZE-1:WINDOW_SIZE-1 + len(pct_change_pred_unscaled)].values
    actuals_unscaled = df_test['close'].iloc[WINDOW_SIZE:WINDOW_SIZE + len(pct_change_pred_unscaled)].values

    min_len = min(len(close_price_T_minus_1), len(pct_change_pred_unscaled))

    close_price_T_minus_1 = close_price_T_minus_1[:min_len]
    pct_change_pred_unscaled = pct_change_pred_unscaled[:min_len]
    actuals_unscaled = actuals_unscaled[:min_len]
    test_dates = test_dates[:min_len]

    # Giá dự đoán ngày T = Giá T-1 * (1 + Pct_Change T / 100)
    predictions_unscaled = close_price_T_minus_1 * (1 + pct_change_pred_unscaled / 100)

    # Tính toán lỗi
    mse = mean_squared_error(actuals_unscaled, predictions_unscaled)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((actuals_unscaled - predictions_unscaled) / (actuals_unscaled + 1e-6))) * 100

    print("\n" + "="*90)
    print(f"KẾT QUẢ ĐÁNH GIÁ TRÊN TẬP TEST ({symbol}: CẤU HÌNH ENCODER-DECODER MỚI)")
    print("="*90)
    print(f"Cấu hình: Window={WINDOW_SIZE}, d_model=128, Encoder (5 Alpha + Temporal), Decoder (Close + Temporal)")
    print(f"MSE:  {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"MAPE: {mape:.2f}%")
    print("="*90)

    results = {
        "model": model,
        "df_processed": df_processed,
        "test_dates": test_dates,
        "predictions_unscaled": predictions_unscaled,
        "actuals_unscaled": actuals_unscaled,
        "scalers": {
            "src": scaler_src,
            "tgt_input": scaler_tgt_input,
            "tgt_output": scaler_tgt_output
        },
        "columns": {
            "source": source_columns,
            "target_input": target_input_columns
        },
        "metrics": {
            "mse": mse,
            "rmse": rmse,
            "mape": mape
        }
    }
    return results

# =====================================
# PHẦN 6. VẼ BIỂU ĐỒ SO SÁNH
# =====================================
def plot_predictions_vs_actuals(test_dates, predictions, actuals, symbol, mape):
    min_len = min(len(test_dates), len(actuals), len(predictions))
    test_dates = test_dates[:min_len]
    actuals = actuals[:min_len]
    predictions = predictions[:min_len]

    plt.figure(figsize=(15, 8))
    try:
        dates_for_plot = [datetime.combine(d, datetime.min.time()) for d in test_dates]
    except TypeError:
        dates_for_plot = test_dates

    plt.plot(dates_for_plot, actuals, label='Giá Thực Tế (Actual Price)', color='green', linewidth=2, alpha=0.8)
    plt.plot(dates_for_plot, predictions, label='Giá Dự Đoán (Model Prediction)', color='darkorange', linestyle='--', linewidth=1.5)

    plt.title(f'SO SÁNH DỰ ĐOÁN VÀ THỰC TẾ CỔ PHIẾU {symbol} (CẤU HÌNH ENCODER-DECODER MỚI)\n(Test Set 2025+ | MAPE: {mape:.2f}%)', fontsize=16, fontweight='bold')
    plt.xlabel('Thời Gian', fontsize=12)
    plt.ylabel('Giá Đóng Cửa (VNĐ)', fontsize=12)

    plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
    plt.gcf().autofmt_xdate()

    plt.legend(fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# =====================================
# THỰC THI CHÍNH
# =====================================
if __name__ == '__main__':
    try:
        training_results = main(FILE_PATH, SYMBOL)

        if training_results:
            print("\n6. Đang vẽ biểu đồ kết quả...")
            plot_predictions_vs_actuals(
                test_dates=training_results["test_dates"],
                predictions=training_results["predictions_unscaled"],
                actuals=training_results["actuals_unscaled"],
                symbol=SYMBOL,
                mape=training_results["metrics"]["mape"]
            )

            print("\n7. Dự đoán giá cho ngày giao dịch tiếp theo...")
            next_day_price = predict_next_day(
                model=training_results["model"],
                scaler_src=training_results["scalers"]["src"],
                scaler_tgt_input=training_results["scalers"]["tgt_input"],
                scaler_tgt_output=training_results["scalers"]["tgt_output"],
                recent_data=training_results["df_processed"],
                src_cols=training_results["columns"]["source"],
                tgt_in_cols=training_results["columns"]["target_input"]
            )

            if next_day_price is not None:
                last_date = training_results["df_processed"]['date'].iloc[-1].date()
                last_price = training_results['df_processed']['close'].iloc[-1]
                price_change = next_day_price - last_price
                change_percent = (price_change / last_price) * 100

                print("\n" + "="*50)
                print("DỰ ĐOÁN GIÁ TIẾP THEO")
                print("="*50)
                print(f"Mã cổ phiếu: {SYMBOL}")
                print(f"Ngày cuối cùng trong dữ liệu: {last_date}")
                print(f"Giá đóng cửa cuối cùng: {last_price:,.0f} VNĐ")
                print(f"Giá dự đoán cho ngày tiếp theo: {next_day_price:,.0f} VNĐ")
                print(f"Thay đổi: {price_change:+,.0f} VNĐ ({change_percent:+.2f}%)")
                print("="*50)
    except Exception as e:
        print(f"\nLỖI CHƯƠNG TRÌNH: Đã xảy ra lỗi không mong muốn trong quá trình thực thi: {e}")
        traceback.print_exc()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM as KerasLSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import warnings
import os
import traceback
import matplotlib.pyplot as plt
from tabulate import tabulate

# Bỏ qua các cảnh báo không quan trọng
warnings.filterwarnings('ignore')

# Cấu hình để tắt cảnh báo TensorFlow
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

# =====================================
# PHẦN 1. CHUẨN BỊ VÀ HẰNG SỐ
# =====================================
TARGET_STOCK_CODE = "VCB"
FILE_PATH = f"/content/drive/MyDrive/cafef_news/{TARGET_STOCK_CODE}_final/{TARGET_STOCK_CODE}_final.csv"
SYMBOL = TARGET_STOCK_CODE
WINDOW_SIZE = 5
BATCH_SIZE = 64
EPOCHS = 50
LEARNING_RATE = 5e-4
# Kích thước dữ liệu cho TCN
TCN_FILTERS = 32
TCN_KERNEL_SIZE = 3
TCN_STACK_SIZE = 2 # Số lượng khối Dilated Conv

# =====================================
# PHẦN 2. XỬ LÝ DATA: TÍNH TIs, ALPHA VÀ TEMPORAL FEATURES (GIỮ NGUYÊN)
# =====================================

def z_score(series, window=None):
    """Tính toán Z-score an toàn."""
    if window:
        std_val = series.rolling(window=window).std().replace(0, 1e-6)
        return (series - series.rolling(window=window).mean()) / std_val
    else:
        mean_val = series.mean()
        std_val = series.std()
        safe_std = np.where(std_val == 0, 1e-6, std_val)
        return (series - mean_val) / safe_std
z = z_score

def rankpct(series):
    """Tính toán rank của series và chuẩn hóa về [0, 1]."""
    return series.rank(pct=True).fillna(0)


def calculate_missing_technical_indicators(df):
    """
    Tính toán các chỉ báo kỹ thuật còn thiếu, đã thêm shift(12) và ret_3d.
    """
    df = df.copy()

    # 1. Xử lý và Đổi tên cột Sentiment
    df['vcb_sent'] = df[f'{TARGET_STOCK_CODE}_daily_new_sentiment_score'].fillna(0)
    df['tcb_sent'] = df['TCB_daily_new_sentiment_score'].fillna(0)

    if 'VIC_daily_new_sentiment_score' in df.columns:
        df['vic_sent'] = df['VIC_daily_new_sentiment_score'].fillna(0)
    else:
        df['vic_sent'] = 0.0

    # 2. Các cột giá/volume an toàn và shift
    df['close_safe'] = df['close'].replace(0, 1e-9)

    df['close_safe_shift_1'] = df['close_safe'].shift(1)
    df['close_safe_shift_3'] = df['close_safe'].shift(3)
    df['close_safe_shift_5'] = df['close_safe'].shift(5)
    df['close_safe_shift_6'] = df['close_safe'].shift(6)
    df['close_safe_shift_11'] = df['close_safe'].shift(11)
    df['close_safe_shift_12'] = df['close_safe'].shift(12)
    df['close_safe_shift_20'] = df['close_safe'].shift(20)

    df['ret_3d'] = ((df['close'] / df['close'].shift(3)) - 1).fillna(0)

    df['volume_rolling_20_mean'] = df['volume'].rolling(window=20).mean()

    # 3. Tính toán các Chỉ báo Kỹ thuật Phụ
    window_bb = 20

    df['SMA_20'] = df['close'].rolling(window=window_bb).mean()
    df['Std_20_safe'] = df['close'].rolling(window=window_bb).std().replace(0, 1e-9).fillna(0)
    df['BB_Upper'] = df['SMA_20'] + 2 * df['Std_20_safe']
    df['BB_Lower'] = df['SMA_20'] - 2 * df['Std_20_safe']

    df['vol_sma_20'] = df['volume'].rolling(window=window_bb).mean().replace(0, 1e-9).fillna(0)
    # MACD đã có trong file CSV gốc, chỉ tính MACD Hist
    if 'MACD' in df.columns and 'MACD_Signal' in df.columns:
        df['MACD_Hist'] = df['MACD'] - df['MACD_Signal']
    else:
        df['MACD_Hist'] = 0.0

    df['bb_width'] = (df['BB_Upper'] - df['BB_Lower']) / df['SMA_20'].replace(0, 1e-9)

    if 'OBV' not in df.columns:
        df['OBV'] = 0.0
    df['OBV'] = df['OBV'].fillna(0)

    # Fill NA cho tất cả
    df = df.fillna(0)

    return df


def load_stock_data_from_df(file_path, symbol):
    # Giữ nguyên logic tải và xử lý ngày
    print(f"Đang đọc file: {file_path}")
    if not os.path.exists(file_path):
        print(f"LỖI: KHÔNG TÌM THẤY FILE tại đường dẫn: {file_path}")
        return pd.DataFrame()

    try:
        df = pd.read_csv(file_path, encoding='utf-8-sig')
    except:
        df = pd.read_csv(file_path, encoding='latin-1')

    df.rename(columns={'Thời gian': 'date'}, inplace=True)

    try:
        df['date'] = pd.to_datetime(df['date'], dayfirst=True)
    except:
        df['date'] = pd.to_datetime(df['date'])

    df = df.sort_values('date').reset_index(drop=True)
    start_date = pd.to_datetime('2024-01-01')
    df = df[df['date'] >= start_date].copy()

    if df.empty:
        print("LỖI: Dữ liệu bị rỗng sau khi lọc ngày bắt đầu.")
        return pd.DataFrame()

    print("\n-> Đang kiểm tra và tính toán các Chỉ báo Kỹ thuật còn thiếu...")
    df = calculate_missing_technical_indicators(df)

    print(f"Đã tải {len(df)} dòng dữ liệu")
    return df

def calculate_alpha_features(df):
    """
    Tính toán 5 trường alpha mới cho VCB dựa trên công thức cập nhật của người dùng.
    """
    df = df.copy()

    # --- CÁC CÔNG THỨC ALPHA MỚI ---

    # Alpha 1: Momentum/Volatility điều chỉnh bởi Sentiment và Tốc độ thay đổi Sentiment
    df['alpha1'] = ((0.65 * ((df['close_safe'].shift(1) / df['close_safe_shift_6']) - 1) +
                     0.35 * ((df['close_safe'].shift(1) / df['close_safe_shift_12']) - 1)) /
                    (df['Std_20_safe'].replace(0,1e-9))) * \
                   (1 + 0.98 * df['vcb_sent'].shift(1).fillna(0)) * \
                   (1 - 0.45 * np.tanh(-df['vcb_sent'].diff(3).fillna(0)))

    # Alpha 2: Volume Power/MACD Power, điều chỉnh Vị thế BB và Chiều rộng BB
    df['alpha2'] = ( ( (df['volume'] / df['vol_sma_20'].replace(0,1e-9)).replace(0,1e-9) )**0.6 * (np.sign(df['MACD_Hist'].fillna(0)) * np.abs(df['MACD_Hist'].fillna(0))**0.75) ) * \
                   ((df['close_safe'] > df['BB_Upper']).astype(float) -
                    (df['close_safe'] < df['BB_Lower']).astype(float)) * \
                   (1 + 0.55 * df['vcb_sent'].shift(1).fillna(0)) / \
                   (1 + df['bb_width'].fillna(0))

    # Alpha 3: Giá/SMA_20 ngược, điều chỉnh bởi Sentiment VCB, Z-score BB Width, Tốc độ thay đổi OBV
    df['alpha3'] = -1 * ((df['close_safe'] - df['SMA_20']) / df['SMA_20'].replace(0,1e-9)) * \
                   (1 + 0.65 * df['vcb_sent'].shift(1).fillna(0)) * \
                   (1 - z(df['bb_width'].fillna(0))) * \
                   ((df['OBV'] - df['OBV'].shift(1)).fillna(0) / df['volume'].replace(0,1e-9))

    # Alpha 4: Tương đối Sentiment/Momentum 3d/Biến động 20d, điều chỉnh bởi Correlation và Sign của Ret 3d
    df['alpha4'] = ( (df['vcb_sent'].shift(1).fillna(0) - df['tcb_sent'].shift(1).fillna(0)) * ((df['close_safe'] / df['close_safe_shift_3']) - 1).fillna(0) ) / \
                   (1 + np.abs(((df['close_safe'] / df['close_safe_shift_20']) - 1).fillna(0))).replace(0,1e-9) * \
                   (1 + 0.35 * df['vcb_sent'].rolling(7).corr(df['tcb_sent']).fillna(0)) * \
                   np.sign(df['ret_3d'].fillna(0))

    # Alpha 5: Z-score Momentum 5d/Std, MACD Hist, Momentum Sentiment 7d, điều chỉnh theo Vị thế so với SMA 20
    df['alpha5'] = np.tanh(
                       0.5 * z(((df['close_safe'] / df['close_safe_shift_5']) - 1).fillna(0) / (df['Std_20_safe'].replace(0,1e-9))) +
                       0.25 * z(df['MACD_Hist'].fillna(0)) +
                       0.25 * z(df['vcb_sent'].rolling(7).mean().diff().fillna(0))
                   ) * (1 + 0.25 * ((df['close_safe'] / df['SMA_20']) - 1).fillna(0))

    # --- Dọn dẹp và làm sạch ---
    cols_to_drop = [col for col in df.columns if col.startswith('close_safe') or col.endswith('_safe') or col.startswith('vol_') or col.endswith('_Hist') or col.startswith('bb_') or col.startswith('close_shift_')]
    cols_to_drop.extend(['volume_rolling_20_mean', 'ret_3d', 'Alpha4_Avg_Sent'])

    df = df.drop(cols_to_drop, axis=1, errors='ignore')

    df = df.replace([np.inf, -np.inf], 0)
    df = df.fillna(0)

    print(f"Đã tính toán xong 5 alpha features mới (Version 2) cho {TARGET_STOCK_CODE}.")
    return df


def add_temporal_features(df):
    """
    Tạo các tính năng liên quan đến thời gian.
    """
    df['day_of_week'] = df['date'].dt.dayofweek + 1
    df['day_of_month'] = df['date'].dt.day
    df['month_of_year'] = df['date'].dt.month
    return df

def prepare_data_with_sentiment(df):
    df = df.copy()

    df = calculate_alpha_features(df)
    df = add_temporal_features(df)

    # Target output là Pct_Change
    df['Pct_Change'] = df['close'].pct_change() * 100
    df['Pct_Change'] = df['Pct_Change'].fillna(0)

    initial_len = len(df)
    MIN_ROWS_TO_DROP = 30

    if initial_len > MIN_ROWS_TO_DROP:
        df_cleaned = df.iloc[MIN_ROWS_TO_DROP:].reset_index(drop=True)
        # Loại bỏ NaN nếu còn sót lại ở các cột quan trọng
        df_cleaned = df_cleaned.dropna(subset=['close', 'alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5'])
    else:
        df_cleaned = pd.DataFrame()

    rows_dropped = initial_len - len(df_cleaned)

    print(f"Sau khi làm sạch: {len(df_cleaned)} dòng dữ liệu (Đã loại bỏ {rows_dropped} dòng do thiếu dữ liệu lịch sử).")
    return df_cleaned


# =====================================
# PHẦN 3. CHUẨN BỊ DATA CHO CÁC LOẠI MODEL
# =====================================

def create_sliding_windows(data, window_size=5):
    X, y = [], []
    # Các features của Window [T-W+1, ..., T] dự đoán giá trị của T+1
    for i in range(len(data) - window_size):
        X.append(data[i:(i + window_size), :])
        y.append(data[i + window_size, -1]) # Lấy Pct_Change (cột cuối cùng)
    return np.array(X), np.array(y)

def create_ml_samples(df, features_cols, target_col):
    """
    Tạo mẫu cho các model ML truyền thống (SVR, RF).
    Features của ngày T-1 dự đoán Target của ngày T.
    """
    X_raw = df[features_cols].iloc[:-1].values
    y_raw = df[target_col].iloc[1:].values
    return X_raw, y_raw

class StockDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.FloatTensor(X)
        self.y = torch.FloatTensor(y)
    def __len__(self):
        return len(self.X)
    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

# =====================================
# PHẦN 4. ĐỊNH NGHĨA VÀ TRAIN CÁC MODEL
# =====================================

# --- Temporal Convolutional Network (TCN) ---
# TCN yêu cầu TensorFlow/Keras cho việc triển khai đơn giản
from tensorflow.keras.layers import Conv1D, Input, Activation, Dense, Dropout, Lambda, Add, Multiply
from tensorflow.keras.models import Model
from tensorflow.keras import backend as K

def TCN_Block(input_layer, filters, kernel_size, dilation_rate, dropout_rate):
    # Dilated Causal Convolution
    conv = Conv1D(filters, kernel_size, dilation_rate=dilation_rate, padding='causal', kernel_initializer='he_normal')(input_layer)
    conv = Activation('relu')(conv)
    conv = Dropout(dropout_rate)(conv)

    # Second Dilated Causal Convolution
    conv = Conv1D(filters, kernel_size, dilation_rate=dilation_rate, padding='causal', kernel_initializer='he_normal')(conv)
    conv = Activation('relu')(conv)
    conv = Dropout(dropout_rate)(conv)

    # Skip Connection (1x1 Conv nếu cần thay đổi chiều)
    if K.int_shape(input_layer)[-1] != filters:
        skip = Conv1D(filters, 1, padding='same')(input_layer)
    else:
        skip = input_layer

    # Add skip connection
    output_block = Add()([skip, conv])
    output_block = Activation('relu')(output_block)
    return output_block

def build_tcn_model(input_shape, filters, kernel_size, stack_size, dropout):
    inputs = Input(shape=input_shape)

    x = inputs

    for s in range(stack_size):
        # Tăng dilation rate theo lũy thừa của 2
        dilation_rate = 2 ** s
        x = TCN_Block(x, filters, kernel_size, dilation_rate, dropout)

    # Aggregate (lấy output cuối cùng của chuỗi)
    x = Lambda(lambda x: x[:, -1, :])(x)

    # Prediction head
    outputs = Dense(1, activation='linear')(x)

    model = Model(inputs=inputs, outputs=outputs)
    model.compile(optimizer=Adam(learning_rate=LEARNING_RATE), loss='mse')
    return model

# --- LSTM Model (Keras) ---
def build_lstm_model(input_shape):
    model = Sequential([
        KerasLSTM(64, activation='relu', input_shape=input_shape, return_sequences=True),
        Dropout(0.2),
        KerasLSTM(32, activation='relu'),
        Dropout(0.2),
        Dense(1, activation='linear')
    ])
    model.compile(optimizer=Adam(learning_rate=LEARNING_RATE), loss='mse')
    return model

# --- Evaluation Function ---
def evaluate_model(y_true, y_pred):
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mape = np.mean(np.abs((y_true - y_pred) / (y_true + 1e-6))) * 100
    return mse, rmse, mape

# =====================================
# PHẦN 5. MAIN EXECUTION
# =====================================

def main(file_path, symbol):
    print(f"=== SO SÁNH DỰ BÁO GIÁ CỔ PHIẾU {symbol} VỚI 4 MÔ HÌNH: LSTM, TCN, SVR, RF ===")

    # 1. Tải và chuẩn bị dữ liệu
    print("\n1. Đang tải và xử lý dữ liệu...")
    df = load_stock_data_from_df(file_path, symbol)
    if df.empty: return

    df_processed = prepare_data_with_sentiment(df)

    temporal_columns = ['day_of_week', 'day_of_month', 'month_of_year']
    alpha_columns = ['alpha1', 'alpha2', 'alpha3', 'alpha4', 'alpha5']
    features_columns = alpha_columns + temporal_columns
    target_output_column = 'Pct_Change'

    if len(df_processed) <= WINDOW_SIZE:
        print(f"LỖI: Dữ liệu sau khi làm sạch không đủ ({len(df_processed)} ngày).")
        return

    # --- Chia Train/Test ---
    TRAIN_END_DATE = pd.to_datetime('2024-12-31')

    df_train = df_processed[df_processed['date'] <= TRAIN_END_DATE].copy()
    df_test = df_processed[df_processed['date'] > TRAIN_END_DATE].copy()

    if df_train.empty or len(df_train) <= WINDOW_SIZE:
        print("LỖI: Dữ liệu training không đủ.")
        return
    if df_test.empty or len(df_test) <= WINDOW_SIZE:
        print("LỖI: Dữ liệu test không đủ.")
        return

    print(f"Số ngày dữ liệu Train (2024): {len(df_train)}")
    print(f"Số ngày dữ liệu Test (2025+): {len(df_test)}")

    # Thêm cột target vào dataframe gốc cho việc chuẩn hóa dễ dàng
    all_cols = features_columns + [target_output_column]

    # --- Chuẩn hóa (Fit trên Train set) ---
    scaler = StandardScaler()
    train_data_scaled = scaler.fit_transform(df_train[all_cols].values)
    test_data_scaled = scaler.transform(df_test[all_cols].values)

    # --- Kết quả so sánh ---
    comparison_results = []


    # =====================================
    # 2. LSTM & TCN (Sử dụng Sliding Window)
    # =====================================
    print("\n2. Huấn luyện LSTM và TCN...")

    # Tạo cửa sổ trượt cho DL models
    X_train_dl_scaled, y_train_dl_scaled = create_sliding_windows(train_data_scaled, WINDOW_SIZE)
    X_test_dl_scaled, y_test_dl_scaled = create_sliding_windows(test_data_scaled, WINDOW_SIZE)

    y_test_raw = df_test[target_output_column].values[WINDOW_SIZE:]
    y_test_close_prev = df_test['close'].values[WINDOW_SIZE-1:-1]

    # --- LSTM ---
    lstm_model = build_lstm_model(input_shape=(WINDOW_SIZE, X_train_dl_scaled.shape[2]))
    lstm_model.fit(X_train_dl_scaled, y_train_dl_scaled, epochs=EPOCHS, batch_size=BATCH_SIZE, verbose=0, callbacks=[EarlyStopping(monitor='loss', patience=5, restore_best_weights=True)])
    lstm_pred_scaled = lstm_model.predict(X_test_dl_scaled, verbose=0).flatten()

    # --- TCN ---
    tcn_model = build_tcn_model(input_shape=(WINDOW_SIZE, X_train_dl_scaled.shape[2]), filters=TCN_FILTERS, kernel_size=TCN_KERNEL_SIZE, stack_size=TCN_STACK_SIZE, dropout=0.2)
    tcn_model.fit(X_train_dl_scaled, y_train_dl_scaled, epochs=EPOCHS, batch_size=BATCH_SIZE, verbose=0, callbacks=[EarlyStopping(monitor='loss', patience=5, restore_best_weights=True)])
    tcn_pred_scaled = tcn_model.predict(X_test_dl_scaled, verbose=0).flatten()

    # --- Inverse Transform và Đánh giá DL Models ---

    # Helper function để Inverse Transform Pct_Change
    def inverse_transform_pct(scaled_pct, scaler, all_cols):
        # Tạo mảng tạm với Pct_Change ở cột cuối cùng và 0 ở các cột khác
        temp_array = np.zeros((len(scaled_pct), len(all_cols)))
        temp_array[:, -1] = scaled_pct
        return scaler.inverse_transform(temp_array)[:, -1]

    lstm_pred_raw = inverse_transform_pct(lstm_pred_scaled, scaler, all_cols)
    tcn_pred_raw = inverse_transform_pct(tcn_pred_scaled, scaler, all_cols)

    # Giá dự đoán ngày T = Giá T-1 * (1 + Pct_Change T / 100)
    lstm_pred_price = y_test_close_prev * (1 + lstm_pred_raw / 100)
    tcn_pred_price = y_test_close_prev * (1 + tcn_pred_raw / 100)

    # Lấy giá trị thực tế ngày T (Target)
    y_test_actual_price = df_test['close'].values[WINDOW_SIZE:]

    # Đánh giá và lưu kết quả
    comparison_results.append(("LSTM", *evaluate_model(y_test_actual_price, lstm_pred_price)))
    comparison_results.append(("TCN", *evaluate_model(y_test_actual_price, tcn_pred_price)))


    # =====================================
    # 3. SVR & Random Forest (Sử dụng 1 ngày T-1)
    # =====================================
    print("\n3. Huấn luyện SVR và Random Forest...")

    # Tạo mẫu cho ML models (X_T-1 -> Y_T)
    X_train_ml_raw, y_train_ml_raw = create_ml_samples(df_train, features_columns, target_output_column)
    X_test_ml_raw, y_test_ml_raw = create_ml_samples(df_test, features_columns, target_output_column)

    # Cần chuẩn hóa riêng cho ML models vì chúng không dùng window
    scaler_ml_X = StandardScaler()
    X_train_ml_scaled = scaler_ml_X.fit_transform(X_train_ml_raw)
    X_test_ml_scaled = scaler_ml_X.transform(X_test_ml_raw)

    scaler_ml_y = StandardScaler()
    y_train_ml_scaled = scaler_ml_y.fit_transform(y_train_ml_raw.reshape(-1, 1)).flatten()

    # --- SVR ---
    svr_model = SVR(kernel='rbf', C=100, gamma=0.1)
    svr_model.fit(X_train_ml_scaled, y_train_ml_scaled)
    svr_pred_scaled = svr_model.predict(X_test_ml_scaled)

    # --- Random Forest ---
    rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
    rf_model.fit(X_train_ml_scaled, y_train_ml_scaled)
    rf_pred_scaled = rf_model.predict(X_test_ml_scaled)

    # --- Inverse Transform và Đánh giá ML Models ---
    svr_pred_raw = scaler_ml_y.inverse_transform(svr_pred_scaled.reshape(-1, 1)).flatten()
    rf_pred_raw = scaler_ml_y.inverse_transform(rf_pred_scaled.reshape(-1, 1)).flatten()

    # Lấy giá đóng cửa T-1 và giá thực tế T cho ML models (đã được tạo ở create_ml_samples)
    y_test_close_prev_ml = df_test['close'].values[:-1]
    y_test_actual_price_ml = df_test['close'].values[1:]

    svr_pred_price = y_test_close_prev_ml * (1 + svr_pred_raw / 100)
    rf_pred_price = y_test_close_prev_ml * (1 + rf_pred_raw / 100)

    # Đánh giá và lưu kết quả
    comparison_results.append(("SVR", *evaluate_model(y_test_actual_price_ml, svr_pred_price)))
    comparison_results.append(("Random Forest", *evaluate_model(y_test_actual_price_ml, rf_pred_price)))


    # =====================================
    # 4. In Bảng So Sánh
    # =====================================
    print("\n" + "="*80)
    print(f"KẾT QUẢ SO SÁNH HIỆU SUẤT DỰ BÁO CỔ PHIẾU {symbol} (TẬP TEST 2025+)")
    print("="*80)

    headers = ["Model", "MSE", "RMSE", "MAPE (%)"]

    # Sắp xếp theo MAPE (tăng dần)
    comparison_results.sort(key=lambda x: x[3])

    # Làm tròn kết quả trước khi in
    rounded_results = [
        (model, f"{mse:.2f}", f"{rmse:.2f}", f"{mape:.2f}")
        for model, mse, rmse, mape in comparison_results
    ]

    print(tabulate(rounded_results, headers=headers, tablefmt="fancy_grid"))
    print("="*80)

if __name__ == '__main__':
    try:
        main(FILE_PATH, SYMBOL)
    except Exception as e:
        print(f"\nLỖI CHƯƠNG TRÌNH: Đã xảy ra lỗi không mong muốn trong quá trình thực thi: {e}")
        traceback.print_exc()
```
